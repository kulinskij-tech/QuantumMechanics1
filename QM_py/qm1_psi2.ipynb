Станом динамічної системи ми називаємо сукупність характеристик, яка достатня для опису еволюції і результатів вимірювань, при відомих зовнішніх умовах.

У квантовій механіці (точніше в [картині Шредінгера](https://colab.research.google.com/github/kulinskij-tech/QuantumMechanics1/blob/main/QM_py/qm1_scheiz_theor.ipynb#schrodrepr)), як постулат, приймається, що хвильова функція саме і є найбільш повним способом опису стану системи. Зокрема, квадрат модуля хвильової функції (інтенсивність хвилі) визначає густину ймовірності знаходження квантової системи в даному стані. Якщо стан системи можна характеризувати тільки координатою, то:
<!--  -->
\begin{equation}
\label{eq3/1} p\,(\mathbf{r},t) =
C\,|\,\psi(\mathbf{r},t)|^{\,2}\,.
\end{equation}
<!--  -->
Це означає, що модуль хвильової функції є експериментально спостережувальною величиною. Матеріальне існування частинки у просторі (в певний момент часу) визначається умовою нормування:
<!--  -->
\begin{equation}
\label{eq4/1} \int\limits_{V} p\,(\mathbf{r},t) \,dV =
1\,\,\,,\,\,\,\,C = \left( \int\limits_{V} |\,\psi
(\mathbf{r},t)|^{\,2} dV \right)^{ - 1}\,,
\end{equation}
<!--  -->
при виконанні умови:
<!--  -->
\begin{equation}\label{norm}
\int\limits_{V}
|\,\psi (\mathbf{r},t)\,|^{\,2} d\,V = 1\,.
\end{equation}
<!--  -->
Умова \eqref{norm} називається __умовою квадратичної інтегрованості__. Множина всіх квадратично інтегрованих функцій, тобто тих, які задовольняють \eqref{norm} і можуть бути нормовані, позначається $L_2$.

<!--  -->
Вони відображають дуальний характер опису квантової системи у термінах або координат або імпульсу. Введемо функцію
<!--  -->
\begin{equation}
\label{eq8/1} \varphi (\mathbf{p},t) = c(\mathbf{p},t)
\exp \left( - i\frac{E\,t}{\hbar} \right)\,,
\end{equation}
<!--  -->
за допомогою якої (\ref{eq6/1}) і (\ref{eq7/1}) можуть бути
записані у вигляді:
<!--  -->
\begin{equation}
\label{ph1} \psi(\mathbf{r},t)=\frac{1}{(2\pi\hbar)^{3/2}}
\int\limits_{\Lambda}\phi(\mathbf{p},t)
\exp{\left(i\frac{\mathbf{p}\cdot{\mathbf{r}}}{\hbar}\right)}d\Lambda\,.
\end{equation}
<!--  -->
Тут $\Lambda$ позначає об'єм імпульсного простору.
<!--  -->
\begin{equation}
\label{ph2}
\phi(\mathbf{p},t)=\frac{1}{(2\pi\hbar)^{3/2}}\int\limits_{V}\psi(\mathbf{r},t)
\exp{\left(-i\frac{\mathbf{p}\cdot{\mathbf{r}}}{\hbar}\right)}d\,V\,.
\end{equation}
<!--  -->
Ці співвідношення з математичної точки зору відображують той факт, що функції $\varphi $ і $\psi $ пов'язані перетворенням Фур'є за просторовими змінними. Відзначимо, що умова квадратичної інтегрованості хвильової функції забезпечує існування як прямого, так і зворотного перетворення Фур'є і приводить до виконання __рівності Планшереля__:
<!--  -->
\begin{equation} \label{eq10/1}
\int\limits_{V}|\psi(\mathbf{r},t)|^2
dV=\int\limits_{\Lambda}|\,\phi(\mathbf{p},t)|^2 d\Lambda\,.
\end{equation}
<!--  -->
Хвильова функція не обов'язково повинна бути функцією координат. Якщо замість фотоплівки, яка фіксує (з певною точністю) координати електронів, використовувати дифракційну гратку (сортування частинок за різними значенням імпульса), то, діючи електростатичним полем у певному напрямку, можна визначити імпульс електрона. Тож, стан квантової системи може бути описаний не з точки зору ймовірності просторового положення, а з точки зору ймовірності знаходження у стані з певним імпульсом (за допомогою дифракційної гратки). Такий дуальний характер опису стану у термінах пари "імпульс-координата" знаходить своє математичне відображення у співвідношеннях:
<!--  -->
\begin{equation}\label{eq6/1} 
\psi(\mathbf{r},t) = \int\limits_{\Lambda} c(\mathbf{p},t)\, 
\psi_{\mathbf{p}}(\mathbf{r},t)\,d\,\mathbf{p}
\end{equation}
<!--  -->
та
<!--  -->
\begin{equation}\label{eq7/1} 
c(\mathbf{p},t) = \int\limits_{V} \psi(\mathbf{r},t)\,\psi^*_{\mathbf{p}}(\mathbf{r},t)d\,V\,.
\end{equation}
<!--  -->
Вони відображають дуальний характер опису квантової системи у термінах або координат чи імпульсу.

Тому довільний стан (за умови \eqref{norm}) може бути представлений у виді суперпозиції хвиль де Бройля -- станів з певним значенням імпульсу. І навпаки, довільна суперпозиція хвиль де Бройля із скінченною сумарною інтенсивністю зображує певний стан квантової системи. Функції $c(\mathbf{p},t)$ (або $\varphi (\mathbf{p},t)$) і $\psi (\mathbf{r},t)$ дуальним чином можна інтерпретувати як хвильову функцію стану в імпульсному та координатному представленні відповідно.

На відміну від електромагнітних (ЕМ) хвиль у вакуумі або пружних виль в однорідному середовищі, характерною рисою хвиль де Бройля є нелінійний закон дисперсії. Використовуючи співвідношення:
<!--  -->
\begin{equation}
\label{eq11/1} E = \hbar\, \omega \,,\,\,E =
\frac{\mathbf{p}^{2}}{2\,m}\,\,\,,\,\,p = \hbar \,k\,,
\end{equation}
<!--  -->
які відображують дуалізм "хвиля-частинка", отримуємо:
<!--  -->
\begin{equation}\label{disp}
\omega = \frac{\hbar\,k^{\,2}}{2\,m}\,\,\,\,.
\end{equation}
<!--  -->


Таким чином, хвильовий пакет ("сигнал") із хвиль де~Бройля змінює свою форму з часом, оскільки хвилі з різними $|\mathbf{k}|$ мають різну швидкість. Співвідношення \eqref{disp} повністю визначають еволюцію станів -- хвильових пакетів за відсутності зовнішніх впливів (а також меж, що також може розглядатись як зовнішнє поле окремого виду). Дійсно, нехай у початковий момент часу $t=0$ є стан квантової системи, який описується хвильовою функцією $\psi(\mathbf{r},0)$. За даною початковою хвильовою функцією визначаємо величини $c(\mathbf{p})$. Підстановка цих даних у \eqref{eq6/1} враховуючи \eqref{eq11/1}
визначає $\psi(\mathbf{r},t)$ для довільного $t$:
<!--  -->
\begin{equation}
\label{eq13/1} 
\psi(\mathbf{r},t) = \int\limits_{V}
c(\mathbf{p})\exp \left( i\frac{\mathbf{p}\cdot\mathbf{r}}{\hbar} -
i\frac{\mathbf{p}^{\,2}}{2\,m\,\hbar}\,t
\right)
\frac{d\,\mathbf{p}}{(2\,\pi\, \hbar )^{n/2}}\,.
\end{equation}
<!--  -->
Відмітимо, що для вільного хвильового пакету, розподіл результатів вимірювань імпульсу від часу не залежить. Він дається функцією $c(\mathbf{p})$, яка визначається тільки початковими умовами.

Розглянемо одномірну групу хвиль -- стан, у формуванні якого беруть участь хвилі лише з невеликого інтервалу хвильових чисел $[k_{0} - \Delta k,k_{0} + \Delta k]$, мало відрізняючись за
амплітудою. Згідно з (\ref{eq13/1}) маємо:
<!--  -->
\begin{align}\label{grp}
\psi(x,t)=\frac{1}{(2\pi)^{1/2}}\int\limits_{k_{0}-\Delta
k}^{k_{0}+\Delta k} c(k)\exp\left(i\,k x-
i\,\frac{k^2}{2m}t\right)dk= &  \notag\\
\frac{1}{(2\pi)^{1/2}}c(k_{0})\exp\left(i\,k_{0} x-
i\,\frac{k_{0}^2}{2m}t\right)\int\limits_{-\Delta k}^{\Delta k}
\exp\left(i\,k x- i\,\frac{k}{m}t\right)dk= & \notag\\
\frac{1}{(2\pi)^{1/2}}2c(k_{0})\exp\left(i\,k_{0} x-
i\,\frac{k_{0}^2}{2m}t\right) \frac{\sin\left(x-\left.\frac{d\omega}{d
k}\right|_{k=k_{0}}t\right)}
{x-\left.\frac{d\omega}{dk}\right|_{k=k_{0}}t}&\notag\,.\\
\end{align}
<!--  -->
Отриманий результат показує, що вершина пакета рухається зі швидкістю
<!--  -->
$$
\left. v = \frac{d\omega}{dk}\right|_{k = k_{0}}\,,
$$
<!--  -->
яка називається \textbf{груповою швидкістю}. Дійсно, цей результат справедливий тільки для малих часів поки можна знехтувати іншими членами розкладу $\omega $ за хвильовим вектором.

Ймовірнісна інтерпретація квадрата модуля хвильової функції в координатному та імпульсному представленнях дозволяє обчислювати середні від функцій, які залежать або від координат, або від
імпульсів у будь-якому стані, якщо відповідні середні існують (наприклад для функцій з класу $\Im$):
<!--  -->
\begin{equation}
\label{eq15/1}
\begin{array}{l}
\langle F(\mathbf{r}) \rangle \, =
\,{\int\limits_{V} {F(\mathbf{r})\,
|\,\psi (\mathbf{r},t)|^{\,2}\,dV}} \, \\
\langle \Phi (\mathbf{p}) \rangle \, =
\,{\int\limits_{\Lambda} {\Phi (\mathbf{p})\,{\left|
{\,c(\mathbf{p},t)\,} \right|}^{\,2}\,d\Lambda}}\,.   \\
\end{array}
\end{equation}
<!--  -->
Більше того, використовуючи властивості перетворення Фур'є, можна записати наступні співвідношення:
<!--  -->
\begin{equation}
\label{eq16/1}
\begin{array}{l}
\langle x^{n} \rangle \, = \int\limits_{V}
\psi^*(\mathbf{r},t) \,x^{n}\,\psi
(\mathbf{r},t)\,dV = 
\int\limits_{\Lambda} 
c^*(\mathbf{p},t)\,\left(i\hbar \,\frac{\partial}{\partial \,p_{x}}\right)^n\,
c(\mathbf{p},t)\,d\Lambda \\
\langle p_{x}^{n} \rangle \, = \,\int\limits_{V}
c^* (\mathbf{p},t)\,p_{x}^{n}\,
c(\mathbf{p},t)\,d\Lambda = 
\int\limits_{V} \psi^*(\mathbf{r},t)\,\,\left(- i\hbar \,\frac{\partial}
{\partial \,x} \right)^{n}\,\psi (\mathbf{r},t)\,dV\,, \\
\end{array}
\end{equation}
<!--  -->
якщо функції та їх похідні достатньо швидко зникають на нескінченності. Цей результат демонструє те, що у квантовій механіці [фізичні величини, можуть представлятися операторами, які діють на вектори станів.](https://colab.research.google.com/github/kulinskij-tech/QuantumMechanics1/blob/main/QM_py/qm1_ermitop_theor.ipynb)

Розв'язки задач цього розділу використовують розклад стану, який задається нормованою хвильовою функцією, яка задовільняє умові нормування:
<!--  -->
\begin{equation}\label{eq:nrm}
\int\limits_{V}|\,\psi(\mathbf{r},\,t)|^2 dV = 1\,,
\end{equation}
<!--  -->
як функції координат, у інтеграл Фур'є:
<!--  -->
\begin{equation}\label{eq:psifourier}
\psi (\mathbf{r},\,t) = {\int\limits_{\Lambda}
{c(\mathbf{p},t)}}\, \psi_{\mathbf{p}}
(\mathbf{r},\,t)\,d\,\mathbf{p}
\end{equation}
<!--  -->
за хвилями де Бройля:
<!--  -->
\begin{equation}\label{eq:psipdevroigl}
\psi_{\mathbf{p}}({\rm r},\,t) =\frac{1}{(2\,\pi\,\hbar )^{d/2}}
 \exp\left(\frac{i\mathbf{p} \cdot \mathbf{r}-i\,E\,t}{\hbar}\right) =
\frac{1}{(2\pi \hbar)^{d/2}}\,\exp\left(i\,\mathbf{k}\cdot\mathbf{r} - i\,\omega(\mathbf{k}) \,t\right)\,,
\end{equation}
<!--  -->
які відповідають станам, вільної частинки яка має певний імпульс $\mathbf{p}$ у просторі із $n$ трансляційними ступенями вільності $\mathbb{R}^n$. Формула \eqref{eq:psifourier} виражає фундаментальний принцип суперпозиції станів заснований на експериментальних спостереженнях дифракційних та інтерференційних явищ при регістрації (суб)атомних частинок. Співвідношення \eqref{eq:nrm} визначає ймовірносний зміст величини $\rho(\mathbf{r},t)=|\psi(\mathbf{r},t)|^2$ як густини ймовірності регістрації частинки у точці $\mathbf{r}$ у момент часу $t$.

Зворотнє до \eqref{eq:psifourier} перетворення
***
\begin{equation}\label{eq:cfourier}
c(\mathbf{p},t) = {\int\limits_{V} {\psi (\mathbf{r},\,t)\,\psi^{*}_{\mathbf{p}}(\mathbf{r},\,t)\,d\,V}}
\end{equation}
***
визначає амплітуди плоських хвиль де Бройля, які формують
стан визначений хвильовою функцією $\psi(\mathbf{r},\,t)$.

Наявність дисперсії призводить до розпливання хвильових пакетів у координатному просторі, окільки хвилі із більшими $k$ мають більші фазові швидкості. Зворотнє до \eqref{eq:psifourier} перетворення
***
\begin{equation}\label{eq:psifourierinv} 
c(\mathbf{p},t) = {\int\limits_{V} {\psi
(\mathbf{r},\,t)\,\psi^*_{\mathbf{p}}(\mathbf{r},\,t)\,d\,V}}
\end{equation}
***
визначає амплітуди плоских хвиль де Бройля, які формують стан із хвильовою функциєю $\psi(\mathbf{r},\,t)$. Відповідно $|c(\mathbf{p},t)|^2$ має зміст густини ймовірності регістрації частинки із імпульсом $\mathbf{p}$.

Введемо функцію

***
\begin{equation}
\label{eq:cphi}
\phi(\mathbf{p},t)=c(\mathbf{p},t)\exp{\left(-\,i\,\frac{E\,t}{\hbar}\right)}
\end{equation}
***

тоді співвідношення \eqref{eq:psifourier} і \eqref{eq:psifourierinv} можуть бути записані у
вигляді:

***
\begin{align}
\psi(\mathbf{r},\,t)=&\frac{1}{(2\pi\,\hbar)^{3/2}}
\int\limits_{\Lambda}\phi(\mathbf{p},t)
\exp{\left(i\,\frac{\mathbf{p}\cdot{\mathbf{r}}}{\hbar}\right)}
d\Lambda \notag\\
\label{eq:psiphifourier}\\
\phi(\mathbf{p},t)=&\frac{1}{(2\pi\,\hbar)^{3/2}}\int\limits_{V}\psi(\mathbf{r},\,t)
\exp{\left(-\,i\,\frac{\mathbf{p}\cdot{\mathbf{r}}}{\hbar}\right)}d\,V \notag
\end{align}
***
Тут $\Lambda$ означає об'єм імпульсного простору. Отримані співвідношення \eqref{eq:psiphifourier} із математичної точки зору виражають  факт того що функції $\psi$ і $\phi$ пов'язані перетворенням Фур'є.
Відмітимо, що умова квадратичної інтегрованості забезпечує існування як прямого так і зворотнього перетворень при чому виконується __рівність Планшереля__:
***
\begin{equation}\label{eq:plnsh} 
||\psi||^2=\int\limits_{V}|\,\psi(\mathbf{r},\,t)|^2\,
dV=\int\limits_{\Lambda}|\,\phi(\mathbf{p},t)|^2\, d\Lambda = ||\phi||^2\,.
\end{equation}
***
тобто нормування не залежить від представлення. Таким чином, простір станів квантової системы з точки зору опису імпульсних чи просторових ступенів вільності є простором квадратично інтегрованих функцій $L_{2}$. Цей простір є нескінченовимірним лінійним нормованим простором. Більш того, він також є характерним прикладом гильбертова (комплексний аналог евклідова) простору, оскільки норма \eqref{eq:plnsh} породжується скалярним добутком
***
\begin{equation}\label{eq:l2sc} 
\langle \psi_{1}\vert \psi_{2}\rangle =\int\limits_{V}\psi_{1}^{*}\psi_{2}\,dV = 
\int\limits_{\Lambda}\phi_{1}^{*}(k)\phi_{2}(k)\,d\Lambda \,.
\end{equation}
***
У випадку дискретного Фур'є-перетворення (якщо накладаються певні граничні умови) маємо знайомий вигляд суми за компонентами "векторів":
$$
||\psi||^2=\sum\limits_{n=1}^{\infty} |c_{n}|^2\,,\\
\langle \psi_{1}\vert \psi_{2}\rangle = \sum\limits_{n=1}^{\infty} c^{(1)^*}_{n}c^{(2)}_{n}
$$
Важливими характеристиками просторового розподілу є середнє $ \langle\, x\,\rangle$
та середнє квадратичне відхилення:
***
$$\langle \Delta x^2 \rangle = \langle\,
x^2\,\rangle - \langle\, x\,\rangle^2 = \left\langle\, x-
\left\langle\, x \,\right\rangle\,
 \right\rangle^2\,.
$$
***
яке визначаеє дисперсію $\sigma_{x} = \sqrt{\langle \Delta x^2 \rangle}$ (у ймовірносному сенсі) розподілу вимірювань.

Зворотність перетворення Фур'є на множині функцій яка визначена умовою \eqref{eq:nrm} ($L_{2}$-простір), означає еквівалентність імпульсного і координатного представлень. Більш того, можна показати (див. [Задачу](https://colab.research.google.com/github/kulinskij-tech/QuantumMechanics1/blob/main/QM_py/qm1_wavepack_zad.ipynb#zad_heiswvpack)), що функція та її Фур'є-зображення не можуть бути одночасно бути сконцентровані у обмежених границях своїх аргументів: чим
більш сконцентрована функція $\psi(\mathbf{r},\,t)$ у $x$-просторі, тим більш розмито у імпульсному просторі її Фур'є-зображення $\phi(\mathbf{p},t)$ (див. [Задачу](https://colab.research.google.com/github/kulinskij-tech/QuantumMechanics1/blob/main/QM_py/qm1_wavepack_zadevolut.ipynb#zad_gaussevol)). Ця властівіть Фур'є перетворення призводить до співвідношення невизначеності між проекцією імпульса та координатою, яке вперше було запропоновано В. Хайзенбергом:
***
\begin{equation}\label{hei}
\Delta k_{i}\, \Delta x_{i}  \ge \frac{1}{2}\quad \Leftrightarrow \quad 
\Delta p_{i}\, \Delta x_{i}  \ge \frac{\hbar}{2}
\end{equation}
***
де
***
\begin{align*}
 \Delta k_{i} =& \sqrt{\int_{\Lambda} k_{i}^2
|\phi(\mathbf{k},t)|^2\,d\mathbf{k}- \left(\int_{\Lambda} k_{i}
|\phi(\mathbf{k},t)|^2\,d\mathbf{k}\right)^2}\\
  \Delta x_{i} =& \sqrt{\int_{V} x_{i}^2
|\psi(\mathbf{r},t)|^2\,d\mathbf{r}- \left(\int_{V} x_{i}
|\psi(\mathbf{r},t)|^2\,d\mathbf{r}\right)^2}\,.
\end{align*}

Якщо густина розподілу швидко зникає, наприклад швидше ніж довільна ступінь аргумента, то існують
__моменти__ розподілу:

***
\begin{equation}\label{eq:moments}
   \langle x^n\rangle \, =
 \,{\int\limits_{V} x^n\,
 |\,\psi (x,t)|^{\,2}\,dx}\,.
\end{equation}
***
Загальне правило для обчислення середніх значень від функцій, які залежать або від координат, або від імпульсів базується на відомих властивостях зображення Фур'є функції та її похідної.

Еволюція стану вільного хвильового пакета $\psi(\mathbf{r},\,t)$ визначається початковим станом
$\psi_{0}(\mathbf{r})=\psi(\mathbf{r},0)$. Для того, що б отримати $\psi(\mathbf{r},\,t)$ необхідно:
- визначити амплітуди хвиль де Бройля, які формують початковий стан:
***
\begin{equation}\label{eq:ck} 
c(\mathbf{k}) = \frac{1}{(2\pi)^{\frac{n}{2}}} \int\limits_{V}
\psi_{0} (\mathbf{r})\,\exp\left(-\,i\,\mathbf{k}\cdot
\mathbf{r}\right)\,d^{n}\,\mathbf{r}
\end{equation}
***
- обчислити функцію $\psi(\mathbf{r},\,t)$ згідно \eqref{eq:psifourier} використовуючи амплітуди $c(\mathbf{k})$.

Слід зауважити, що __для вільного хвильового пакета амплітуди хвиль де Бройля__ $c(\mathbf{p})$ __можна вважати незалежними від часу__,  оскільки за відсутності зовнішних сил, як і у класичній механиці, зміна імпульсних (динамічних) характеристик системи не можлива. Це є квантовий аналог класичного закона інерції.

### Рівняння Шредінгера вільної частинки та густина потоку ймовірності
Виходячи з закону еволюції \eqref{eq13/1} вільних хвильових пакетів легко дістаємо рівняння Шредінгера для цього випадка:

$$i\,\hbar \frac{\partial\psi }{\partial t} = \int\limits_{\Lambda}
\frac{\mathbf{p}^2}{2\,m}\,c(\mathbf{p})\, 
\psi_{\mathbf{p}} (\mathbf{r},t)\,d\,\mathbf{p} = -\frac{\hbar^2}{2\,m}\,\Delta \psi$$

З іншого боку, факт береження повної ймовірності і те що її локальна зміна відбуається за рахунок потоку частинок крізь стінки, дозволяє отримати диференціальну форму закона збереження аналогічну до відомих аналогів з механіки суцільних середовищ та ввести густину струму ймовірності:

\begin{equation}
\frac{d}{d \,t}\int\limits_{V}\rho\, dV = 
\int\left(\frac{\partial \psi^*}{\partial t}\,\psi +
\psi^*\frac{\partial \psi}{\partial t}
\right)\,dV = \frac{\hbar}{i\,2\,m}\int \left(\psi\,\Delta \psi^* -
\psi^* \,\Delta \psi\right)dV=\\
-\frac{\hbar}{m}\int  {\rm div}\,\left(\frac{\psi^*\,\nabla \psi -
\psi \,\nabla \psi^*}{2\,i}\right)dV = - \int \mathbf{j}\cdot \mathbf{n}\,dS\\
\mathbf{j} = \frac{\hbar}{m}{\rm Im}\, \psi^*\nabla \psi
\end{equation}
