__Задача__
Знайдемо закон перетворення хвильової функції у координатному і імпульсному представленнях при переході з однієї інерціальної системи відліку в іншу.<br>
__Розв'язок__ Зв'язок координат між інерціальними системами відліку $K$ і $K'$ в нерелятивістській теорії дається
перетворенням Галілея:
$$
\mathbf{r}_{K'} = \mathbf{r}_{K} - \mathbf{v}_{KK'} t\,,
$$
де $\mathbf{v}_{K\,K'}$  - швидкість руху $K '$ відносно $K$. <a id='momenrggalil'>Імпульс та енергія перетворюється згідно з правилом складання швидкостей</a>:
$$
\mathbf{p}_{K'} = \mathbf{p}_{K} -  m\,\mathbf{v}_{KK'}\,,\quad 
E = E' + \mathbf{p}_{K'}\cdot \mathbf{v}_{KK'} + \frac{m\,\mathbf{v}^2_{KK'}}{2}
$$
Стан квантовой системи є її внутрішньою властивістю, яка не залежить від системи відліку. Тому проекційний оператор 
$$
\hat{P}_{\psi} = \vert \psi \rangle
\, \langle \psi \vert
$$
є інваріантом тому вектор стану $\langle \psi \vert$ при зміні системи відліку може змінити тільки фазу. Хвильова функція у координатному представленні є проекцією на власні стани оператору координати:
$$
\psi(\mathbf{r},t) = \langle \mathbf{r},t \vert \psi \rangle
$$
Розкладемо її за хвилями де Бройля
$$
\langle \mathbf{r},t \vert \psi \rangle = \langle \mathbf{r}'+\mathbf{v}_{KK'} t,t \vert \psi \rangle = 
\int \langle \mathbf{r}'+\mathbf{v}_{KK'} t,t \vert\mathbf{p}\rangle\langle\mathbf{p}\vert \psi 
\rangle\,d\mathbf{p}
$$
Зкориставшись координатним представленням хвиль де Бройля:
$$
\langle \mathbf{r},t \vert \mathbf{p}\rangle = 
\exp\left(i\,\frac{\mathbf{p}\cdot\mathbf{r} - E\,t}{\hbar}\right)
$$
та [правилами трансформації](#momenrggalil) отримаємо:
$$
\exp\left(i\,\frac{\mathbf{p}\cdot\mathbf{r} - E\,t}{\hbar}\right) = 
\exp\left(i\,\frac{\left(\mathbf{p}'+m\mathbf{v}_{KK'}\right)\cdot\left(\mathbf{r}'+\mathbf{v}_{KK'} t\right) -\left(E' + \mathbf{p}'\cdot \mathbf{v}_{KK'} + \frac{m\,\mathbf{v}^2_{KK'}}{2}\right)\,t}{\hbar}\right) = \\
\exp\left(i\,\frac{\mathbf{p}'\cdot\mathbf{r}'-E' t}{\hbar}\right)\,\exp\left(\frac{m\mathbf{v}_{KK'}\cdot\mathbf{r}'+t\,\frac{m\mathbf{v}^2_{KK'}}{2}}{\hbar}\right)
$$
Тому
$$
\langle \mathbf{p}'\vert\psi\rangle = \Psi_{K'}(\mathbf{p'},t) = \Psi_{K}(\mathbf{p}-m\,\mathbf{v}_{KK'},t)\,,\quad 
\Psi_{K}(\mathbf{r},t) = \Psi_{K'}(\mathbf{r}',t)\,\exp\left(\frac{m\mathbf{v}_{KK'}\cdot\mathbf{r}'+t\,\frac{m\mathbf{v}^2_{KK'}}{2}}{\hbar}\right)
$$

__Задача__ У просторі векторів станів визначимо сукупність перетворень які залежать від векторного параметра $\mathbf{v}$: 
$$
\left\langle\,\mathbf{r}\left|\,\hat{U}_{\mathbf{v}}\,\right|\,\psi\,\right\rangle
 = \exp\left(\frac{m\,
\mathbf{r}\cdot
\mathbf{v}}{i\,\hbar}\right)\,\psi(\mathbf{r})\,.$$ 
Нехай
$\hat{\mathbf{P}}$ - оператор імпульса системи. Знайдіть
$$ 
\hat{U}_{\mathbf{v}}\,\hat{\mathbf{P}}\,\hat{U}_{ \mathbf{v} }^{-1}\,.
$$
<br>
__Розв'язок__ Очевидно, що матричні елементи оператора у координатному представленні є:
$$
\left\langle\,\mathbf{r}\left|\,\hat{U}_{\mathbf{v}}\,\right|\,\mathbf{r}'\,\right\rangle
= \exp\left(\frac{m\, \mathbf{r}'\cdot \mathbf{v}}{i\,\hbar}\right)\,
\delta(\mathbf{r} - \mathbf{r}')\,.
$$
Зрозуміло, що оператор $\hat{U}_{\mathbf{v}}$ - унітарний
$$
\left\langle\,\mathbf{r}\left|\,\hat{U}_{\mathbf{v}}^{\dagger}
\hat{U}_{\mathbf{v}}\,\right|\,\mathbf{r}'\,\right\rangle =
\left\langle\,\mathbf{r}\left|\,\hat{U}_{\mathbf{v}}
\hat{U}^{\dagger}_{\mathbf{v}}\,\right|\,\mathbf{r}'\,\right\rangle =
\delta(\mathbf{r}'-\mathbf{r})\,.
$$
Крім того,
$$\left\langle\,\mathbf{p}_1\,\vert\,
\hat{U}_{\mathbf{v}} \vert\,\mathbf{p}_2\right\rangle = \int
d\mathbf{r}_1\,d\mathbf{r}_2 \left\langle\,\mathbf{p}_1\,\vert\,
\mathbf{r}_1\right\rangle\left\langle\,\mathbf{r}_1\,\vert\,
\hat{U}_{\mathbf{v}} \vert\,\mathbf{r}_2\right\rangle
\left\langle\,\mathbf{r}_2 \vert\,\mathbf{p}_2\right\rangle =
\delta(\mathbf{p}_2 - m\mathbf{v} - \mathbf{p}_1)\,.$$
$$\left\langle\,\mathbf{p}_1\,\vert\,
\hat{U}^{\dagger}_{\mathbf{v}} \vert\,\mathbf{p}_2\right\rangle =
\delta(\mathbf{p}_1 - m\mathbf{v} - \mathbf{p}_2)\,.$$
тому
$$\left\langle\,\mathbf{p}\,\vert\,
\hat{U}_{\mathbf{v}} \hat{\mathbf{P}} \hat{U}_{ \mathbf{v}
}^{-1}\vert\,\mathbf{p}'\right\rangle = \int
d\mathbf{p}_1\,d\mathbf{p}_2 \left\langle\,\mathbf{p}\,\vert\,
\hat{U}_{\mathbf{v}}
\vert\,\mathbf{p}_1\right\rangle\left\langle\,\mathbf{p}_1\,\vert\,
\hat{\mathbf{P}} \vert\,\mathbf{p}_2\right\rangle
\left\langle\,\mathbf{p}_2\,\vert\, \hat{U}^{-1}_{\mathbf{v}}
\vert\,\mathbf{p}'\right\rangle = $$
$$\int
d\mathbf{p}_1\left\langle\,\mathbf{p}\,\vert\,
\hat{U}_{\mathbf{v}} \vert\,\mathbf{p}_1\right\rangle \mathbf{p}_1
\left\langle\,\mathbf{p}_1\,\vert\, \hat{U}^{\dagger}_{\mathbf{v}}
\vert\,\mathbf{p}'\right\rangle = \int
d\mathbf{p}_1\delta(\mathbf{p}_1 - m\mathbf{v} - \mathbf{p})
\mathbf{p}_1 \delta(\mathbf{p}_1 - m\mathbf{v} - \mathbf{p}') =(\mathbf{p}' + m\mathbf{v})\delta(\mathbf{p} - \mathbf{p}')\,.$$
Таким чином
$$
\hat{U}_{\mathbf{v}}\left|\,\psi\,\right\rangle 
$$ 
описує миттєвий процес приведения системи у стан рівномірного руху зі швидкістю $\mathbf{v}$ (перехід у миттєву супроводжуючу інерціальну систему відліку). Зауважимо, що в силу унитарности оператора $\hat{U}_{\mathbf{v}}$, квадрати модулей волнових функций
$$\left\langle\,\mathbf{r}\,\vert\,\psi\right\rangle $$ та
$$\left\langle\,\mathbf{r}\left|\,\hat{U}_{\mathbf{v}}\,\right|\,\psi\,\right\rangle$$
у точці $\mathbf{r}$ однакові.

__Задача__ Доведіть, що якщо $\mathbf{V}$ - оператор
величини векторного типу, наприклад, імпульсу $\mathbf{P}$ або зміщення
$\mathbf{R}$, то:
$$
[L_m,V_n] = i\,\hbar\,\varepsilon_{m\,n\,k} V_{k}\,.
$$
__Розв'язок__ Нехай $\hat{U}_{R(\mathbf{n},\theta)}$ - унітарне перетворення векторів стана, яке відповідає повороту $R(\mathbf{n},\theta)$ системи у реальному просторі. Тоді спостережувальна величина, яка зображується оператором $\hat{A}$, при новому положені системи, буде мати вигляд:
\begin{equation}\label{au}
\tilde{\hat{A}} = \hat{U}_{R(\mathbf{n},\theta)} \hat{A}
\hat{U}^{-1}_{R(\mathbf{n},\theta)}\,.
\end{equation}
Запишемо оператор $\hat{U}$ у  інфінітезимальній формі:
$$\hat{U}_{R(\mathbf{n},\theta)} = 1 +
\frac{\mathbf{n}\cdot \hat{\mathbf{L}}}{i\,\hbar}\theta
+o(\theta)\,,\quad \hat{U}^{-1}_{R(\mathbf{n},\theta)}=\hat{U}_{R^{-1}(\mathbf{n},\theta)} = 1 -
\frac{\mathbf{n}\cdot \hat{\mathbf{L}}}{i\,\hbar}\theta +o(\theta).
$$ 
тоді інфінітезимальна зміна такого оператора:
\begin{equation}\label{i1}
  d \hat{A} =\frac{\theta}{i\,\hbar} [\mathbf{n}\cdot \hat{\mathbf{L}},\hat{A}]
\end{equation}
З іншого боку для оператора __векторного__ типу:
$$\mathbf{V} = V_{i} \mathbf{e}_i\,,$$
його інфінітезимальна зміна при обертанні системи визначається зміною напрямка базисних векторів:
\begin{equation}\label{i2}
  d\mathbf{V} = d\left(R^{-1}(\mathbf{n},\theta) \mathbf{V}\right) =
- \theta\, \mathbf{n}\times \mathbf{V}
\end{equation}
Прирівнюючи \eqref{i1} і \eqref{i2}, отримаємо:
$$
-  \mathbf{n}\times \mathbf{V} = \frac{1}{i\,\hbar} [\mathbf{n}\cdot
\hat{\mathbf{L}},\hat{\mathbf{V}}]
$$
Переходячи до декартових компонент, маємо результат:
$$
-  \varepsilon_{ijk}\, V_k = \frac{1}{i\,\hbar} [L_j,V_i]\,
\Rightarrow \, [L_i,V_j] = i\hbar \varepsilon_{ijk} V_k\,.
$$

__Задача__ Побудуйте вирази для операторів $L_{i}$ у сферичній системі координат.<br>
__Рішення__ Будемо використовувати означення інфінітезимального генератора і запишемо зміну кутів $\theta, \varphi$ при (зворотньому) повороті точки навколо вісі $x$:
$$
x' = x\,,\quad y' = y + \delta\,\alpha\,z\,,\quad z' = z-\delta\,\alpha\,y
$$
Використовуючи стандартні формули переходу:
$$
x = r\,\sin\theta\,\cos\varphi\,,\quad y = r\,\sin\theta\,\sin\varphi\,,\quad z =r\,\cos \theta
$$
і порівнюючи малі першого порядка знайдемо що зміна кутів $\theta\,,\varphi$ при такому повороті дорівнює:
$$
\delta\,\theta = \delta\alpha\,\sin{\varphi}\,,\quad \delta\,\varphi =   \delta\alpha\,\cos{\varphi}\,{\rm ctg}\,\theta
$$
тому, в сферичних координатах зміна хвильової функції:
$$\psi(r,\theta',\varphi') = U_{x} \psi(r,\theta,\varphi) =
\left(\,1+\frac{\delta\alpha}{i\,\hbar}\,L_{x}+\ldots\,\right)\,
\psi(r,\theta,\varphi)
$$
Таким чином:
$$
L_{x} = i\,\hbar
\left(\,\sin{\varphi}\,
\frac{\partial\,}{\partial\, \theta} + \cos{\varphi}\,{\rm ctg}\,\theta\,\frac{\partial\,}{\partial\, \varphi}\,\right)
$$
Аналогічно, для $L_{y}$:
\begin{eqnarray}
   x' &=& x - \delta\,\alpha\,z = r\,\sin \theta\,\cos \varphi - \delta\,\alpha\,r\,\cos \theta \\
  y' &=& y \\
  z' &=& z+\delta\,\alpha\,x = r\,\cos \theta + \delta\,\alpha\,r\,\sin \theta\,\cos \varphi
\end{eqnarray}
$$\delta\,\theta = -\delta\alpha\,\cos{\varphi}\,,\quad \delta\,\varphi =   \delta\alpha\,\sin{\varphi}\,{\rm ctg}\,\theta
$$
і
$$
\psi(r,\theta',\varphi') = U_{y} \psi(r,\theta,\varphi) =
\left(\,1+\frac{\delta\alpha}{i\,\hbar}\,L_{y}+\ldots\,\right)\,
\psi(r,\theta,\varphi)\,.
$$
Результат:
\begin{equation}\label{ly}
  L_{y} = i\,\hbar\,\left(-\cos \varphi\, \frac{\partial\,}{\partial\, \theta}
  + \sin{\varphi}\,{\rm ctg}\,\theta \,\frac{\partial\,}{\partial\, \varphi} \right)
\end{equation}
Сходинкові оператори:
$$
L_{\pm} = L_{x}\pm i\,L_{y}= \hbar\,e^{\pm\,i\,\varphi}\left(\pm\frac{\partial\,}{\partial\, \theta} + i\,{\rm ctg}\,\theta \,\frac{\partial\,}{\partial\, \varphi}\right)
$$
Очевидно, стан $l=0,m=0$ представляється функцією $Y_{0,0} = \frac{1}{\sqrt{4\,\pi}}$. Стан $l=1,m=0$:
$$
Y_{1,0} = \sqrt{\frac{3}{4\,\pi}}\,\cos \theta\sim z
$$
Функції інших станів отримаємо застосовуючи оператори $\hat{L}_{\pm}$:
$$
\sqrt{2}\,Y_{1,\pm 1} =  L_{\pm} \,Y_{1,0} \to Y_{1,\pm 1} = \mp \sqrt{\frac{3}{8\,\pi}}\,\sin \theta\,e^{\pm\,i\,\varphi}\sim x\pm i\,y
$$
Цей результат можна отримати також розглядаючи неприводиму систему функцій $x,y,z$ з т.з. оператора момента імпульса. Зрозуміло що лінійні комбінації $x \pm i\,y = r\,\sin{\theta}\,e^{\pm i\,\varphi}$ є власними для $\hat{L}_{z}$ с $m=\pm 1$, $z=r\,\cos \theta$  - $m=0$. Т.ч. система неприводимих функцій порядка $l=1$ має вигляд:
$$
x\pm i\,y = e^{\pm\,i\,\varphi}\sin{\theta}\,,\quad  z = \cos{\theta}
$$
Друга гармоніка будується на основі ортогональності до перших двох:
$$
Y_{2,0} = \sqrt{\frac{5}{16\,\pi}}\,\left(3\cos^2\theta -1\right)
$$
В результаті отримаємо:
$$
\sqrt{6}\,Y_{2,\pm 1} =  L_{\pm} \,Y_{2,0} \to Y_{2,\pm 1} =  \mp \sqrt{\frac{15}{8\,\pi}}\,\sin \theta\,\cos \theta\,e^{\pm\,i\,\varphi}\,,
$$
і
$$
Y_{2,\pm 2} = \frac{1}{4}
   \sqrt{\frac{15}{2 \pi }}\, e^{\pm 2 i \phi }\, \sin^2\theta\,.
$$
Ці 5 функцій є комбінаціями поліномів другого порядка: $(x\pm i\,y)^2,z^2,z\,(x\pm i\,y)$. Не важко здогадатися, що ці ж комбінації виникають при обчисленні компонент тензора квадрупольного моменту:
$$
Q_{ij} = 3\,n_{i}\,n_{j} - \delta_{i\,j}\,,\quad \mathbf{n} = \frac{\mathbf{r}}{r}
$$
В термінах змінних $z, w=x+i\,y$ оператори кутового момента є:
$$
L_{+} = z\,\partial_{w^*}+\frac{w}{2}\,\partial_{z}\,,\quad
L_{-} = z\,\partial_{w}+\frac{w^*}{2}\,\partial_{z}\,,\quad
L_{z} = w\,\partial_{w} - w^*\,\partial_{w^*}
$$
де
$$
\partial_{w} = \frac{1}{2}\left(\frac{\partial}{\partial x} -
i\, \frac{\partial}{\partial y}\right)
$$
