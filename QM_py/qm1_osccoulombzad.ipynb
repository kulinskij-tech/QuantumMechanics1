__Задача__ За допомогою методу сходинкових операторів,
обчислите спектр стац. станів частинки на піввісі із гамільтоніаном:
$$\hat H = \hat{p}^2 - \frac{\lambda}{\hat{x}}\,, \quad x>0$$ <br>
__Рішення__ По суті даний гамільтоніан представляє радіальну частину гамільтоніана кулонівської заачи для станів з $l=0$ (відсутність обертального внеску). Запишемо пеший крок ітерації:
\begin{align*}
  \hat{a}_{0} =& \hat{p} + i\,\alpha_{0}\,f_{0}(x) \\
  \hat{H} = & \left(\hat{p}-i\,\alpha_{0}\,f_{0}(x)\right)\,\left(\hat{p}+i\,\alpha_{0}\,f_{0}(x)\right)+E_{0}
\Rightarrow \alpha^2_{0}\,f^2_{0}(x)+\alpha_0\,f'_{0}(x)+E_0+\lambda/x = 0
\end{align*}
За характером рівняння видно що функцію $f_{0}(x)$ доцільно шукати у вигляді:
$$f_{0}(x) = 1+a_0/x$$
отже
\begin{align*}
   \alpha^2_0+ E_0= &\, 0 \Rightarrow E_0 = -\lambda^2/4\\
  2\,\alpha^2_0\,a_0+\lambda = &\, 0 \Rightarrow a_0 = -2/\lambda\\
  a^2\,\alpha^2_0 - a_0\,\alpha_0 =&\, 0 \Rightarrow \alpha_0 = 1/a_0
\end{align*}
Наступний крок:
$$\left(\hat{p}+i\,\alpha_{0}\,f_{0}(x)\right)\,\left(\hat{p}-i\,\alpha_{0}\,f_{0}(x)\right)+E_{0} = \hat{p}^2 + \alpha^2_{0}\,f^2_{0}(x)-\alpha_0\,f'_{0}(x)+E_0=
\hat{p}^2+\alpha^2_1\,f^2_{1}(x) +\alpha_1\,f'_{1}(x) + E_{1}\,.
$$
c $f_{1}(x) = 1+a_1/x$\,.

\begin{align*}
  a_1\alpha^2_1 =&\, -\gamma/2 \Rightarrow \alpha_1 = -\lambda/4\\
  a^2_1\alpha^2_1 - a_1\alpha_1 =& \,2 \Rightarrow a_1\alpha_1 = 2 \\
  E_1=&\,-\alpha^2_1 \Rightarrow E_1 = - (\lambda/4)^2 = E_0/4
  \end{align*}
  
откуда видно, что реккурентная последовательность операторов $\hat{a}_{n}$ задается равенствами:
$$  
a_{n+1}\alpha^2_{n+1} =\, a_{n}\alpha^2_{n}\\
  a^2_{n+1}\alpha^2_{n+1} - a_{n+1}\alpha_{n+1} = \, a^2_{n}\alpha^2_{n}+ a_{n}\alpha_{n}
  \Rightarrow  a_{n+1}\alpha_{n+1} = 1 + a_{n}\alpha_{n}
  \\
  E_{n+1}+\alpha^2_{n+1}=\, E_{n}+\alpha^2_{n} = 0
$$  
Рішення цих рекурентних співвідношень дает спектр:
$$\alpha_{n+1}a_{n+1} = n+1\,,\quad \alpha_{n+1} = \frac{1}{n+1}\,,\quad \Rightarrow E_{n} = -\frac{E_0}{(n+1)^2}$$
який співпадає з відомим результатом після переходу до стандартних позначень. Враховуючи те що 3-вимірне радіальне рівняння
Шредінгера еквівалентно о 1-мірного із заміною $\psi \to r\,\psi$:
$$-\frac{1}{r}\left(r\,\psi\right)''+
\frac{l(l+1)}{r^2}\psi+\frac{2\,m}{\hbar^2}\,(V(r)-E)\psi = 0$$
та виродження енергии за квантовим числом $l$, знайдений спектр 1-мірної задачі дає відповідь і у 3-вимірному випадку.

__Задача__ Покажемо, що спектр частинки у потенціальной ямі:
$V(x) = V_0\left(\frac{a}{x} - \frac{x}{a}\right)^2$ є еквідистантним,
тобто, фактично, співпадає із спектром гармонічного
осцилятора.<br>
__Рішення__  Для спрощення будемо вважати, що $V_0 = 1,\, a=1$. Згідно з процедурою введемо оператор
$\hat{a}_0 = \hat{p} +i\,\alpha_0\, f_0(x)\,,$ і отримаєм
рівняння для визначення $f_0$:

\begin{equation}\label{f0}
  \alpha_0^2\,f^2_0(x) + \alpha_0\,f^{'}(x) + E_0 =\left(\frac{1}{x} - x\right)^2\,.
\end{equation}

Очевидно, що функцію $f_0$ можна шукати у вигляді:

$$f_0(x) = \frac{\beta_0}{x}-x\,.$$

Підставляючи до \eqref{f0} отримаєм:

\begin{align}
  E_0 - 2\,\alpha_0\,\beta_0 - \alpha_0 + 2 =& 0 \notag\\
  \alpha_0^2 -1 = & 0\\
  \alpha_0^2\,\beta_0^2 - \alpha_0\,\beta_0 - 1 = & 0 \notag
\end{align}

Корінь $\alpha_0 = - 1$ призводить до $E_0 < 0$, що не можливо. Для $\alpha_0 = 1$ маємо:
$$
\beta_0 = \frac{1+\sqrt{5}}{2}\,, \quad E_0 = 2\,\beta_0 - 1 =
\sqrt{5}\,.
$$
Далі визначимо:
$$
\hat{a}_1 = \frac{\beta_1}{x} - x
$$
і використовуючи вираз \eqref{e2} знаходимо:
$$
\left(\frac{\beta_1}{x}-x\right)^2-\frac{\beta_1^2}{x^2}+E_1 =
\left(\frac{\beta_0}{x}-x\right)^2 + \frac{\beta_0}{x^2} + E_0\,.
$$
Звідки :
$$
\beta_1 = 1+\beta_0\,,\quad E_1 = E_0 + 2\,.
$$

Діючи за індукцією, неважко переконатися що має місце рекурентне співвідношення $E_n = E_{n-1} + 2$. Різниця зі спектром гармонічного осцилятора не суттєва, вона пов'язана із обраною параметризацією енергії.
