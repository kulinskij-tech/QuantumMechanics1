__Задача__
Довести, що нормований вектор $ \left|\,\psi\,\right\rangle$
є власним для оператора фізичної величини $\hat{A}$ тільки якщо:
\begin{equation}\label{eq:sobst}
  \left\langle\,\psi\, \left\vert \hat{A}^2 \right\vert\, \psi\,\right\rangle =
\left\langle\,\psi\,\left \vert \hat{A}\right \vert\, \psi\,\right\rangle^2
\end{equation}
Дайте фізичну інтерпретацію цьому твердженню.
<br> 
__Розв'язок__ Знайдемо норму вектора:
$$
  (\hat{A}-\lambda)
\left|\,\psi\,\right\rangle\,,
$$

где $\lambda$ - дійсне число:
$$
\forall\, \lambda \in \mathbb{R}\,,\quad  0 \le
\braket{(\hat{A}-\lambda)\psi}{(\hat{A}-\lambda)\psi}
=\av{\psi}{\hat{A}^2}{\psi} - 2\,\lambda \av{\psi}{\hat{A}}{\psi} +
\lambda^2\,.
$$
Остання рівність виконується завдяки ермітовості оператора. Умова невід'ємності:
$$
 \av{\psi}{\hat{A}}{\psi}^2 - \av{\psi}{\hat{A}^2}{\psi}\le 0
$$
дає питоме твердження. З фізичної точки зору рівність \eqref{eq:sobst} означає нульову дисперсію результатів вимірювання для фізичної величини у стані $\vert \psi\rangle $, що і виділяє стан як власний.

__Задача__ Знайдіть матричні елементи операторів координати
$\hat{\mathbf{r}}$ та імпульса $\hat{\mathbf{p}}$ в координатному та
імпульсному представленнях.<br> 
__Розв'язок__ За означенням :
$$
\psi(\mathbf{r})=
\left\langle \mathbf{r}\, \vert\,\psi\,\right\rangle\,.
$$
Розглянемо наприклад $\hat{X}$ и $\hat{P}_{x}$, тоді
\begin{equation}\label{x1}
\langle\, x\,\vert \hat{X} \vert\,\psi\,\rangle = \int \langle\,
x\,\vert \hat{X} \vert\,x'\,\rangle \langle\,
x'\,\vert\,\psi\,\rangle\, dx'= \int \langle\, x\,\vert \hat{X}
\vert\,x'\,\rangle \psi(x')\, dx' = x\, \psi(x)\,.
\end{equation}
тому використовуючи означення $\delta$-функціоналу приходимо до висновку:
\begin{equation}\label{x11}
\langle\, x\,\vert \hat{X} \vert\,x'\,\rangle = x'\delta(x'-x)\,.
\end{equation}
Зрозуміло, що те ж саме із відповідними переозначеннями
виконеється і для оператора імпульса його власному представлені. Визначимо
теперь вид оператора імпульса у координатному представлені ( у
базисі власних узагальнених станів оператора $\hat X$). Діючи аналогічно до 
\eqref{x1} запишемо:
$$
\langle\, x\,\vert \hat{P}_{x} \vert\,\psi\,\rangle = \int
\langle\, x\,\vert \hat{P}_{x} \vert\,x'\,\rangle \langle\,
x'\,\vert\,\psi\,\rangle\, dx'= \int \langle\, x\,\vert
\hat{P}_{x} \vert\,x'\,\rangle \psi(x')\, dx' = - i\,\hbar
\frac{\partial\,\psi}{\partial\, x}\,.
$$
що згідно з означенням похідної від $\delta$-функціоналу призводить до:
$$
\langle\, x\,\vert \hat{P}_{x} \vert\,x'\,\rangle = i\,\hbar\,
\frac{\partial\, }{\partial\, x'}\,\delta(x'-x)\,.
$$
Для оператора координати в імпульсному представлені:
$$
\langle\, p\,\vert \hat{X}\vert\,\psi\,\rangle = \int \langle\,
p\,\vert \hat{X} \vert\,p\,'\,\rangle \langle\,
p\,'\,\vert\,\psi\,\rangle\, dp\,'= \int \langle\, p\,\vert
\hat{X} \vert\,p\,'\,\rangle \tilde{\psi}(p\,')\, dp\,' = i\,\hbar
\frac{\partial\, \tilde{\psi}}{\partial\, p}\,.
$$
$$
\langle\, p\,\vert \hat{X}\vert\,p\,'\,\rangle = -\, i\,\hbar\,
\frac{\partial\, }{\partial\, p\,'}\,\delta(p\,'-p)\,,
$$
що узгоджується з \eqref{x11}:
$$
\langle\, x\,\vert \hat{X} \vert\,x'\,\rangle = \int
\langle\, x \vert\,p\,\rangle\langle\, p\,\vert \hat{X}
\vert\,p'\,\rangle\langle\, p'\vert\,x'\,\rangle\,d p\,d p' 
$$

__Задача__
Визначіть спектр оператора "імпульса" для частинки на
відрізку $x\in[0,a]$.
<br> 
__Розв'язок__ Формальний вираз для оператора проекції імпульса ($\hbar=1$) $\hat{p}_{x} = -i\frac{d}{d\,x}$ слід зробити ермітовим:

$$\label{ermp}
\braket{\psi_{1}}{\hat{p}_{x}\psi_{2}} = \braket{\hat{p}_{x}\psi_{1}}{\psi_{2}}=
\int\limits_{0}^{a}\,\psi^*_{1}\left(-\,i\,\frac{d\,\psi_{2}}{d\,x}\right)\,dx\,.
$$

Інтегруючи за частками отримуємо:

$$\label{ermp1}
\int\limits_{0}^{a}\,\psi^*_{1}\left(-\,i\,\frac{d\,\psi_{2}}{d\,x}\right)\,dx
=\left.-\,i\,\psi^*_{1}\psi_{2}\right|_{0}^{a}+
\int\limits_{0}^{a}\,\psi_{2}\left(-\,i\,\frac{d\,\psi_{1}}{d\,x}\right)^*\,dx
$$

Видається, що природня умова на граничні значеня хвильової функції $\psi(0)=\psi(a)=0$ не є задовільною, оскільки спряжений оператор визначений на більш широкій множині функцій, які не обовязково прямують до 0 на границях проміжка. Якщо припустимо $\psi_1=\psi_2$ (см. [Задачу](#hermitavrg), хоча оператор не обмежений), то вимога ермітовості виглядатиме як:

$$
\vert\psi(0)\vert = \vert\psi(a)\vert
$$

що накладає обмеження на граничні значеня хвильової функції:

$$
 \psi(a) = \exp(i\,\alpha)\,\psi(0)\,.
$$ 

Фазовий множник граничної умови може бути обрано вільно, він і визначатиме імпульс. Цей параметр характеризує т.з. самосопряжені розсширеня формального операторного виразу 

$$
\hat{p}_{x} = -i\frac{d}{d\,x}
$$  

(див. доповнення до Лекцій). В результаті власні функції і спектр оператора (квази)імпульса є:

$$-\,i\,\frac{d\,\psi}{d\,x} = p\,\psi(x)\,,\quad \psi(a) =
\exp(i\,\alpha)\,\psi(0)\,,\quad \alpha\in[0,2\,\pi)
\\
\psi(x) = C\,\exp\left(i\,p_{n;\alpha}\,x\right)\,,\quad
p_{n;\alpha} = \frac{\alpha+2\,\pi\,n}{a}
$$
