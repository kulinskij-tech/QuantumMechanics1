Задачі цього разділу спрямовані на отримання аналітичних розв'язків стаціонарного рівняння Шредінгера:

$$-\frac{\hbar^{2}}{2m}\frac{d^2}{d\,x^2}\psi + V(x)\psi = E\,\psi $$

розв'язкі задачі - спектр оператора энергії та його власні стани - визначються граничними умовами. В точках, де потенціал має розриви мають накладатись граничні умови виходячи із закона збереження повної ймовірності (неперервність потоку ймовірності). У випадку розривів I роду достатньо (але не обов'язково) вимагати неперервності хвильової функції та її похідної. Для одновимірних задач стаціонарне hівняння Шредінгера є лінійним ОДР 2-го порядку, тому важливу роль відіграє <a href='https://uk.wikipedia.org/wiki/Визначник_Вронського'> визначник Вронського (вронскіан)</a> $W$ для пари розв'язків $\psi_1$ и $\psi_2$:
$$W[\psi_1,\psi_2] = \begin{vmatrix}
  \psi_1 & \psi_2 \\
  \psi_1' & \psi_2'
\end{vmatrix} = \psi_1\,\psi_2' - \psi_1'\,\psi_2\,
$$
рівняння Шредінгера. Якщо потенціал має розриви тільки 1-го роду, то $W[\psi_1,\psi_2]$ є постійною для розв'язків які відповідають одному і тому ж власному значеню $E$. Дійсно:
\begin{equation}\label{wrs}
\frac{d}{d\,x}W[\psi_1,\psi_2] = \psi_1\,\psi_2'' - \psi_1''\,\psi_2
= \frac{2m}{\hbar^2}\,(E-U(x))\psi_1\,\psi_2 -
\frac{2m}{\hbar^2}(E-U(x))\psi_1\,\psi_2 =0
\end{equation}
Для одномірних задач про зв'язані стани у потенціальних ямах мають місце наступні факти.

- хвильова функція основного стану може прямувати в 0 тільки на границях області (в координатному представлені). 
- хвильові функціи збуджених станів мають вузли.
- стани (т.з. зв'язані) дискретного спектру ($\psi(\pm \infty)\to 0$) є невирожденними.
- стани (т.з. розсіяння) неперервного спектру є 2-кратно вирожденними (за напрямом).

У випадку обмеженого потенціалу стани с $E<0$ відповідають дискретному спектру, а стани $E>0$ - неперервному. Стани дискретного спектру - це стани локалізовані, яким відповідають функції з простору $L_2$:
$$
\int|\psi|^2 dV< \infty
$$
Cтани неперервного спектра називаються станами розсіяння, оскільки ймовірність спостереження частинки на $\infty$. Процес пружнього розсіяння може бути представлений як (унітарне) перетворення станів приходящих частинок в стани уходящих. Розсіяння хвиль характеризується амплітудою і зсувом фаз по відношенню до падаючого пучка. Зобразимо  стан частинок що приходять хвилями де Бройля:
\begin{equation}\label{sc1}
  \psi_{in} = \left\{\begin{array}{lr}
    e^{i\,k\,x}  & \text{частинки з $-\infty$} \\
    e^{-i\,k\,x} & \text{частинки з $+\infty$}
  \end{array}
\right.
\end{equation}
Ми положили амплітуди таких хвиль одиничними, що відповідає одиничній густині частинок у потоці. Після взаємодії
з потенціалом кожна з таких хвиль породжає віддзеркалену та прошедшу хвилі. В результаті маємо відповідність:
\begin{eqnarray}
  \underbrace{\psi^{(+)}_{I} = e^{ikx} +
  B e^{-ikx}}_{\text{падаюча + віддзеркалена}}\rightarrow& \underbrace{\psi^{(+)}_{II} = A
  e^{ikx}}_{\text{розсіяна вперед}} \notag\\
\underbrace{ \psi^{(-)}_{I} = D e^{-ikx}}_{\text{розсіяна вперед}}
\leftarrow\,\, & \underbrace{\psi^{(-)}_{II} = e^{-ikx}+ C
e^{ikx}}_{\text{падаюча + віддзеркалена}} \notag\,.
\end{eqnarray}
Між амплітудами можна встановити співвідношення, які витікають з основної властивості вронскіана \eqref{wrs}:

\begin{eqnarray}
  \left. W\left[\psi^{(+)}_{I},\overline{\psi^{(+)}_{I}}\right]\right|_{x\to-\infty} =
    \left. W\left[\psi^{(+)}_{II},\overline{\psi^{(+)}_{II}}\right]\right|_{x\to+\infty}
    \,: & |A|^2+|B|^2=1 \notag\\
&\notag\\
  \left. W\left[\psi^{(-)}_{I},\overline{\psi^{(-)}_{I}}\right]\right|_{x\to-\infty} =
    \left. W\left[\psi^{(-)}_{II},\overline{\psi^{(-)}_{II}}\right]\right|_{x\to+\infty}
    \,: & |C|^2+|D|^2=1\notag\\
    & \label{s4}\\
      \left. W\left[\psi^{(+)}_{I},\psi^{(-)}_{I}\right]\right|_{x\to-\infty} =
    \left. W\left[\psi^{(+)}_{II},\psi^{(-)}_{II}\right]\right|_{x\to+\infty}
    \,:  & A=D\,\notag\\
   &\notag\\
  \left. W\left[\psi^{(+)}_{I},\overline{\psi^{(-)}_{I}}\right]\right|_{x\to-\infty} =
    \left. W\left[\psi^{(+)}_{II},\overline{\psi^{(-)}_{II}}\right]\right|_{x\to+\infty}
    \,:  & \overline{A}\,B+A\,\overline{C}=0 \notag\\ \notag
\end{eqnarray}

```python
import sympy 
from sympy import *
# 
# psi1,psi2= symbols('psi1 psi2',cls=Function)
x,k = symbols('x k',real=True)
a,b,c,d = symbols('a b c d')
# 
psi1=[exp(I*k*x)+b*exp(-I*k*x),a*exp(I*k*x)]
psi2=[d*exp(-I*k*x),exp(-I*k*x)+c*exp(I*k*x)]
#
```

Обчислимо вронскіани за допомогою Sympy:

```python
simplify((wronskian([psi1[0],conjugate(psi1[0])],x)-wronskian([psi1[1],conjugate(psi1[1])],x))/(2*I*k))
```

```python
simplify((wronskian([psi2[0],conjugate(psi2[0])],x)-wronskian([psi2[1],conjugate(psi2[1])],x))/(2*I*k))
```

```python
simplify((wronskian([psi1[0],psi2[0]],x)-wronskian([psi1[1],psi2[1]],x))/(2*I*k))
```

```python
simplify((wronskian([psi1[0],conjugate(psi2[0])],x)-wronskian([psi1[1],conjugate(psi2[1])],x))/(2*I*k))
```

Відношення густин струмів відбитої і падаючої хвиль визначає коефіцієнт проходження (відбиття) бар'єру:

\begin{equation}\label{coeff}
  T = \frac{|j_{\text{прош}}|}{|j_{\text{пад}}|}\,,\quad
  R = \frac{|j_{\text{отраж}}|}{|j_{\text{пад}}|}
\end{equation}

Закон збереження числа частинок:

$$R+T = 1\,.$$

Бачимо, що перші два співвідношення \eqref{s4} відповідають цієї рівності. Третє співвідношення означає, що амплітуди прохождения вправо и вліво співпадають, що означає інваріантність системи відносно зміни напрямка часу. Останнє можна поєднати з попереднім так що матриця:
\begin{equation}\label{scmat}
 \hat{S} = \begin{pmatrix}
    A & C \\
    B & D
  \end{pmatrix}
\end{equation}
є унітарною (див. \eqref{s4}). Фізичний зміст такої матриці можна зясувати, якщо розглядати її в базисі падаючих частинок:
\begin{eqnarray}
e^{ikx}\rightarrow \begin{pmatrix}
    1 \\
    0
  \end{pmatrix}\,, & \,\,\, \hat{S}\,\begin{pmatrix}
    1 \\
    0
  \end{pmatrix} =\begin{pmatrix}
    A \\
    B
  \end{pmatrix}\,,\notag\\
  &\\
e^{-ikx}\rightarrow \begin{pmatrix}
    0 \\
    1
  \end{pmatrix}\,,&\,\,\,\,  \hat{S}\,\begin{pmatrix}
    0 \\
    1
  \end{pmatrix} =\begin{pmatrix}
    C \\
    D
  \end{pmatrix}
  \notag\,,
\end{eqnarray}
тобто оператор $ \hat{S}$ перетворює хилі що приходять з $\infty$ на хвилі що уходять до $\infty$ після взаємодії. Матрица \eqref{scmat} називается __S-матрицею розсіяння__.<br>
