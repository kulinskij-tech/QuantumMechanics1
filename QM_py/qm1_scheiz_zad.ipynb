__Задача__
Квантова система має базисні стани
$$
\left|\,0\,\right\rangle\,,\quad \left|\,1\,\right\rangle
$$ 
в яких фізична величина $\hat{A}$ приймає значення 0,1 відповідно. Оператор Шредінгера задано наступним чином:
\begin{equation*}
  \hat H \left|\,0\,\right\rangle = \alpha \left|\,0\,\right\rangle +
  \beta \left|\,1\,\right\rangle\,,\quad \hat H \left|\,1\,\right\rangle =
  \beta \left|\,0\,\right\rangle + \alpha
  \left|\,1\,\right\rangle\,,
\end{equation*}
де $\alpha,\,\beta \in R$ - не залежать від часу. Знайдіть еволюцію початкового стана:
$$
\left|\,\psi_{0}\,\right\rangle = c_{0}\,\left|\,0\,\right\rangle+c_{1}\,\left|\,1\,\right\rangle
$$
та еволюцію середнього значення величини $\hat{A}$.
<br>
__Розв'язок__ Матриця оператора Шредінгера:
$$
\hat{H} = \begin{pmatrix}
  \alpha & \beta \\
  \beta & \alpha
\end{pmatrix}
$$
Власні вектори і значення обчислюються стандартним образом:

```python
from sympy import *
a,b =symbols('a b')
ham=Matrix([[a,b],[b,a]]);
ham.eigenvals()
```

```python
ham.eigenvects()[0][2][0]
```

```python
ham.eigenvects()[1][2][0]
```

хоча воні очевидні  завдяки простій структурі матриці:
\begin{align*}
E_{1} = \alpha +\beta\,,\quad& \left|\,\psi_{1}\,\right\rangle =
\frac{1}{\sqrt{2}}\,( \left|\,0\,\right\rangle +
\left|\,1\,\right\rangle )
 \\
E_{2} = \alpha -\beta\,,\quad& \left|\,\psi_{2}\,\right\rangle =
\frac{1}{\sqrt{2}}\,( \left|\,0\,\right\rangle -
\left|\,1\,\right\rangle )
\end{align*}
Еволюція вектора стана визначається його розкладом за власними станами гамильтоніана:
\begin{multline*}
\left|\,\psi(t)\,\right\rangle = \exp\left(-i\,\frac{E_{1}}{\hbar}\,t\right)
\left\langle\,\psi_{1}\,\vert\,\psi_{0}\right\rangle
\left|\,\psi_{1}\,\right\rangle +\exp\left(-i\,\frac{E_{2}}{\hbar}\,t\right)
\left\langle\,\psi_{2}\,\vert\,\psi_{0}\right\rangle
\left|\,\psi_{2}\,\right\rangle = \\
\exp\left(-i\,\frac{E_{1}}{\hbar}\,t\right)\frac{c_0+c_1}{\sqrt{2}}
\left|\,\psi_{1}\,\right\rangle +\exp\left(-i\,\frac{E_{2}}{\hbar}\,t\right)
\frac{c_0-c_1}{\sqrt{2}}
\left|\,\psi_{2}\,\right\rangle
\end{multline*}

__Задача__
Знайдіть часову еволюцію дисперсій координати та імпульсу для вільного хвильового пакета та поясніть їх асимптотичну поведінку
при $t\,\to 0+$ и $t\,\to \infty$.<br>
__Розв'язок__ Для вільної частинки, імпульс є інтегралом руху оскільки
гамільтоніан представлений суто кінетичною енергією:
$$
\hat H = \frac{\hat{p}_{x}^2}{2\,m}\,.
$$
Отже для такої системи, середнє значння довільної функції імпульсу не змінюється з часом, тому:
\begin{equation}
  \frac{d}{d t}
\left\langle \hat{p}_{x}^2 \right\rangle= \left\langle
\frac{1}{i\,\hbar}[\hat{p}_{x}^2,\hat H] \right\rangle=0\,.
\end{equation}
Для оператора координати запишемо:
$$
[\hat{x},\hat H] =
\frac{1}{2\,m}[\hat{x},\hat{p}^2]=\frac{i\,\hbar}{m}\hat{p}
$$
і
\begin{equation}\label{u0}
  \frac{d}{d t}
\left\langle\, \hat{x}^2 \right\rangle= \left\langle
\frac{1}{i\,\hbar}[\,\hat{x}^2,\hat H] \right\rangle=\frac{1}{m}
\left\langle\, \hat{x}
\hat{p}_x+\hat{p}_x\hat{x}\,\right\rangle\,= \frac{2}{m}
\left\langle\, \hat{x} \hat{p}_x\,\right\rangle - \frac{i\hbar}{m} .
\end{equation}
Знайдемо ''прискорення'' для дисперсії:
\begin{equation}
  \frac{d^2}{d t^2}
\left\langle\, \hat{x}^2 \right\rangle= -\left\langle
\frac{1}{\hbar^2}[[\,\hat{x}^2,\hat H],\hat H]
\right\rangle=\frac{1}{i\,\hbar\, m} \left\langle\, [\hat{x}
\hat{p}_x+\hat{p}_x\hat{x},\hat H]\,\right\rangle = \frac{2}{m^2}
\left\langle p_{x}^2 \right\rangle\,.
\end{equation}
Це свідчить про те що середнє квадратичне відхилення
координати зростає з часом за законом:
$$
\Delta x^2(t) =
\Delta x^2(0)+u_0\,t+\frac{t^2}{m^2} \left\langle p_{x}^2
\right\rangle\,,
$$
де $u_0$ визначається початковим значенням величини \eqref{u0}. Т.ч. дисперсія координати асимптотично веде себе
як:
\begin{align}
  \sigma(t)&=\sqrt{\Delta x^2(t)} = \sigma(0)\left(1+A\,t+B\,t^2+
  o(t^2)\right)\,, \quad t\,\to\,0\\
  \sigma(t)& =  \frac{t}{m} \sqrt{\left\langle p_{x}^2
\right\rangle}\,, \quad t\,\to\,\infty \,,
\end{align}
Це демонструє загальний характер  поведінки дисперсії для вільних хвильових пакетів, [результата який було отримано раніше для конкретного пакета.](https://colab.research.google.com/github/kulinskij-tech/QuantumMechanics1/blob/main/QM_py/qm1_wavepack_zadevolut.ipynb#gaussdispasympt)

Реалізація символьних обчислень наведена нижче ($m=1$)

```python
from sympy.physics.quantum.constants import hbar
# 
psi = symbols('psi',cls=Function)
x,y,z = symbols('x y z',real=True) 
pos = [x,y,z] # coordinates
#
def coordf(c,f): # coordinate operator
    p= c*f
    return p
# 
def imp(f):  # momentum operator
    p=[]
    for x in pos:
        p.append(-I*hbar*diff(f,x))
    return p

def ham(f): # free hamiltonian
    h=-hbar**2*(diff(f,x,2)+diff(f,y,2)+diff(f,z,2))/2
    return h
# 
```

Обчислюємо комутатор $\frac{1}{i\,\hbar}[\hat{x},\hat{H}]$

```python
xham=simplify(coordf(x**2,ham(psi(x,y,z)))-ham(coordf(x**2,psi(x,y,z))))/(I*hbar)
xham
```

Обчислюємо комутатор $\frac{1}{(i\,\hbar)^2}[[\hat{x},\hat{H}],\hat{H}]$

```python
simplify(simplify((xham.subs(psi(x,y,z),ham(psi(x,y,z))))-ham(xham))/(I*hbar))
```

__Задача__ Знайдіть зв'язок між матричними елементами операторів координати
і імпульса для системи "частинка у потенціальному полі", якщо в
якості базиса обрана множина власних векторів оператора
Шредінгера (енергетичне представлення).
<br>
__Розв'язок__ Система задається гамільтоніаном:

$$\hat{H} = \frac{\hat{\mathbf{p}}^2}{2\,m}+\hat{V}\,.$$

Звязок між операторами координати і імпульса визначається рівнянням Гейзенберга:

$$\frac{d}{d\,t}\hat{\mathbf{r}}=\frac{\hat{\mathbf{p}}}{m} =
\frac{1}{i\,\hbar}[\hat{\mathbf{r}},\hat{H}]\,.$$

таким чином отримаємо:

$$
\langle\psi_{E}\vert\hat{\mathbf{p}}\vert\psi_{E'}\rangle
=\frac{m}{i\,\hbar}\langle\psi_{E}\vert[\hat{\mathbf{r}},\hat{H}]\vert\psi_{E'}\rangle
=\frac{m}{i\,\hbar}\langle\psi_{E}\vert\left(\hat{\mathbf{r}}\,\hat{H}-
\hat{H}\,\hat{\mathbf{r}}\right)\vert\psi_{E'}\rangle=
\frac{m\,(E'-E)}{i\,\hbar}\av{\psi_{E}}{\hat{\mathbf{r}}}{\psi_{E'}}\,.
$$

__Розв'язок__ З рівняння Шредінгера безпосередньо отримуємо:

$$

=\av{\varphi_{E}}{\hat{H}}{\psi} = E\,\braket{\varphi_{E}}{\psi}\\
% 
\frac{\partial }{\partial\,t}\left( \braket{\varphi_{E}}{\psi}\vphantom{\frac{1}{2}}\right)= 0
$$
