## Спін.
<!--  -->
$\newcommand{\ket}[1]{\left.\left\vert\, #1\,\right. \right \rangle}$
$\newcommand{\bra}[1]{\left. \langle \, #1\, \right\vert}$
$\newcommand{\braket}[2]{\langle \,#1 \vert #2 \, \rangle}$

Спін є додатковою внутрішньою характеристикою частинки яка проявляється через наявність власного магнітного моменту, наприклад в <a href='https://en.wikipedia.org/wiki/Stern–Gerlach_experiment'>досліді Штерна-Герлаха</a>. Експеримент вказує, що магнітний момент пов'язаний із спіном існує навіть якщо частинка нейтральна (наприклад <a href='https://en.wikipedia.org/wiki/Neutron_magnetic_moment'>нейтрон</a>), що свідчить про її складну (кваркову) внутрішню структуру. Спін є прикладом внутрішньої ступені вільності аналогічної до заряду чи маси покою. Прикладом інших внутрішніх квантових чисел, якими характеризується стан частинок є ізотопічний, баріонний, лептоний заряди, а також дивність та чарівність. Вони проявляються при аналізі процесів зіткнення та взаємоперетворення частинок. Приклад спіна надає розуміня загальнгого механізму введення такого типу квантових чисел та звязок із властивостями симетрії. Послідовне введення поняття спіну засноване на релятивістській інваріантості основних фізичних законів. Однак, завдяки тісному зв'язку цього поняття із властивостями інваріантості системи відносно поворотів, спін також природнім чином виникає у нерелятивістській квантовій механіці. Нагадаємо, що оператор орбітального моменту імпульсу є інфінітезимальним генератором унітарного перетворення хвильової функції при повороті. При цьому стани з різними проекціями моменту імпульсу на виділену вісь грали роль базисних векторів у просторі станів з даним значенням квадрата моменту імпульсу. З математичної точки зору хвильова функція може розглядатися як $2l + 1$ - мірний вектор в якості координат якого виступають коефіцієнти розкладу за вказаною базисною системою станів. При розгляді дії групи поворотів був використаний той факт, що поворот на кут $2\,\pi $ співпадає з тотожним. Завдяки цій умові проекція кутового моменту на вісь є цілою кратною $\hbar $. Тому величина $2l + 1$ і опинилась непарним числом. Однак існує можливість задати дію (строго кажучи, незводиме лінійне представлення) групи поворотів і в просторі розмірності кратній 2. Розглянемо простіший випадок частинки із спіном $s = \frac{1}{2}$. Це означає, що існує два базисних стани, які характеризуються протилежними значеннями проекції спіну на виділену вісь, скажімо $z$. Оскільки спінова ступінь свободи не залежить від положення частинки у просторі (виключаючи релятивістські ефекти), то будемо розглядати тільки залежність хвильової функції від спінових ступенів свободи. Таким чином, простір спінових ступенів свободи для частинки із спіном $s = \frac{1}{2}$ являє собою 2-вимірний (комплексний) гільбертовий простір. В загальному випадку розмірність $2s+1$.
<br>
<!--  -->
З точки зору обертальної інваріантності зміна проекції спіна пов'язана при повороті системи (чи системи координат) як цілого навколо центру мас. Операція
повороту реалізується у вигляді унітарного оператора, який не діє на трансляційні (зовнішні) ступені вільності, на відміну від  оператора обертального момента, який явним чином діє на координатні чи імпульсні ступені вільності:
<!--  -->
\begin{equation}
\label{trspin} \left\langle\,\mathbf{r}\,\vert\,\psi\right\rangle
=\left\langle\,\mathbf{r}\left|\,
\hat{\mathcal{U}}_{\mathcal{R}(\mathbf{n},\alpha)}\, \right|\,\psi\,\right\rangle=
\left\langle\,\mathbf{r}\left|\,\tilde{\psi}\,\right.\right\rangle\,,
\end{equation}
<!--  -->
де вісь обертання проходить через точку $\mathbf{r}$. Густина ймовірності не змінюється при повороті вимірувача в точці $\mathbf{r}$. На відміну від оператора обертального момента, спін описує зміну хвильової функції (вектора стану) при повороті системи яке не пов'язане із зміною радіус-вектору $\mathbf{r}$. Це означає, що вектор $\ket{\psi}$
повинен характеризувати простір внутрішніх ступенів вільності,
яке у випадку спіна є скінченовимірним гільбертовим простором. В ньому реалізується представлення группи обертань
унітарними перетвореннями. Таке представлення визначається розмірністю $2s+1$, де $s$ є спін системи. Тобто хвильова
функція має структуру вектора в просторі внутрішніх (спінових) ступенів вільності:
<!--  -->
$$
\left\langle\,\mathbf{r}\,\vert\,\Psi\right\rangle
 =\begin{pmatrix} \psi_1(\mathbf{r})
   \\
  \ldots \\
\psi_{2s+1}(\mathbf{r})
\end{pmatrix}
$$
<!--  -->
$2s+1$-мерного вектора. <a id='moms'>Унітарний оператор, який представляє обертання $\mathcal{R}(\mathbf{n},\alpha)$ у спіновому просторі</a> має вигляд аналогічний [до того що ми отримали раніше для кутового момента](https://colab.research.google.com/github/kulinskij-tech/QuantumMechanics1/blob/main/QM_py/qm1_intsymom_theor.ipynb#unitrot):
<!--  -->
\begin{equation}\label{moms} 
\hat{\mathcal{U}}_{\mathcal{R}(\mathbf{n},\alpha)}=
\exp\left( \alpha\,\frac{\mathbf{n}\cdot\hat{\mathbf{S}}}{i\,\hbar}
\right)
\end{equation}
<!--  -->
У якості інфінітезимального генератору виступає оператор спіна, точніше його проекція на вісь повороту $\mathbf{n}\cdot\hat{\mathbf{S}}$. Це ермітов оператор, і у певному базисі, він представляється у вигляді ермітовій матриці $(2s+1)\times (2s+1)$ з нульовим слідом. У випадку полуцілого спіна, обертанням $\mathcal{R}(\mathbf{n},\alpha)$ і $\mathcal{R}(\mathbf{n},\alpha+2\pi)$, які тотожні з геометричної точки зору, відповідають різні оператори 
$$
\hat{\mathcal{U}}_{\mathcal{R}(\mathbf{n},\alpha)} = -
\hat{\mathcal{U}}_{\mathcal{R}(\mathbf{n},\alpha+2\pi)}
$$
що відповідає зміні фази стана при обертанні на непарну кількість повних обертань $ = 1\,({\rm mod}\, 2\,\pi)$. <br>
Комутаційні співвідношення для проекції спіна і правило суми
спінового момента складеної системи співпадають з аналогічними для обертального момента:
$$
[\hat{S}_{i},\hat{S}_{j}] = i\,\hbar\,\varepsilon_{i\,j\,k}\,\hat{S}_{k}
$$
Зазвичай простор спінових станів будується за допомогою сходинкових операторів:
<!--  -->
$$
\hat{S}_{\pm} = \hat{S}_{x}\pm i\,\hat{S}_{y}\\
\hat{S}_{\pm}\,\ket{s,m} = \sqrt{s(s+1)\pm m(m+1)}\,\ket{s,m\pm 1}\,,\,\,\,|m|\le s
$$
У випадку спіна $s=1/2$ декартови компоненти оператора спіна визначаються матрицями Паулі:
\begin{equation}\label{pau}
\mathbf{n}\cdot \hat{\mathbf{S}} = n_{i}\,\hat{S}_{i} = \frac{\hbar}{2}\,n_{i}\,\sigma_{i}\\
  \sigma_{x}=\left(\begin{array}{cc}
  0 & 1 \\
  1 & 0
\end{array}\right)\,,\,\,
\sigma_{y}=\left(\begin{array}{cc}
  0 & -i \\
  i & 0
\end{array}\right)\,,\,\,
  \sigma_{z}=\left(\begin{array}{cc}
  1 & 0 \\
  0 & -1
\end{array}\right)
\end{equation}
***
\begin{gather}
 \sigma _{i}^{\,2} = 1\,,\quad \{\sigma _{i} ,\sigma _{j} \}
 = \sigma _{i} \sigma_{j} + \sigma _{j} \sigma _{i} = 2\,\delta _{ij} \label{eq_paulimtrx0} \\
[\sigma _{i} ,\sigma _{j}] = 2\,i\,\varepsilon_{i\,j\,k}\, \sigma _{k}\,,\Rightarrow
\sigma_{i}\sigma_{j} = \delta _{ij} + i\,\varepsilon_{i\,j\,k}\, \sigma _{k}\,. \label{eq_paulimtrx1}
\end{gather}

Виходячи з цих властивостей неважко переконатись, що

$$
\hat{\mathcal{U}}_{\mathcal{R}(\mathbf{n},\alpha)}=
\exp\left( \alpha\,\frac{\mathbf{n}\cdot\hat{\mathbf{S}}}{i\,\hbar}
\right) = \cos{\frac{\alpha}{2}} - i\,\sin{\frac{\alpha}{2}}\,\mathbf{n}\cdot\boldsymbol{\sigma}
$$

звідси витікає, що ймовірність спостереження проекції вздовж напрямка $\mathbf{n}_{1}$ якщо частинка перебуває у стані з проекцією спіна на напрямок $\mathbf{n}_{2}$ дорівнює:

$$
p(\mathbf{n}_{1};\mathbf{n}_{2})  = \left|\braket{\mathbf{n}_{1}}{\mathbf{n}_{2}}\right|^2 =  \cos^2{\frac{\alpha}{2}}
$$

Очевидно, що оператор проекції спіна може бути представлений у вигляді свого спектрального розкладу за проекційними операторами:

$$
\mathbf{n}\cdot \hat{\mathbf{S}} = \frac{\hbar}{2}\ket{\mathbf{n}}\bra{\mathbf{n}}
-\frac{\hbar}{2}\ket{-\mathbf{n}}\bra{-\mathbf{n}}
$$

## Гамільтоніан Паулі
Рівняння Шредінгера перетворюється у систему рівняннь для компонент хвильової функції, індексуємих проекцією спіна на певну вісь. Осакільки спін взаємодіє з магнітним полем, то гамільтоніан частинки зі спіном у зовнішньому ЕМП має вигляд
<!--  -->
\begin{equation}\label{hampauli}
H = \frac{1}{2m}
\left(- i\,\hbar \nabla -
\frac{q}{c} \mathbf{A} \right)^{\,2} + q\,\varphi -
\mathbf{M}_s \cdot \mathbf{H}\\
\mathbf{M}_s = \frac{q}{mc}\mathbf{S}
\end{equation}
<!--  -->
гамільтоніан \eqref{hampauli} називається гамільтоніаном Паулі.<br>
Густина струму ймовірності для \eqref{hampauli} будується відомим чином на основі рівнянь Шредінгера для хвильових функцій $\Psi $ та $\Psi^{\dagger} $, які розглядаються вже як вектори у гільбертовому просторі $\left(L_{2} \right)^{2s + 1}$ векторів-стовпців, елементами яких є координатні хвильові функції з простору $L_{2}$:
<!--  -->
\begin{equation}\label{eq_jpauli} 
\mathbf{j} = \frac{\hbar}{m}{\rm Im}\left(\Psi^{\dagger} \nabla \Psi \right) - 
\frac{q}{m\,c}\mathbf{A}\,\Psi^{\dagger}\,\Psi\,.
\end{equation}
<!--  -->
Струм $\mathbf{J}$ густини заряду $\rho = q\,\Psi^{\dagger} \Psi $ для такого гамільтоніана повинен, однак, крім звичайного вкладу $q\,\mathbf{j}$ враховувати також струм намагніченості:
<!--  -->
\begin{equation}\label{eq_jmag} 
\mathbf{J}_{\rm m} = q\,c{\rm rot}\left(\Psi^{\dagger} \mathbf{m}\Psi  \right) = \frac{q\,\hbar}{2\,m}{\rm rot}\left(\Psi^{\dagger} 
\boldsymbol{\sigma} \Psi \right)\,,
\end{equation}
<!--  -->
який, як відомо, має властивість ${\rm div} \mathbf{J}_{\rm m} = 0$, тобто чисто вихрового походження, і тому не дає внеска у перенос заряда. Таким чином, повний вираз для струму густини заряду є:
<!--  -->
\begin{equation}\label{eq_jpaulimag} 
\mathbf{J} = q\,\mathbf{j}\, +\mathbf{J}_{\rm m} = 
\frac{q\,\hbar}{m}{\rm Im}\left(\Psi^{\dagger} \nabla \Psi\right) - 
\frac{q^{\,2}}{m\,c^{\,2}}\,\mathbf{A}\Psi^{\dagger} \Psi +
\frac{q\,\hbar}{2\,m}\,{\rm rot}\left(\Psi^{\dagger}\boldsymbol{\sigma} \Psi 
\right)\,.
\end{equation}
<!--  -->
На його підставі можна виписати відповідний струм густини ймовірності:
<!--  -->
\begin{equation}\label{eq_jpaulimagw} 
\mathbf{J}_{w} = \frac{\hbar}{m}\,
{\rm Im}\left(\Psi^{\dagger} \nabla \Psi\right) - 
\frac{q}{m\,c^{\,2}}\,\mathbf{A}\Psi^{\dagger} \Psi +\frac{\hbar}{2\,m}\,{\rm rot}\left(\Psi^{\dagger} \sigma\Psi \right)\,.
\end{equation}
<!--  -->
Ці вирази, як і сам спін можна послідовно отримати виходячи з релятивістського рівняння Дірака, у нерелятивістському наближенні.
Оскільки, як виявляється, густина струму ймовірності явно залежить від вектор-потенціалу $\mathbf{A}$ виникає питання калібровочної інваріантності квантової механіки. Як відомо з класичної електродинаміки діючи сили визначаються напруженостями полів
$$
\mathbf{E} = -\frac{1}{c}\,\frac{\partial \mathbf{A} }{\partial t} - \nabla\, \varphi\,,\,\,\,\, \mathbf{B} = {\rm rot}\, \mathbf{A}
$$
які не залежать від калібровки потенціалів:
$$\label{edcalibr}
\tilde{\mathbf{A}} = \mathbf{A}+\nabla\,f\,,\,\,\, \tilde{\varphi} = \varphi - \frac{1}{c}\,f 
$$
<!--  -->
Відмітимо, що спіновий внесок в густину струма \eqref{eq_jpaulimagw} є калібровочно інваріантним, тому достатньо забезпечити калібрувальну інваріантність для інших внесків. Аналог калібрувального перетворення якій не змінює стан системи природньо задати увигляді зміни фази хвильової функції: 
$$
\tilde{\psi}  = \psi \exp \left( {i\,\theta}\right)
$$
Виходячи з:
<!--  -->
\begin{equation}\label{emeq10} 
\mathbf{ \tilde{j} } = \frac{\hbar}{m}{\rm Im}
\left(\tilde{\psi}^*\nabla \tilde{\psi} \right) - 
\frac{q}{m\,c}\tilde {\mathbf{A}}\vert \tilde{\psi}
\vert^{2} = \mathbf{j} + \left(\frac{\hbar}{m}\mathbf{\nabla} \theta- 
\frac{q}{m\,c}\mathbf{\nabla} f \right)\vert \psi \vert^{2}
\end{equation}
<!--  -->
та принципу калібровочної інваріантності отримаємо що при:
$$
\tilde{\psi} = \psi \exp \left(\frac{\,i\,q}{\,\hbar \,c}\,f \right)
$$
і
$$
\theta = \frac{q}{\,\hbar \,c}\,f\,,\,\,\,\,\, 
$$
струм ймовірності не зміниться.
<br>

На фундаментальних виразах \eqref{eq_jpauli}, \eqref{eq_jpaulimagw}  грунтується багато квантових ефектів.

<!--  -->
$$
\frac{m\,v^2_{\perp}}{R_{\perp}} = \frac{q}{c}\,v_{\perp}\,B\,,\,\,\,v_{\perp} = \omega_{0}\,R_{\perp}
$$
<!--  -->
дає рух по колу із характерним циклотроним радіусом $R_{\perp} = \frac{p_{\perp}\,c}{q\,B}$ та циклотронною частотою $\omega_{0} = \frac{q\,B}{m\,c}$. В квантовій механіці це буде відповідати тому що спектр енергії такої частинки має дві складові:
<!--  -->
$$
E = E_{\perp}+E_{\parallel}\,,\\
E_{\parallel} =  \frac{p^2_{\parallel}}{2\,m}\,,\,\,\, 
E_{\perp} =  \hbar\,\omega_{0}\left(n+\frac{1}{2}\right)
$$

### [Ефект Ааронова-Бома](http://link.aps.org/doi/10.1103/PhysRev.115.485)
<!--  -->
<img src="./figs/fig_abeffpaper1959.png" alt="Image" width="600">
<!--  -->

Це суто квантовий ефект фазового зсуву інтерференційної картини реєстрації зарядженної частинки в залежності від магнітного потоку, який може бути локалізований (теоретично) у нескінченно вузькому соленоїді.
<!--  -->
<img src="./figs/fig_abeffect.jpg" alt="Image" width="400">
<!--  -->
Поле соленоїда у найпростішої калібровці визначається виразом
<!--  -->
$$
\mathbf{A} = \frac{\Phi}{2\pi\,r}\,\mathbf{e}_{\varphi}\,\,\Rightarrow\,\, \mathbf{B} = {\rm rot} \mathbf{A} = \Phi\,\delta(r)\,\mathbf{e}_{z}
$$
<!--  -->
Зазвичай найпростіше еврістичне "поянення" цього ефекту полягає у розгляді інтерференції хвиль вздовж різних шляхів, які утворюють нестягуєму петлю навколо соленоїда. Тоді потік є єдиним параметром який обумовлює різницю фаз для таких хвиль, що і призводить до інтерференційного зсуву: 
<!--  -->
$$
\phi_{ABF}  = -\frac{q}{\hbar\,c}\,\int\limits_{ABF} \mathbf{A}\cdot d\mathbf{l}\,,\,\,\,\,\,
\phi_{ACF}  = -\frac{q}{\hbar\,c}\,\int\limits_{ACF} \mathbf{A}\cdot d\mathbf{l}'
$$
<!--  -->
Різниця фаз:
<!-- -->
\begin{equation}\label{abphasediff}
\Delta \phi = \frac{q}{\hbar\,c}\,\oint
\mathbf{A}\cdot d\mathbf{l} = \frac{e}{\hbar\,c}\,\Phi\,\,.
\end{equation}
<!--  -->
Слід однак пам'ятати, що електрон "сам із собою" не інтерферує і це тільки хвильова інтерпретація статистичної картини великої кількості незалежних актів реєстрації. З фізичної точки зору зсув обумовлений (локальною) взаємодією із точковою сингулярністю магнітного поля аналогічно до точкового потенціала. Це прямо витікає з виразу для гамільтоніана який не є калібровочно еквівалентним до вільного. У випадку квантованого потоку $\Phi = m\,\frac{\hbar\,c}{q}$ ми маємо таку еквівалентність і фазовий зсув відсутній. Звичайно це просто означає, що магнітне поле слід реєструвати іншим чином.

Експеримент підтверджує наявність зсуву в залежності від магнітного потоку.

```python
from ipywidgets import interactive
x0=4
def ints(x,a):
    ii=(1+np.cos(2*np.pi*x+a))/(2*(1+x**2))
    return ii
# Silly example data
x = np.linspace(-x0, x0, num=200, endpoint=True)
fluxdict = {0:['0','r','--',1],np.pi/2:['$\pi/2$','g','-',2],np.pi:['$\pi$','b','-',2]}      
# Make plot
for aa in fluxdict:
    plt.plot(x, ints(x,aa), linewidth=fluxdict[aa][3], 
             linestyle=fluxdict[aa][2],color=fluxdict[aa][1],
             label='flux = '+fluxdict[aa][0])
    plt.legend()
#     
plt.title('Two slit interference with AB flux shift')
plt.figure(figsize=(50,30))
plt.show()

def pii(a):
    x = np.linspace(-x0, x0, num=200, endpoint=True)
    plt.plot(x, ints(x,a),x,ints(x,0))
    plt.title('Two slit interference with AB flux shift')
#     plt.legend()
    plt.show()

interactive_plot = interactive(pii, a=(-np.pi, np.pi, 0.1))
interactive_plot
```
