Стан квантовой системи в даний момент часу несе всю інформацію про її поведінку між процесами вимірювання. Зміна вектора стану з часом описується лінійним оператором (еволюції) який діє на векторному просторі станів $L_{2}$
неперервно і бієктивно. Прийняття такого постулата приводит до того, що квантова механіка є динамічною теориєю. При цьому довільний стан з $L_2$ може бути як початковим так і кінцевим станом еволюції. Оператор еволюції $\hat{S}$ визначається як:
\begin{equation}\label{shft} 
\ket{\psi(t)}=\hat
S(t,0)\,\ket{\psi(0)}\,,\quad \ket{\psi(0)}=\hat
S^{-1}(t,0)\,\ket{\psi(t)}
\end{equation}
Оскільки впродовж еволюції факт існування системи є "інвариантом", то виконується умова збереження
норми вектора стану:
\begin{equation}
\label{shftn}
\braket{\hat{S}\,\psi}{
\hat{S}\,\psi}=\braket{\psi}{\psi}\,\,.
\end{equation}
Оператор еволюції визначений на всьому гільбертовому просторі і з \eqref{shftn} витікає, 
$$
\hat{S}^{\dagger}\hat{S} = \hat{S}\hat{S}^{\dagger} = 1 \Rightarrow \hat{S}^{\dagger} = \hat{S}^{-1}
$$ 
Такі оператори називають __унітарними__. Разом із ермітовими унітарні оператори грают важливу роль у квантовій теорії. Вони описують зв'язок між двома фізично еквівалентним представленнями фізичної реальності. Традиційно  математичная форма динамичного постулата формулюється в термінах швидкості зміни вектора стану. Ця швидкість є результом дії лінійного оператора:

\begin{equation}
\label{schr} i\,\hbar \frac{\partial}{\partial\, t}
\ket{\psi} =\hat{H}\, \ket{\psi}\\
\end{equation}

Оператор $\hat{H}$ називається __гамільтоніаном__ системи чи __оператором Шредінгера__ і є ермітовим, тому пов'язаний із сопостережуваною фізичною величиною - енергією. Гамільтоніан для найпростійшої системи -  вільної частинки є:

\begin{equation}
\label{h0} \hat{H}=\frac{\hat
p^{\,2}}{2m}=-\frac{\hbar^{\,2}}{2m}\Delta\,.
\end{equation}

Таким чином, гамільтониан квантової системи відповідає функції гамільтона її класичного аналога. Слід зауважити, що не для всіх квантових систем існують класичні аналоги. Крім того, операція співставлення класичній функції гамільтона квантового оператора не є однозначною тому що координата та імпульс не комутують. Вигляд гамільтоніана може бути легко знайдено лише для простих систем типу "частинка у зовнішньому потенціальному полі", де координатні і імпульсні змінні "розділені":

\begin{equation}
\label{hext} \hat H=\frac{\hat p^{\,2}}{2m}+\hat
V(\mathbf{r})=-\frac{\hbar^{\,2} }{2m}\Delta+\hat V(\mathbf{r})\,.
\end{equation}

Суттєво те що еволюція квантової системи описується унітарним оператором, а процесс вимірювання ні, оскільки спирається на взаємодію квантової системи і вимірюючого приладу. В цьому випадку квантова система не є замкненою тому еволюція стана при вимірюванні носить незворотній, не унітарний характер.

Визначимо вигляд оператора еволюції для стаціонарних систем тобто гамільтоніан яких не залежить від часу. Рівняння Шредінгера можно записать у вигляді:

\begin{equation}
\label{sev} i\,\hbar \frac{\partial}{\partial\, t}\hat S(t,0)=\hat
H\hat S(t,0)\,\, .
\end{equation}

Розв'язуючи  \eqref{sev} методом послідовних наближень, маємо:

\begin{equation}
\label{itt} \hat S(t,0)=1+\frac{t}{i\,\hbar}\hat H+
\frac{1}{2}\left(\frac{t}{i\,\hbar}\hat H\right)^{\,2}+\ldots+
\frac{1}{n!}\left(\frac{t}{i\,\hbar}\hat H\right)^{n}+\ldots=
\exp\left(\frac{t}{i\,\hbar}\hat H\right)
\end{equation}

Зміна ймовірності регістрації частинки у певному об'ємі $V$ визначається згідно 
\eqref{schr}:

\begin{equation}
\label{h} \frac{d }{d\, t}\int\limits_{V}\,|\,\psi|^{\,2}dV
=\frac{1}{i\,\hbar}\int\limits_{V} \left[\psi^{*}(\hat H\psi)-\psi(\hat H\psi)^{*}\right]dV
\end{equation}

Якщо $V$ це вся область регістрації, то \eqref{h} разом з умовою збереження, означает
ермітовість гамільтониана, тобто еквівалентність цієї властивості до вимоги збереження нормування (ймовірності). Підставляючи \eqref{hext} до \eqref{h} отримаємо:

\begin{eqnarray}
\label{consint} \frac{d}{d \,t}\int\limits_{V}\rho\, dV
&=&-\int\limits_{S}\mathbf{j}\cdot\mathbf{n}\,dS\\ \label{consdiff}
\frac{\partial \rho}{\partial\, t}&+&{\rm div \mathbf{j}}=0 ,\\
\label{cons} \quad\rho=&|\,\psi|^{\,2}&,\,\,\,\,
\mathbf{j}=\frac{\hbar}{m}{\rm Im}(\psi^{*}\nabla \psi)
\end{eqnarray}

Зауважимо, що \eqref{consint} і \eqref{consdiff} фактично є <a id='jcons'>законом збереження повної ймовірності у інтегральній і
диференціальній формі відповідно</a>. Якщо квантова система знаходиться у сінченному обємі, то зникнення інтегрального потоку ймовірності на границі цього обєму є умовою ермітовості оператора Шредінгера.

Якщо гамільтониан системи не залежить від часу, у системи є т.з. стаціонарні стани - власні стани гамільтоніана: 
<!--  -->
$$
\hat{H}\ket{\psi_E} = E\, \ket{\psi_E}
$$
<!--  -->
вони характеризуються тим, що їх часова залежність может бути виділена явним чином: 
<!--  -->
\begin{equation}
\label{stac} \psi_{E}(\mathbf{r},t) =\exp\left(\frac{E\,t}{i\,\hbar}\right)\phi(\mathbf{r})
\end{equation}
<!--  -->
Густина ймовірності $\rho$ в таких станах не залежить від часу і $\rm{div} \mathbf{j} =0$ у кожній внутрішній точці області руху. Зокрема, для хвиль де Бройля :
<!--  -->
$$
\psi_{\mathbf{k}}(\mathbf{r})=C\exp(i\,\mathbf{k}\cdot \mathbf{r}-
i\,\omega\, t)
$$
<!--  -->
згідно з \eqref{cons}
<!--  -->
\begin{equation}\label{dBrl}
  \mathbf{j}=\frac{\hbar\, \mathbf{k}}{m}\,|C|^2\,,
\end{equation}
<!--  -->
що відповідає змісту густини потока як числа частинок, що проходять через одиничную площадку в одиницю часу. Величина $\mathbf{v} = \frac{\hbar \mathbf{k}}{m}$ природньо ототожнюється із швидкістю потоку.

Спектр гамільтоніана є найважливішою характеристикою квантової системи. Як витікає з властивостей власних векторів для операторів фізичних величин, стаціонарні стани утворюють повний набір векторів так що довільний стан квантовой системи може бути представлено у вигляді їх суперпозиции:

\begin{equation}
\label{stat1} \bra{\psi}
=\sum\limits_{n,\ldots}c_{n}\bra{\phi_{n}}+\int\limits_{cont}\,a(E)\,\bra{\phi_{E}}\,dE\,\ldots
\end{equation}

тут крапки позначають інші змінні, які можуть бути задані в стані зпевною енергією у випадку виродженя. Використання бра-вектору замість кет тут є більш коректним оскільки стани неперервного спектру є функціоналами (бра-векторами) на просторі кет-векторів.

Якщо стаціонарні стани відомі то еволюція довільного стана $\ket{\psi}$ визначається суперпозицією 
еволюцій стаціонарних станів за \eqref{stat1}.

Квантова механіка є нерелятивістською теорією,
тому розглянемо питання про інвариантність рівняння Шредінгера при зміні інерціальної системи відліку. Перехід від ІСВ$\to$ІСВ' описується перетворенням Галілея <a id="galiley"></a>
<!--  -->
\begin{equation}\label{galileotransf}
\mathbf{r}' = \mathbf{r}-\mathbf{v}\,t\,,\quad t'=t
\end{equation}
<!--  -->
Імпульс та енергія змінюються відповідно:
<!--  -->
\begin{equation}\label{galileotransfenergymomnt}
\mathbf{p}' = \mathbf{p}-m\,\mathbf{v}\,,\quad E' = E - m\,\mathbf{v}\cdot \mathbf{p} + m\,\frac{\mathbf{v}^2}{2}\,.
\end{equation}
<!-- % -->
$$\exp\left( i\,\frac{\mathbf{p}\cdot \mathbf{r}}{\hbar} - i\,\frac{E\,t}{\hbar} \right)
\to \exp\left( i\,\frac{\mathbf{p}'\cdot \mathbf{r}'}{\hbar} - i\,\frac{E'\,t}{\hbar} \right)\,\exp\left( i\,\frac{m\,\mathbf{v}\cdot \mathbf{r}'}{\hbar} + i\,\frac{m\,v^2\,t}{2\,\hbar} \right)$$
<!-- % -->
Відповідно, оскільки довільний стан представляється у вигляді суперпозиції плоских хвиль, то остаточно:
<!-- % -->
\begin{equation}\label{galileopsi}
\Psi(\mathbf{r},t) = \Psi'(\mathbf{r}',t)\,\,\exp\left( i\,\frac{m\,\mathbf{v}\cdot \mathbf{r}'}{\hbar} + i\,\frac{m\,v^2\,t}{2\,\hbar} \right)
\end{equation}
<!-- % -->
Очевидно, це є унітарним перетворенням, яке відповідає зміні ІСВ.

## Зміна середніх з часом

Оскільки з часом стан квантової системи змінюється, то змінюються і середні значення фізичних величин. Розглянемо швидкість зміни середнього значення фізичної величини:

\begin{equation}
\frac{d}{d\,t} \langle{A}\rangle_{\psi} = \frac{d}{d\,t} \av{\psi}{A}{\psi} = 
\frac{d}{d\,t}\int {\psi^*} {A}\,\psi
\,dV =  
\end{equation}
$$\frac{1}{i\hbar} \int \left[- (H\psi)^* A\,\psi +\right. \psi^* i\hbar \,\left.\left(\frac{\partial}{\partial \,t} A
\right)\,\psi + \psi^*A\,H\psi \right] \,dV = 
\left\langle \frac{\partial}{\partial \,t}\,A +
\frac{i}{\hbar}\left[H,A \right] \right\rangle_{\psi}\,.
$$

Таким чином, зміна середнього значення може бути описана як
середнє значення деякого оператора. Відмітимо, що останній вираз може розглядатися як означення символу повної похідної за часом від оператора (в представленні Шредінгера):
<!--  -->
\begin{equation}
\label{eq2/6} {\frac{d}{{d\,t}}}{A} = {\frac{{\partial}}
{{\partial \,t}}}{A} + {\frac{{i}}{{\hbar}} }{\left[ {{H},{A}} \right]}\,.
\end{equation}
<!--  -->
В стаціонарних станах середнє від будь якого оператору не змінюється з часом, якщо сам оператор явно не залежить від $t$.

Вираз $${\frac{1}{{i\hbar}} }{\left[ {{A},{H}} \right]} =
\{{A},{H}\}$$ будемо називати квантовою дужкою Пуасона, оскільки
вона має такі ж формально алгебраїчні властивостями, що й аналогічна класична величина. Нагадаємо:
<!--  -->
$$\frac{d}{dt}f(t,q(t),p(t)) = \frac{\partial f}{\partial t} + \dot{q}\,\frac{\partial f}{\partial q} + \dot{p}\,\frac{\partial f}{\partial p} = \frac{\partial f}{\partial t} + \frac{\partial H}{\partial p}\,\frac{\partial f}{\partial q} -  \frac{\partial H}{\partial q}\,\frac{\partial f}{\partial p} = \frac{\partial f}{\partial t} + \{f,H\}$$

$$ \frac{\partial H}{\partial p}\,\frac{\partial f}{\partial q} -  \frac{\partial H}{\partial q}\,\frac{\partial f}{\partial p}  = \{f,H\}$$

$$\dot{q}  = \,\frac{\partial H}{\partial p}, \dot{p} = -\frac{\partial H}{\partial q}$$ 

$$\{f_1,f_2\} = - \{f_2,f_1\}\,,\,\,\,\{\{f_1,f_2\},f_3\} + \{\{f_3,f_2\},f_1\} + \text{cycle} = 0$$

Використовуючи явні вирази для операторів імпульсу, координати і гамільтоніан \eqref{hext} отримаємо співвідношення:

\begin{equation}
\label{eq3/6} {\frac{d}{{d\,t}}}\,\langle\mathbf{r}\rangle = \langle{\left\{ {\mathbf{r},{H}}
\right\}}\rangle = \langle \mathbf{p}/m\rangle\,\,,\,\frac{d}{d\,t}\,\langle \mathbf{p}\rangle =
\langle \left\{\mathbf{p},H \right\}\rangle = - \langle\vec {\nabla} V(\mathbf{r})\rangle\,,
\end{equation}
які можна назвати квантовим аналогом рівнянь Гамільтона.

Співвідношення подібні (\ref{eq3/6}) приводять до теореми Еренфеста, яка стверджує, що закони класичного руху справедливі для локалізованих пакетів у середньому ( __Теорема Еренфеста__ ). Якщо хвильова функція є фінітним пакетом, на просторовим протязі якого сила
приблизно постійна, то:
\begin{equation} \label{eq4/6}
m{\frac{{d^{\,2}}}{{d\,t^{\,2}}}}\,\langle \mathbf{r} \rangle \,
\approx \vec {F}( \langle \mathbf{r} \rangle )\,.
\end{equation}
<!--  -->
Виходячи з розгляду зміни середнього значення фізичної величини може бути введено поняття інтеграла руху у квантовій механіці. Відповідно до (\ref{eq2/6}) назвемо величину ${A}$ інтегралом
руху, коли:
<!--  -->
\begin{equation}
\label{eq5/6} {\frac{d}{{d\,t}}}{A} = {\frac{{\partial}}
{{\partial \,t}}}{A} + {\frac{{i}}{{\hbar}} }{\left[ {{H},{A}}
\right]} = 0\,.
\end{equation}
<!--  -->
Якщо така величина явно не залежить від часу, то інтеграл руху комутує з гамільтоніаном. Таким чином, довільний інтеграл руху може бути одночасно вимірюваний з енергією, тож існує можливість вибору спільних з гамільтоніаном власних станів.

Як бачимо. у такому підході довільна зміна значення фізичної величини пов'язана з
еволюцією вектора стану. Картина еволюції, яку було використано вище, відоме під
назвою шредінгеровської картини або <a id='shcrodrepr'>__представленням Шредінгера__</a>. Воно характеризується тим, що вектор стану (хвильова функція) явним чином залежить від часу, але спостережувальні величини від часу не залежать (якщо відсутня явна залежність
оператора від часу, що виражає той факт, що експериментальна процедура визначення величини не змінюється з часом). У такому підході зміна значення фізичної величини пов'язана з еволюцією вектора стану. Однак, оскільки вектор стану безпосередньо не спостерігається, його можна зробити другорядним елементом опису динаміки квантової системи, а всю часову залежність перенести прямо на фізичні величини. Таке представлення називається
хайзенберговською картиною або __представленням Хайзенберга__.

### Представлення Хайзенберга

Спосіб представлення еволюції, при якому вектор стана явно залежить від часу, а такі оператори як координата і імпульс не залежать від нього, відомий як представлення Шредінгера. В такому підході зміна значення фізичної величини пов'язано з еволюцією вектора стана (при умові відсутності залежності від часу відповідного оператора, що виражає факт того що експериментальная процедура її вимірювання не змінюється з часом). Однак, завдяки тому що вектор стана не є спостережуваною величиною всю часову залежність можна перенести на фізичні величини. Така картина називається __представленням Хайзенберга__. Оскільки еволюція векторів стана \eqref{shft} представляєтся унітарним перетворенням, то при перетворенні векторів
\eqref{shft} оператори змінюються як:

\begin{equation}
\label{top} \hat{A}_{H}=\hat{S}^{-1}(t)\hat{A}_{Sch}\,\hat{S}(t)
\end{equation}

Це співідношення визначає оператори фізичних величин в представлені Хайзенберга. Тобто якщо в представлені Шредінгера оператори спостережуваних які не залежать від часу діють на зміний вектор стану $\ket{\psi}$, то у хайзенберговському представленні еволюції оператори фізичних величин явно залежать від часу, і діять на статичний хайзенберговський вектор стана $\ket{\psi(0)}$ (він може бути обраний довільно, оскільки всі стани гільбертова простору унітарно-еквівалентні). Рівняння

$$
\frac{d}{d\,t} \hat{A}_{H} =
  \frac{\partial\, }{\partial\, t} \hat{A}_{H} +\frac{1}{i\,\hbar}
  [\hat{A}_{H},\hat{H}]
$$

є <a id='heizeq'>рівнянням руху для операторів у представленні Хайзенберга</a>.
Спостережувальні величини, средні значення яких не залежать від часу називаються __інтегралами руху__. Хайзенберговський оператор інтеграла руху задовільняє рівнянню:

\begin{equation}\label{hop}
  \frac{d}{d\,t} \hat{A}_{H} = 0\,.
\end{equation}

Такий підхід, очевидно, майже ідентичний до канонічного формалізму класичної механіки в якому роль комутатора відіграє дужки Пуассона, яка визначає еволюцію значення динамічної функції $f(q,p,t)$ на фазовому просторі. Така еволюція виникає внаслідок "гамільтонівського потоку" точок $(q,p)$ фазового простору:

$$
\frac{d\,f}{d\,t} = \frac{\partial f}{\partial t} + \{f,H\}
$$


Таким чином, комутатор операторів можна розглядати як "квантові дужки Пуассона":

$$
\frac{1}{i\,\hbar}\,[\hat{A},\hat{B}] \Leftrightarrow \{f_{A},f_{B}\}
$$
$$
\{\{f_a,f_b\},f_c\}+cycle=0\\
[[A,B],C]+cycle = 0
$$

Зокрема канонічні "рівняння руху" для хайзенбергівських операторів координати та імпульса мають вигляд:

$$
i\,\hbar \,\frac{d\,\hat{x}_{H}}{d\,t} = [\hat{x}_{H}, \hat{H}]\,,\,\,\,
i\,\hbar \,\frac{d\,\hat{p}_{H}}{d\,t} = [\hat{p}_{H}, \hat{H}]\\
\hat{x}_{H}(t) = \hat{S}^{\dagger}(t)\,\hat{x}\,\hat{S}(t)\,,\,\,\,\,\hat{p}_{H}(t) = \hat{S}^{\dagger}(t)\,\hat{p}\,\hat{S}(t)
$$

де $\hat{x},\hat{p}$ - оператори координати та імпульса в представленні Шредінгера, які були введені раніше.

У нерелятивістській теорії представлення (інтерпретації) динаміки Хайзенберга та Шредінгера є унітарно еквівалентними, тобто їх неможливо розрізнити на рівні експеримента. Але з теоретичної точки зору хайзенберговський підхід заснований на операторах спостережуваних величин є більш загальним і застосовується у релятивістській квантовій теорії (квантовій теорії поля) де введення одночастинкової хвильової функції з ймовірносною інтерпретацією квадрата модуля втрачає фізичний зміст.

### Співвідношення енергія-час
Так само як і у класичній механіці у квантовій час є виділеним параметром (навіть не оператором!) від якого залежать всі інші величини. Це прояв  нерелятивістської концепції абсолютного часу де ми нехтуємо скінченністю та локальним характером фізичних процесів росповсюдження енергії та сигналів. Наявність гамільтоново спряжених величин в класичній механіці

з алгебраіної точки зору відповідає наявності некомутуючих операторів у квантовій.

Раніше ми вивели співвідношеня невизначенностей для спостережуваних величин, які зображуються операторами. Аналогічним чином розглядаючи нерівність: 

$$
 \forall\, \lambda  \in  \mathbb{R}\,,\,\,\,
0 \le \left|\left|\left(i\,\lambda\,\hat{A}+\hat{H}\right)\psi \right|\right|^2
$$

де $\hat{A}$ -  величина яка явно не залежіть від часу, $\hat{H}$ - гамільтоніан, можна отримати наступне співвідношення:

$$
\tau_{A} \sigma_{H} \ge \frac{\hbar}{2}\,,\,\,\,\,\,\,
\tau_{A} = \,\left| \frac{d\,\left\langle \hat{A} \right\rangle}{d\,t} \right|^{-1}\,\sigma _{A} 
$$

Характеристичний час $\tau_{A}$ можна вважати мірою нестацінарності стану з точки зору фізичної величини $\hat{A}$.
