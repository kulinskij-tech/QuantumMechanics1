__Задача__ Покажіть, що класичне рівняння руху зарядженої частинки у зовнішньому ЕМП з урахуванням сили Лоренца
може бути записане у гамільтонівській формі з гамільтоніаном
$$
H =
\frac{1}{2m}\left(\mathbf{p} - \frac{q}{c}{\rm \mathbf{A}}
\right)^{\,2} + q\,\varphi
$$
__Розв'язок__ Рівняння Н'ютона має вигляд:
<!--  -->
\begin{equation}\label{n1}
  m\,\dot{\mathbf{v}} = q\, \mathbf{E} + \frac{q}{c}\,\mathbf{v}\times {\rm rot}\,\mathbf{A}
\end{equation}
<!--  -->
$$
\mathbf{E} = -\nabla\varphi - \frac{1}{c}
\frac{\partial\, \mathbf{A}}{\partial\, t}
$$
Визначимо функцію Гамільтона $H$ так, щоби рівняння руху мало би канонічну гамільтонівську форму:
$$
\left\{ \begin{array}{rl}
  \dot{\mathbf{p}}= & -
  \frac{\partial\, H}{\partial\, \mathbf{r}} \\
  \dot{\mathbf{r}}= &
  \frac{\partial\, H}{\partial\, \mathbf{p}}
\end{array}\right.
$$
Виділяємо у \eqref{n1} повну похідну за часом від вектор-потенціалу:
<!--  -->
\begin{equation}\label{n2}
  \frac{d}{d\,t}\left(m\,\mathbf{v}+\frac{q}{c}\mathbf{A} \right) = 
  -q\,\nabla\varphi + \frac{q}{c}\left(\mathbf{v}\cdot\nabla\mathbf{A}
  +\mathbf{v}\times {\rm rot} \mathbf{A}\right)\,.
\end{equation}
<!--  -->
Оскільки  права частина \eqref{n2} теперь не містить повних похідних за часом, позначимо нову змінну: 
$$
\mathbf{p} = m\,\mathbf{v}+\frac{q}{c} \mathbf{A}\,.
$$ 
Отримане співвідношення означає, що в даному випадку канонічний імпульс $\mathbf{p}$, якій у квантовомеханічному формалізмі відповідає оператору $-i\hbar \nabla$ у координатному представлені, не співпадає із "кінематичним" імпульсом $m \mathbf{v}$.
Таким чином:
$$
\begin{array}{rl}
  \dot{\mathbf{r}} =  & \frac{1}{m}\left(\mathbf{p} - \frac{q}{c} \mathbf{A}\right) =
\frac{\partial\, H}{\partial\, \mathbf{p}} \\
  \dot{\mathbf{p}} = & - \frac{\partial\, }{\partial\,
\mathbf{r}}\left(q\,\varphi + \frac{1}{2m}\left(\mathbf{p} - \frac{q}{c}
\mathbf{A}\right)^2\right)\,,
\end{array}
$$
що і означає що функцію Гамільтона слід обрати наступним чином:
$$
H = \frac{1}{2m}\left(\mathbf{p} - \frac{q}{c} \mathbf{A}\right)^2 +q\,\varphi\,.
$$

__<a id='magosc'>Задача</a>__ Знайти спектр зарядженої частинки у однорідному магнітном полі. <br> 
__Рішення__ Магнітне поле будемо вважати спрямованим вздовж вісі $z$: $\mathbf{B}= (0,0,B)$.
Тоді можна вважати що $\mathbf{A}= (0,H\,x,0)$ (оберемо відповідну калібровку). Рівняння Шредінгера має вигляд:
$$
\left(-\frac{\hbar^2}{2m}\frac{\partial^2}{\partial\,x^2}+
\frac{1}{2m}\left(-i\,\hbar\frac{\partial}{\partial\,y}-\frac{q}{c}\,B\,x \right)^2 -\frac{\hbar^2}{2m}\frac{\partial^2}{\partial\,z^2}\right)\psi=E\,\psi
$$
Зробимо розділення змінних:
$$
\psi = X(x)\,e^{i\,k_y y+i\,k_z z}
$$
тоді
$$
\left(-\frac{\hbar^2}{2m}\frac{\partial^2}{\partial\,x^2}+
\frac{1}{2m}\left(p_y-\frac{q}{c}\,B\,x \right)^2 \right)X(x)=\varepsilon\,X(x)
$$
де $p_y = \hbar\,k_y\,,\,\,\varepsilon = E-\frac{\hbar^2 k^2_{z}}{2\,m}$. Введемо
$$
x_0 = \frac{p_y\,c}{q\,B}\,,\quad \omega_{0} = \frac{q}{m\,c}\,B 
$$
так що отримаємо рівняння гармонічного осцилятора (!):
$$
\left(-\frac{\hbar^2}{2m}\frac{\partial^2}{\partial\,\tilde{x}^2}+
\frac{m\,\omega^2_{0}}{2}\tilde{x}^2 \right)\,X=\varepsilon\,X
$$
де $\tilde{x} = x-x_0$. Cпектр цієї задачі нам вже відомий - це спектр гармонічного осцилятора:
$$
\varepsilon_{n} = \hbar\,\omega_{0}\,\left(n+\frac{1}{2}\right)
$$
Це т.з. рівні Ландау. Такі квантові стани відповідають фінітному класичному руху по колам у поперечній до магнітного поля площині. Рух вздовж напрямка магнітного поля не квантується оскільки він є вільним. Також слід звернути увагу на нескінченність кратності виродження рівней Ландау по параметру $x_0$, оскільки всі функції вигляду $X_{n}(\tilde{x})\,e^{i\,k_y\,y}$ відповідають одному і тому ж значенню $\varepsilon_{n}$. Як завжди нескінченність такого виродження повязана з необмеженністю руху (трансляційній інваріантності в площині $xy$). 
Якщо область руху обмежена, то кратність скінченна.

__Задача__ Нехай $A = \mathbf{a}\cdot \boldsymbol{\sigma}$ - оператор спостережувальної величини у спіновому просторі частинки $s=1/2$. Який вид має оператор, якщо систему піддати обертанню $\mathcal{R}(\mathbf{n},\alpha)$?
<br> 
__Рішення__ Перетворення обертання індукує перетворення спінових векторів стана:

$$
\psi \rightarrow \hat{\mathcal{U}}_{\mathcal{R}(\mathbf{n},\alpha)} \psi\,,
$$

де 

$\hat{\mathcal{U}}_{\mathcal{R}(\mathbf{n},\alpha)}$ - оператор [перетворення у спіновому просторі](https://colab.research.google.com/github/kulinskij-tech/QuantumMechanics1/blob/main/QM_py/qm1_spinpauli_theor.ipynb#moms). 
Тому

$$
A \rightarrow \hat{\mathcal{U}}_{\mathcal{R}(\mathbf{n},\alpha)}\, A\,
\hat{\mathcal{U}}_{\mathcal{R}(\mathbf{n},\alpha)}^{\,\dagger}\,.
$$

Використовуючи представлення властивості матриць Паулі, отримуємо перетворення у інфінітезимальному  вигляді:

\begin{eqnarray}
  \hat{\mathcal{U}}_{\mathcal{R}(\mathbf{n},\alpha)}\, A\,
\hat{\mathcal{U}}_{\mathcal{R}(\mathbf{n},\alpha)}^{\,\dagger}
&=&|\mathbf{a}|\, \left(1-\frac{i\,\alpha}{2}\,\mathbf{n}\cdot
\boldsymbol{\sigma} + o(\alpha) \right) \mathbf{m}\cdot
\boldsymbol{\sigma} \left( 1+\frac{i\,\alpha}{2}\,\mathbf{n}\cdot
\boldsymbol{\sigma} + o(\alpha)\right) =\notag\\
&&|\mathbf{a}|\left(\mathbf{m}\cdot \boldsymbol{\sigma} +
i\,\frac{\alpha}{2}\,[\mathbf{m}\cdot \boldsymbol{\sigma},
\mathbf{n}\cdot \boldsymbol{\sigma}]+ o(\alpha)\right) =\notag \\
&&|\mathbf{a}|\left(\mathbf{m}\cdot \boldsymbol{\sigma} -
\alpha\,(\mathbf{m}\times \mathbf{n}) \cdot
\boldsymbol{\sigma}+ o(\alpha)\right) =\notag\\
&&\mathbf{a}\cdot \boldsymbol{\sigma} +\,\alpha\,(\mathbf{n}\times
\mathbf{a}) \cdot \boldsymbol{\sigma}+ o(\alpha)\label{ur}\,.
\end{eqnarray}

Друга складова це як раз інфінітезімальна зміна вектора $\mathbf{a}$ при обертанні навколо вісі $\mathbf{n}$. Остаточно отримаємо:

$$
\hat{\mathcal{U}}_{\mathcal{R}(\mathbf{n},\alpha)}\, A\,
\hat{\mathcal{U}}_{\mathcal{R}(\mathbf{n},\alpha)}^{\,\dagger} =
\mathbf{a}'\cdot \boldsymbol{\sigma}\,,\,\,\,\, \mathbf{a}' =
\mathcal{R}(\mathbf{n},\alpha)\,\mathbf{a}\,.
$$

__Задача__ Показати, що рівняння руху одномірного гармонічгого осцилятору у гамільтонівській формі формально тотожньо рівнянню Шредінгера для спіновой частини хвильової функції частинки зі спіном $s=1/2$ у магнітному полі. <br> 

__Розв'язок__ Рівняння:
\begin{equation}\label{osc}
  \frac{d}{d t} \begin{pmatrix}
    q \\
    p \
  \end{pmatrix}=\begin{pmatrix}
    0 & 1 \\
    -\omega^2 & 0 \
  \end{pmatrix}\begin{pmatrix}
    q \\
    p \
  \end{pmatrix}
\end{equation}
Зробимо заміну $p\to \omega p$ , тоді \eqref{osc} має вигляд:
\begin{equation}\label{oscs}
  \frac{d}{d t} \begin{pmatrix}
    q \\
    p \
  \end{pmatrix}=\omega \begin{pmatrix}
    0 & 1 \\
    -1 & 0 \
  \end{pmatrix}\begin{pmatrix}
    q \\
    p \
  \end{pmatrix}
\end{equation}
що, очевидно, еквівалентно:
\begin{equation}\label{sch}
  i\frac{d}{d t} \begin{pmatrix}
    q \\
    p \
  \end{pmatrix}=\omega \begin{pmatrix}
    0 & i \\
    -i & 0 \
  \end{pmatrix}\begin{pmatrix}
    q \\
    p \
  \end{pmatrix}
\end{equation}
Якщо ввести $\psi \stackrel{\rm def}{=} \begin{pmatrix}
    q \\
    p \
  \end{pmatrix}\,,$ то \eqref{sch} запишеться у вигляді:
\begin{equation}\label{sch1}
  i\,\hbar\,\frac{d}{d t}\,\vert\,\psi\,\rangle=-\hbar\,
  \omega \,\hat{\sigma}_{2}\,\vert\,\psi\,\rangle = -
  \frac{q}{m\,c}\,\mathbf{H}\cdot\hat{\mathbf{S}}\,
  \vert\,\psi\,\rangle\,,
\end{equation}
$\mathbf{H} = \left(0,\frac{2\,m\,c}{q}\,\omega,0\right)
\,,\quad
\hat{\mathbf{S}}=\frac{\hbar}{2}\,\hat{\boldsymbol{\sigma}}$
яке має Розв'язок:
\begin{equation}\label{resch}
  \vert\,\psi_{t}\,\rangle =
  \exp\left(-i\,\omega\,t\,\hat{\sigma}_{2}\right)\vert\,
  \psi_{0}\,\rangle
  =(\cos{\omega\,t}-i\,\hat{\sigma}_{2}\sin{\omega\,t})\,
  \vert\,\psi_{0}\,\rangle
\end{equation}
Слід зауважити що така аналогія має місце лише якщо поле $\mathbf{H}$ не залежить від часу.

__Задача__ Частинка зі спіном $s=1$ описується гамільтоніаном: $\hat H = A\,\hat{S}^2_x+B\,\hat{S}^2_y\,.$ Визначте його спектр. <br> 
__Розв'язок__ Зазначимо, що випадок $s=1/2$ є тривіальним, оскільки завдяки $(\mathbf{n}\cdot \mathbf{S})^2 = 1/4$ і тому:
$$
\hat{H} =
\frac{A+B}{4}\,\hat{I}\,.
$$
Крім того задача легко вирішується, якщо $A=B$:
$$
\hat H =
A\,\left(\hat{S}^2_x+\hat{S}^2_y\,\right) = A\left(\,
\hat{S}^2-\hat{S}^2_z\,\right) = A\,s(s+1) - A\,m^2\,,\quad
m=-s,\ldots,+s\,.
$$
Використовуючи цей факт представимо гамільтоніан у вигляді:
\begin{equation*}
\begin{split}
\hat H = B\,\left(\hat{S}^2_x+\hat{S}^2_y\,\right) +
(A-B)\,\hat{S}^2_x = B\,\left(\hat{S}^2-\hat{S}^2_z\,\right) +
\frac{A-B}{4}\,\left(\,\hat{S}_{+}+\hat{S}_{-}\,\right)^2 =\\
\frac{A+B}{2}\,\left(\,\hat{S}^2-\hat{S}_{z}^2\,\right)
+\frac{A-B}{4}\,\left(\,\hat{S}_{+}^2+\hat{S}_{-}^2\,\right)\,.
\end{split}
\end{equation*}
Зрозуміло, що для $s=1$:
\begin{gather}
\hat{S}^2_{+} \vert\,0\,\rangle =  \hat{S}^2_{-} \vert\,0\,\rangle = 0\,,\notag \\
\hat{S}^2_{+} \vert\,-1\,\rangle =
2\vert\,+1\,\rangle\,, \label{spm}\\
\hat{S}^2_{-} \vert\,+1\,\rangle = 2
\vert\,-1\,\rangle \notag\,.
\end{gather}
Звідки витікає що вектори:
$$ \vert\,0\,\rangle\,,
\frac{1}{\sqrt{2}}\left(\,\vert\,+1\,\rangle+\vert\,-1\,\rangle\,\right)\,,
\frac{1}{\sqrt{2}}\left(\,\vert\,+1\,\rangle-\vert\,-1\,\rangle\,\right)
$$
є власними для оператора $\hat{S}^2_{+}+\hat{S}^2_{-}$ із значениями $0\,,2\,,-2$ відповідно. В результаті знаходимо спектр:
\begin{equation}\label{spec1}
E_{1} = \frac{A+B}{2}\,,\quad E_{2} = A\,,\quad E_{3} = B\,.
\end{equation}
Зауважимо, що геометричне перетворення $x \leftrightarrow y$ для гамільтоніана еквівалентно зміні $A \leftrightarrow B$ і тому не змінює спектра, призводячи до перепозначення власних чисел. Тобто спектр не змінюється при заміні $x \leftrightarrow y\,,\,\,A \leftrightarrow B$.<br>
***
Звичайно можна зробити обчислення безпосередньо використовуючи матричний вид оператора. Сподручніше знов скористатись операторами $\hat{S}_{\pm}$:
$$
\hat{H} = \frac{A}{4}\,(\hat{S}_{+}+\hat{S}_{-})^2 - \frac{B}{4}\,(\hat{S}_{+}-\hat{S}_{-})^2
$$
