```
sinc0[n_Integer]:= Fold[Times, Sqrt[#/Pi] Sinc[k/#]&/@ Range[n]]
sinc3[n_Integer]:= Fold[Times, Sinc[k/(2*# -1 1)]& /@ Range[n]]
isinc[n_]:= Integrate[sinc3[n], {k, 0, \[Infinity]}]
sincseq=isinc/@ Range[8]
```

Скоротимо на $\pi/2$

```
(2*sincseq/Pi)
```

```
n0 = 4; sinc0[n0]
f = InverseFourierTransform[sinc0[#], k, x] & /@ Range[n0];
f3 = InverseFourierTransform[sinc3[#], k, x] & /@ Range[n0];
GraphicsRow[{Plot[f, {x, -2., 2.}, PlotRange -> All, PlotStyle -> Thick],
Plot[f3, {x, -1., 1.}, PlotRange -> All, PlotStyle -> Thick]}]
```
