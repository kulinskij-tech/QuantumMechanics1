## Квантові кореляції. ЕПР пари. Нерівність Белла.

Природа і спосіб опису квантово-механічних об'єктів, які є предметом дослідження квантової механіки, суттєво відрізняються від відповідних класичних аналогів. Основні відмінності пов'язані з існуванням принципу суперпозиції чистих станів і неможливістю одночасного вимірювання фізичних величин, які не комутують. Це має наслідком те, що не існує сумісної функції розподілу ймовірності значень не комутуючих фізичних величин. Для класичних об'єктів така процедура як сума станів, взагалі кажучи, не визначена, а фізичним величинам відповідають дійсні функції на фазовому просторі імпульсів та координат. Стан системи повністю визначається фіксацією точки фазового простору.

Також, на відміну від класичної мехніки де систему можна мислити як набір різних матеріальних точок, квантово-механічні системи є __несепарабельними__, оскільки простір станів та фізичних величин комбінованих систем будується на основі тензорного добутку. Ця специфічна риса квантової механіки проявляється у наявності кореляцій між фізичними величинами для підсистем, обумовлених процедурою приготування початкового стану комбінованої системи який не можна представити як добуток станів окремих підсистем.

Найпростішою квантовою системо є [q-bit](https://colab.research.google.com/github/kulinskij-tech/QuantumMechanics1/blob/main/QM_py/qm1_qbitspace.ipynb#qbit) для якого простір векторів стану є $\mathbb{C}^2$.Сукупність із $N$ таких $q$-бітів являє собою комбіновану систему і має, очевидно, в якості простору станів $\mathbb{C}^{2} \otimes ... \otimes \mathbb{C}^{2}$. Розмірність такого простору є $2^{N}$. При цьому припускається, що не накладається інших умов, пов'язаних з тотожністю $q$-бітів, тобто припускається, що вони принципово
різні (добре локалізовані стани) і обмінними ефектами можна знехтувати.

Розглянемо систему з двох не взаємодіючих $q$-бітів $Q_{1}$\, і $Q_{2} $. Конкретно будемо мати на увазі спінові стани. Тоді повний спін $\mathbf{S} = \mathbf{s}_{1} + \mathbf{s}_{2}$\, комбінованої системи $Q_{1} \otimes Q_{2} $, завдяки правилу додавання моментів, може відповідати тільки значенням $S = 0\,,\,1$ спінового квантового числа. Відповідно, можна говорити про синглетний $S = 0$\, та
триплетний стани $S = 1$\, комбінованої системи. Задля конкреттики візьмемо синглетний стан:

\begin{equation}
\label{eq3/14} \ket{\Sigma} = \frac{1}{\sqrt
{2}}\left( \ket{\uparrow}_{1} \otimes
\ket{\downarrow}_{2} - \ket{\downarrow}_{1} \otimes
\ket{\uparrow}_{2}
\right)\,.
\end{equation}

Відповідний статистичний оператор за означенням є:

$$
\label{rhosingl}
 \hat{\rho}_{\Sigma} = \ket{\Sigma}\bra{\Sigma}
 =\frac{1}{2}\left(\ket{\uparrow}_{1} \otimes \ket{\downarrow}_{2} - \ket{\downarrow}_{1} \otimes \ket{\uparrow}_{2} \right)\,\left(\bra{\uparrow}_{1} \otimes
\bra{\downarrow}_{2} -  \bra{\downarrow}_{1}\otimes
\bra{\uparrow}_{2} \right) = \\
\frac{1}{2}\left(
\ket{\uparrow}_{1}\bra{\uparrow}_{1} \otimes \ket{\downarrow}_{2}\bra{\downarrow}_{2}
+ \ket{\downarrow}_{1}\bra{\downarrow}_{1}\otimes \ket{\uparrow}_{2}\bra{\uparrow}_{2} - 
\ket{\uparrow}_{1}\bra{\downarrow}_{1} \otimes \ket{\downarrow}_{2}\bra{\uparrow}_{2}
-\ket{\downarrow}_{1}\bra{\uparrow}_{1} \otimes \ket{\uparrow}_{2}\bra{\downarrow}_{2} \right)
$$

Вважаємо спін інтегралом руху (ми нехтуємо
релятивістськими ефектами). Якщо проводити будь які вимірювання тільки над першою частинкою, то її стан представляється статоператором:
<!-- $$
\hat{s}^{(z)}_{1} = \frac{\hbar}{2}\left(\ket{\uparrow}_{1}\bra{\uparrow}_{1} - 
\ket{\downarrow}_{1}\bra{\downarrow}_{1}\right)
$$
 -->
$$
\hat{\rho}^{(1)} = {\rm Tr}_{2} (\rho_{\Sigma}) = \frac{1}{2}\left(\ket{\uparrow}_{1}\bra{\downarrow}_{1}+\ket{\downarrow}_{1}\bra{\downarrow}_{1}
\right)
$$

що означає статистичну суміш двох чистих станів з протилежними проекціями спіну. Якщо будемо вимірювати 
певний стан другої частинки, скажімо $\ket{\uparrow}$, то після вимірення система перейде у стан:

$$
\ket{\uparrow}_{2}\bra{\uparrow}_{2}\,\hat{\rho}_{\Sigma} \to \ket{\downarrow}_{1}\otimes \ket{\uparrow}_{2}
$$

Тобто результат вимірювання цієї ж компоненти для іншої частинки визначений. Він відрізняється тільки знаком. Можна одночасно робити вимірювання над просторово розділеними хвильовими пакетами, з тим щоб виключити можливий причинний вплив вимірювань. Результат не зміниться: виявляється 100\% кореляція у результатах вимірювання фізичних величин $s_{z}^{(1)} $ і $s_{z}^{(2)} $. Це ще не є, однак, чимось дивовижним, оскільки для класичної системи така кореляція також має місце - це наслідок закону збереження. Однак, ситуація змінюється якщо виміровати компоненти у різних напрямках. Якбі би спін був просто класичним випадковим вектором, то кореляції проекцій спінів для різних частинок у різних напрямках відповідали певному розподілу компонент випадкового вектора. Ці кореляції визначаються функцією розподілу ймовірностей для компонент випадкового вектора. За квантовою механікою не існує станів в яких різні компоненти спіну мають визначені значення, тому не існує функції сумісного розподілу ймовірності для цих компонент. Вимірюючи кореляції таким чином можна встновити квантовий чи класичний характер розподілу. 

Визначимо ймовірності того, що частинки опиняться у станах з заданими проекціями спінів, скажімо вздовж напрямків, які задані одиничними векторами $\mathbf{n}_{1} ,\mathbf{n}_{2} $. Виберемо вектор $\mathbf{n}_{1}$ вздовж вісі $z$, тоді

\begin{equation}
\label{eq5/14} 
\left\vert\braket{\mathbf{n}_{1} \otimes
\, \mathbf{n}_{2}}{\Sigma}
\right\vert^{2} = \frac{1}{2}\,
\left\vert\braket{\mathbf{n}_{2}}{\downarrow}_2
\right\vert^{\,2} = 
\frac{1}{2}\,
\left\vert\av{\downarrow}{U(\mathbf{n}_2)}{\uparrow}_2
\right\vert^{\,2}=
\frac{1}{2}\sin
^{\,2}\frac{\varphi}{2}
\end{equation}

де $\varphi $ - кут між векторами $\mathbf{n}_{1} ,\mathbf{n}_{2}
$. 
Тут ми скористались виразом

\begin{equation}\label{uu}
  U = \exp\left(\alpha \frac{\mathbf{n}\cdot
\hat{\mathbf{S}}}{i\,\hbar} \right) = \exp \left( - i\frac{\alpha}
{2}\hat{\boldsymbol{\sigma}} \cdot \mathbf{n} \right) = \cos
\left( {{\frac{{\alpha }}{2}}} \right) - i\,\hat{\boldsymbol{\sigma}}
\cdot {\rm \mathbf{n}}\sin \left( {{\frac{{\alpha}} {2}}} \right)\,,
\end{equation}

отриманим раніше (див. ).

Вимірювання середнього значення величини $\mathbf{n}_{1} \cdot
\mathbf{S}_{1} \otimes \mathbf{n}_{2} \cdot \mathbf{S}_{2} $ дає
результат:
$$
\begin{array}{l}
 {\left\langle {\Sigma}  \right|}\mathbf{n}_{1} \cdot \mathbf{S}_{1}
\otimes \mathbf{n}_{2} \cdot \mathbf{S}_{2} {\left| {\Sigma}
\right\rangle}  = \\

\right|_{1} \mathbf{n}_{1} \cdot \mathbf{S}_{1} {\left| \downarrow
\right\rangle}_{1} {\left\langle \uparrow \right|}_{2}
\mathbf{n}_{2} \cdot \mathbf{S}_{2} \left| \uparrow  \right\rangle
_{2} -}
\\

\left| \uparrow  \right\rangle_{2} - \left\langle \downarrow
\right|_{1} \mathbf{n}_{1} \cdot \mathbf{S}_{1} {\left| \uparrow
\right\rangle }_{1} {\left\langle \uparrow \right|}_{2}
\mathbf{n}_{2} \cdot \mathbf{S}_{2} {\left| \downarrow

 \end{array}
$$
$$=- {\frac{{\hbar ^{2}}}{{4}}}\mathbf{n}_{1} \cdot
\mathbf{n}_{2}\,,$$ або безпосередньо використовуючи
(\ref{eq5/14}):
\begin{equation}
\label{eq6/14} {\left\langle {\Sigma}  \right|}\mathbf{n}_{1}
\cdot \mathbf{S}_{1} \otimes \mathbf{n}_{2} \cdot \mathbf{S}_{2}
{\left| {\Sigma} \right\rangle}  = {\frac{{\hbar
^{2}}}{{4}}}\left( {\sin ^{2}{\frac{{\varphi }}{2}} - \sin
^{2}{\frac{{\pi - \varphi}} {2}}} \right) = - {\frac{{\hbar
^{2}}}{{4}}}\mathbf{n}_{1} \cdot \mathbf{n}_{2}\,.
\end{equation}
Відмітимо, що середнє значення проекції спіну кожної з частинок
дорівнює $0$:
\begin{equation}
\label{eq7/14} {\left\langle {\Sigma}  \right|}\,\mathbf{n}_{1}
\cdot \mathbf{S}_{1} \otimes {\mathbf{I}}\,{\left| {\Sigma}
\right\rangle} = {\left\langle {\Sigma}  \right|}\,{\mathbf{I}}
\otimes \mathbf{n}_{2} \cdot \mathbf{S}_{2} \,{\left| {\Sigma}
\right\rangle}  = 0\,.
\end{equation}
З точки зору класичної теорії ймовірностей (\ref{eq6/14}) є кореляційною
функцією відповідних випадкових величин. Дійсно, якщо $N_{s_{1}
,s_{2}}  $ - число випадків, в яких було зареєстровано значення
проекцій $s_{1} ,s_{2} $ для кожної з частинок у напрямках
$\mathbf{n}_{1} ,\mathbf{n}_{2} $ відповідно, то:
\begin{equation}
\label{eq8/14} {\left\langle {\Sigma}  \right|}\,{\rm
\mathbf{n}}_{1} \cdot {\rm \mathbf{S}}_{1} \otimes \, {\rm
\mathbf{n}}_{2} \cdot {\rm \mathbf{S}}_{2} {\left| {\Sigma}
\right\rangle}  = {\frac{{N_{+ \, +}  + N_{- \, -}  - N_{+ \,-} -
N_{- \, +}} } {{N_{+ \, +} + N_{- \, -}  + N_{+\, -}  + N_{- \,
+}} } }\,.
\end{equation}
Наприклад, для ситуації $\mathbf{n}_{1} = \mathbf{n}_{2} $ з (\ref{eq6/14}) витікає $N_{+\, +}  = N_{ - \, -}  = 0$, як і вище. Для $\mathbf{n}_{1} =-\mathbf{n}_{2} $ аналогічним чином отримуємо
$N_{ + \, -}  = N_{ - \, + } = 0$. У подальшому тривіальні множники, пов'язані з модулем проекції спіну ${\frac{{\hbar}} {2}}$, будемо опускати, тобто в якості операторів спіну будемо вибирати матриці Паулі $\sigma_{i}$. Покажемо, що природа квантово-механічного індетермінізму, з точки зору результатів
вимірювання фізичних величин, не зводиться просто до нестачі інформації як у звичайному статистичному описі за допомогою функцій розподілу значень тих чи інших величин.

Припустимо, що ми можемо розглянути невизначеність у вимірюванні двох проекцій спіну, при фіксованій третій, як результат того, що вектор спіну є просто випадковим вектором:

$$
\mathbf{S} = s^{i}\sigma_{i}\,,\quad    s^{i} = \pm 1\,.
$$

Введемо сумісну функцію розподілу

$$
P_{\,3;3} \,\left( {\,\mathbf{n}_{1} \cdot \mathbf{S}_{1}
,\mathbf{m}_{1} \cdot \mathbf{S}_{1} ,\mathbf{l}_{1} \cdot
\mathbf{S}_{1} ;\mathbf{n}_{2} \cdot \mathbf{S}_{2}
,\mathbf{m}_{2} \cdot \mathbf{S}_{2} ,\mathbf{l}_{2} \cdot
\mathbf{S}_{2}}  \right)
$$

величин проекцій спінів частинок у синглетному стані, яка відображує недостаток інформації про точні значення проекцій. Тут одиничні вектора визначають напрямки проекцій.

Часткові функції розподілу можуть бути отримані стандартним чином сумою по іншим змінним, наприклад:

\begin{align}
P_{\,2;2} \,\left( {\,\mathbf{n}_{1} \cdot \mathbf{S}_{1} ,
 \mathbf{m}_{1} \cdot \mathbf{S}_{1} ;
 \mathbf{n}_{2} \cdot \mathbf{S}_{2} ,
 \mathbf{m}_{2} \cdot \mathbf{S}_{2}}  \right)
 =&\notag\\
P_{\,3;3} \,\left( {\,\mathbf{n}_{1} \cdot \mathbf{S}_{1}
,\mathbf{m}_{1} \cdot \mathbf{S}_{1} , + 1;\mathbf{n}_{2} \cdot
\mathbf{S}_{2} ,\mathbf{m}_{2} \cdot \mathbf{S}_{2} , + 1} \right)
+ \notag\\ \,\,\,\,\,\,\,P_{\,3;3} \,\left( {\,\mathbf{n}_{1}
\cdot \mathbf{S}_{1} ,\mathbf{m}_{1} \cdot \mathbf{S}_{1} , -
1;\mathbf{n}_{2} \cdot \mathbf{S}_{2} ,\mathbf{m}_{2} \cdot
\mathbf{S}_{2} , + 1} \right) +& \notag\\ P_{\,3;3} \,\left(
{\,\mathbf{n}_{1} \cdot \mathbf{S}_{1} ,\mathbf{m}_{1} \cdot
\mathbf{S}_{1} , + 1;\mathbf{n}_{2} \cdot \mathbf{S}_{2}
,\mathbf{m}_{2} \cdot \mathbf{S}_{2} , - 1} \right) + \notag\\
P_{\,3;3} \,\left( {\,\mathbf{n}_{1} \cdot \mathbf{S}_{1}
,\mathbf{m}_{1} \cdot \mathbf{S}_{1} , - 1;\mathbf{n}_{2} \cdot
\mathbf{S}_{2} ,\mathbf{m}_{2} \cdot \mathbf{S}_{2} , - 1}
\right)\,.
 \end{align}

У подальшому для простоти ми розглянемо ситуацію з функціями
$P_{\,2;1} \,\left( {\,\mathbf{n}_{1} \cdot \mathbf{S}_{1}
,\mathbf{m}_{1} \cdot \mathbf{S}_{1} ;\mathbf{n}_{2} \cdot
\mathbf{S}_{2}} \right)$ і $P_{\,1;1} \,\left( {\,\mathbf{n}_{1}
\cdot \mathbf{S}_{1} ;\mathbf{n}_{2} \cdot \mathbf{S}_{2}}
\right)$. 

За елементарними міркуваннями теорії ймовірності,
отримаємо:
\begin{align}
\label{eq9/14} P_{\,2;\,0} \,\left(\mathbf{n}
\cdot\mathbf{S}_{1}=\,+ 1,
 \mathbf{m} \cdot \mathbf{S}_{1} = - 1 \right)
 =& \notag\\
 P_{\,1;1}
 \,\left(
\mathbf{n} \cdot \mathbf{S}_{1} = + 1; \mathbf{m} \cdot
\mathbf{S}_{2} = + 1 \right) =& \notag\\
 P_{\,3;0}
 \,\left(\mathbf{n} \cdot \mathbf{S}_{1}=
  + 1,\mathbf{m} \cdot \mathbf{S}_{1} = - 1,
 \mathbf{l} \cdot \mathbf{S}_{1} = + 1 \right) +& \notag\\
 P_{\,3;0} \,\left(\mathbf{n}
\cdot \mathbf{S}_{1} = + 1,\mathbf{m} \cdot \mathbf{S}_{1} = -
1,\mathbf{l} \cdot \mathbf{S}_{1} = - 1 \right) \le\notag
\\
 P_{\,2;\,0} \,\left( \,\mathbf{n} \cdot \mathbf{S}_{1}= +
 1,
 \mathbf{l} \cdot \mathbf{S}_{1} = + 1 \right) +&\notag\\
 P_{\,2;\,0} \,\left(\mathbf{m} \cdot \mathbf{S}_{1} = - 1,
 \mathbf{l} \cdot
\mathbf{S}_{1} = - 1 \right) =&\notag\\
 P_{\,1;1} \,\left(\mathbf{n} \cdot \mathbf{S}_{1}= +
 1,
 \mathbf{l} \cdot \mathbf{S}_{2} = - 1 \right) +&\notag\\
 P_{\,1;1} \,\left(\mathbf{m} \cdot \mathbf{S}_{1} = - 1,\mathbf{l} \cdot
\mathbf{S}_{2} = + 1 \right)\,.&
\end{align}

Тут ми скористалися тим, синглетним характером стана і тому що спін у данному випадку є інтегралом руху. Це приводить до того, що подія $\mathbf{n} \cdot \mathbf{S}_{1} = a$ обумовлює, що $\mathbf{n} \cdot \mathbf{S}_{2} = - \,a$ і навпаки.

Нехай вектори $\mathbf{n}_{1} ,\mathbf{m}_{1} ,\mathbf{l}_{1} $
компланарні, тоді: 

$$\mathbf{n} \cdot \mathbf{m} = \cos \varphi
\,,\mathbf{n}~\cdot~\mathbf{l} = \cos \theta\,. 
$$ 

Виходячи з (\ref{eq5/14}) і (\ref{eq9/14}) отримаємо нерівність:

\begin{equation}
\label{eq10/14} \sin ^{2}{\frac{{\varphi}} {2}} \le \sin
^{2}{\frac{{\pi - \theta}} {2}} + \sin ^{2}{\frac{{\pi - \theta -
\varphi}} {2}} = \cos ^{2}{\frac{{\theta }}{2}} + \cos
^{2}{\frac{{\theta + \varphi}} {2}}
\end{equation}

або

\begin{equation}
\label{eq11/14} - 1 \le \cos \theta + \cos \,(\theta + \varphi ) + \cos \,\varphi \,,
\end{equation}

яка порушується, наприклад, при $\theta\! = \!\varphi =\!
{\frac{{3\,\pi }}{{4}}}$. Це свідчить, що припущення про те, що $\mathbf{n} \cdot \mathbf{S}_{1} = a$ еквівалентно $\mathbf{n} \cdot \mathbf{S}_{2} = - \,a$ у розумінні еквівалентності

$$P_{\,2;\,0} \,\left( {\,\mathbf{n} \cdot \mathbf{S}_{1} = +
1,\mathbf{m} \cdot \mathbf{S}_{1} = - 1} \right)\,\,\,{\rm
i}\,\,\,P_{\,1;1} \,\left( {\,\mathbf{n} \cdot \mathbf{S}_{1} = +
1;\mathbf{m} \cdot \mathbf{S}_{2} = + 1} \right)
$$

не вірне. Завдяки тому, що різні компоненти спіну не комутують, для них функція розподілу $P_{\,2;\,0} \,\left( {\,\mathbf{n} \cdot \mathbf{S}_{1} ,\mathbf{m} \cdot \mathbf{S}_{1}}  \right)$ не має сенсу, що і проявляється у порушенні нерівностей типу (\ref{eq11/14}), які називаються нерівностями Белла. Порушення нерівності \eqref{eq11/14} тепер може бути перевірено експериментальним чином. Експерименти підтверджують справедливість висновків квантової механіки про відсутність станів із заданими значеннями не комутуючих спостережувальних величин.

Аналогічним чином можна отримати нерівності на середні типу (\ref{eq8/14}). Розглянемо одну з них. Для кореляційної функції проекцій спінів двох частинок, які приготували у синглетному стані, очевидно за означенням можна записати:

\begin{equation}
\label{eq1411} {\left\langle {S_{{\rm \mathbf{n}}_{1} ,1} S_{{\rm
\mathbf{n}}_{2} ,2}} \right\rangle}  = {\sum\limits_{S_{1} ,S_{2}
= \pm 1} {S_{1} S_{2}\, P_{\,1;1} (S_{{\rm \mathbf{n}}_{1} ,1} =
S_{1} ;S_{{\rm \mathbf{n}}_{2} ,2} = S_{2} )}}\,.
