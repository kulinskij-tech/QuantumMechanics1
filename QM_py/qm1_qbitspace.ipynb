Найпростійшою квантовою системою є система із 2-вимірним гільбертовим простором векторів
станів $\mathbb{C}^2$. Фізичною системою такого типу є внутрішній момент (спін) електрона або поляризаційні ступені вільності фотона. Дивовижно, що всі суттєві риси некласичної поведінки квантових систем можуть бути продемонстровані на цьому простому прикладі. Така система є нетривіальным об'єктом - елементарною коміркою квантової інформації і називається $q$-бітом по аналогії з класичним елементом - бітом. Таким чином існують реальні фізичні системи які є прикладами цього елементарного об'єкту квантової інформатики.

Нормований вектор стану q-біта має вигляд суперпозиції двох ортогональних базисних станів:<br>

$$
\ket{\psi} = c_{1}\,\ket{1}+c_{2}\,\ket{2}\,,\quad 
c_{i}\in \mathbb{C}\,, \\
\langle 1\vert 2\rangle = 0\,,\\ 
||\psi||^2=\braket{\psi}{\psi} = \vert c_{1}\vert^2+\vert c_{2}\vert^2=1
$$

тобто це просто вектор-стовпець:

$$
\ket{\psi} = 
\begin{vmatrix}
  c_{1} \\
  c_{2}
\end{vmatrix}
$$

спряжений вектор-строка:

$$
\bra{\psi} = \left(\ket{\psi}\right)^{\dagger} = (c^{*}_{1},c^{*}_{2})
$$

В якості базисних станів можуть виступати стани лінійної поляризації ЕМХ (фотону) $\left|\,x\,\right\rangle, \left|\,y\,\right\rangle$. З них можна побудувати стани довільної еліптичної поляризації, зокрема <a id='circ'>кругової (циркулярної)</a>

$$
\vert\, \pm\, \rangle =\frac{\vert\, x\, \rangle \pm i\,\vert\, y\, \rangle}{\sqrt{2}}\,,\\ 
\vert\, x\, \rangle =\frac{\vert\, +\, \rangle + \,\vert\,-\, \rangle}{\sqrt{2}}\,,\quad
\vert\, y\, \rangle =\frac{\vert\, +\, \rangle - \,\vert\,-\, \rangle}{i\,\sqrt{2}}
$$

<img src="./figs/fig_plrz.gif" alt="Image" width="600">

Стани проекції спіна електрона на певну вісь $\ket{\uparrow},\ket{\downarrow}$ також є базисними станами для простору спінових станів такого q-bit. Вектори простору $\mathbb{C}^2$ відображуються на проміні, які знаходяться у 1-1 віповідності із розширеною множиною комплексних чисел:

$$
\left|\,\psi\,\right\rangle =c_{\uparrow}\,\left|\,\uparrow\,\right\rangle+c_{\downarrow}\,\left|\,\downarrow\,\right\rangle \to z = c_{\uparrow}/c_{\downarrow}\in \mathbb{C}+\infty
$$

це і є сфера Рімана, оскільки стереографічна проекція отожює точки цієй сфери із множиною $\mathbb{C}+\infty$.

Якщо скористатись <a id='riemanspherparam'>наступною параметризацією точок сфери Рімана</a> $z = {\rm ctg}\frac{\theta}{2}\,e^{-i\,\varphi}$, то можна вважати, що
$$
\vert \psi \rangle = \vert \theta, \varphi \rangle = \cos{\frac{\theta}{2}}\,\vert\uparrow\rangle +e^{i\,\varphi}\,\sin{\frac{\theta}{2}}\,\vert\downarrow\rangle
$$
Ортопроектор на такий стан:
\begin{equation}\label{eq:orthoprj}
\hat{P} = \vert \theta, \varphi \rangle \langle \theta, \varphi \vert =
\begin{pmatrix}
 \cos^2{\frac{\theta}{2}} & \frac{1}{2}\,\sin{\theta}\,e^{-i\,\varphi}\\
 \frac{1}{2}\,\sin{\theta}\,e^{i\,\varphi}&\sin^2{\frac{\theta}{2}}
\end{pmatrix} = \frac{1}{2}\left(I+\hat{\mathbf{n}}\cdot\hat{\boldsymbol{\sigma}}\right)\\
\boldsymbol{\sigma} = \{\sigma_{1},\sigma_{2},\sigma_{3}\}\\
\end{equation}
Тут $\hat{\mathbf{n}}$ є сукупністю діагональних матриць (матричнозначний вектор) з елементами, які дорівнюють  відповідним проекціям вектору $\mathbf{n}= (\sin{\theta}\cos{\varphi},\sin{\theta}\sin{\varphi},\cos{\theta})$:

$$
\hat{\mathbf{n}} = \left(
\begin{pmatrix}
 n_1 &  0\\
  0 &  n_1
\end{pmatrix}, \begin{pmatrix}
 n_2 &  0\\
  0 &  n_2
\end{pmatrix},\begin{pmatrix}
 n_3 &  0\\
  0 &  n_3
\end{pmatrix}
\right)
$$
Матриці
$$
\sigma_{1} = \begin{pmatrix}
 0 &  1\\
  1 &  0
\end{pmatrix}\,,\quad \sigma_{2}=
\begin{pmatrix}
 0 &  -i\\

\end{pmatrix}\,, \quad \sigma_{3} = 
\begin{pmatrix}

\end{pmatrix}
$$

$$
\mathbf{n}\cdot \hat{\mathbf{S}} = \frac{\hbar}{2}\hat{\mathbf{n}}\cdot \hat{\boldsymbol{\sigma}}
$$
(множник $\hbar/2$ тут знову зображує одиничну матрицю з відповідним матричним елементом).
Таким чином із станом q-bit можна повязати напрямок заданий 3-вектором $\mathbf{n}\in \mathbb{R}^3$:
$$
\vert \theta, \varphi \rangle \leftrightarrow \vert \mathbf{n}\rangle 
$$
Це прояв ___векторного характеру___ __оператора спіна__ і того що існують стани із певним значенням проекції спіна. Власне саме вони і вимірюються завдяки взаємодії спінового магнітного моменту у зовнішньому магнітному полі яке і задає напрям проектування (<a href='https://en.wikipedia.org/wiki/Stern–Gerlach_experiment'>експеримент Штерна-Герлаха</a>). Зазначимо, тим не менше, що спін не є вектором у звичайному класичному сенсі навять якщо припустити що він ж випадковим вектором. Розглянемо тепер побудову унітарно матриці яка відображає базис $\vert \uparrow\rangle,\vert\downarrow \rangle$ на інший, перший вектор якого будемо визначати як $\vert \mathbf{n} \rangle$, тоді другий вектор цього базису можна визначити як:
$$
\vert \mathbf{n}\rangle^{\perp} = \vert - \mathbf{n}\rangle = \sin{\frac{\theta}{2}}\,\vert \uparrow \rangle -e^{-i\,\varphi}\,\cos{\frac{\theta}{2}}\,\vert \downarrow \rangle\,\Rightarrow\, \hat{P}_{\perp} = \frac{1}{2}\left(I-\mathbf{n}\cdot\boldsymbol{\sigma}\right)
$$
Унітарне перетворення зробимо у такі кроки: 
1. зміна фази вектора $\left|\,\uparrow\right\rangle$ відносно фази $\left|\,\downarrow\right\rangle$. 
1. обертання на кут $\theta$ у (2-вимірній) дійсній площині визначеній векторами $\left|\,\uparrow\right\rangle, \left|\,\downarrow\right\rangle$
1. зміна відносної фази образів
1. зміна загальної фази 

В результаті маємо 4-параметричне унітарне перетворення:
$$
\hat T(\theta,\psi,\phi,\alpha)=\exp(i\alpha)\left(\begin{array}{cc}
  \exp\left(-i\frac{\phi}{2}\right) & 0 \\
  0 & \exp\left(i\frac{\phi}{2}\right)
\end{array}\right)\,
\left(\begin{array}{cc}
  \cos{\theta} & -\sin{\theta} \\
  \sin{\theta} & \cos{\theta}
\end{array}\right)
\,\left(\begin{array}{cc}
  \exp\left(-i\frac{\psi}{2}\right) & 0 \\
  0 & \exp\left(i\frac{\psi}{2}\right)
\end{array}\right)\\
=\exp(i\alpha) \left(\begin{array}{cc}
  \exp\left(-i\frac{\psi+\phi}{2}\right)\cos{\theta} &
  -\exp\left(i\frac{\psi-\phi}{2}\right)\sin{\theta} \\
  \exp\left(-i\frac{\psi-\phi}{2}\right)\sin{\theta} &
  \exp\left(i\frac{\psi+\phi}{2}\right)\cos{\theta}
\end{array}\right)
$$

Зрозуміло, що ймовірності є квадратами модулей проекцій стана на базисні стани. Тому з т.з. лінійної алгебри питання зводиться до зміни координат вектора $\vert\psi \rangle$ при переході від базиса $\{\vert\, x\, \rangle\,,\vert\, y\, \rangle\}$ до $\{\vert\, +\, \rangle\,,\vert\, -\, \rangle\}$. Запишемо матрицю такого переходу:
$$
\hat{U}\left(\{\vert x,y\rangle\}\to \{\vert \pm \rangle\}\right) =\begin{pmatrix}
 \left\langle\,+\,\right|x\rangle &  \left\langle\,+\,\right|y\rangle\\
  \left\langle\,-\,\right|x\rangle &  \left\langle\,-\,\right|y\rangle
\end{pmatrix} = 
\frac{1}{\sqrt{2}}\begin{pmatrix}
 1 &  -i\\
 1 &  i
\end{pmatrix}
$$

```python
e1=Matrix([1,0])
e2=Matrix([0,1])
circ10=Matrix([1,I])
circ20=Matrix([1,-I])
circ1=circ10/circ10.norm()#normalization
circ2=circ20/circ20.norm()#normalization
umtrx=Matrix([[(circ1.H*e1)[0],(circ1.H*e2)[0]],[(circ2.H*e1)[0],(circ2.H*e2)[0]]])
umtrx
```

$$
\hat{U}^{-1} = \hat{U}^{\dagger} =\frac{1}{2}\begin{pmatrix}

\end{pmatrix}
$$
