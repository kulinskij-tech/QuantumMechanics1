Як і в класичній в квантовій механиці є тісний звязок між існуванням інтегралів руху і симетрійними властивостями системи. При вирішені проблем квантової теорії ідеї та методи теорії симметрії відіграють навіть більшу роль, оскільки кратності виродження власних значень гамільтоніана визначаються саме симметріями. Кожному перетворенню, яке не змінює динаміку системи відповідає певний оператора - інтеграл руху.

Розглянемо трансляційну інваріантність динаміки системи, тобто симетрію відносно паралельних зсувів:

$$\label{par}
\mathbf{r}'=T_{\mathbf{a}}\, \mathbf{r}=\mathbf{r}+\mathbf{a}\,,
$$

або еквівалентно, при зміщені вимірювачана вектор $- \mathbf{a}$. Така симетрія призводить до інтеграла руху - імпульса. 
Геометричне перетворення $T_{\mathbf{a}}$, яке залежить від набору параметрів $\mathbf{a}$, (трансляція, обертання та інш.) породжає оператор унітарного перетворення у просторі хвильових функцій. Значення нової хвильової функції $\tilde\psi(\mathbf{r})$ у точці $\mathbf{r}$ співпадає зі значенням старої функції $\psi$ в точці прообразі $T_{\mathbf{a}}^{-1}\mathbf{r}$:
<!--  -->
\begin{equation}
\label{trans} \tilde{\psi}(\mathbf{r})=
\left(\hat{U}_{T_{\mathbf{a}}}\psi\right)(\mathbf{r})
=\psi(T_{\mathbf{a}}^{-1}\mathbf{r}) = \psi(\mathbf{r}-\mathbf{a})\,,
\end{equation}
<!--  -->
або в термінах векторів:
<!--  -->
$$
\braket{\mathbf{r}}{\psi} 
= \braket{T_{\mathbf{a}}^{-1}\mathbf{r}'}{\psi} =
\braket{\mathbf{r}'}{\hat{U}_{T_{\mathbf{a}}}\psi} = 
\braket{\mathbf{r}'}{\tilde{\psi}}\,.
$$
<!--  -->
Остання рівність співпадає з \eqref{trans}. Якщо перетворення залежить від параметра гладким чином, то:
<!--  -->
\begin{equation}
\label{mom1} \hat U\left[\hat T(\mathbf{a})\right]= \exp\left(
\frac{\mathbf{a}\cdot\hat{\mathbf{P}}}{i\,\hbar}
\right)=1+\frac{\mathbf{a}\cdot\hat{\mathbf{P}}}{i\,\hbar}+o(\mathbf{a})\,,\,\,\,\,
\hat{\mathbf{P}} = -i\,\hbar\,\nabla_{\mathbf{r}}
\end{equation}
<!--  -->
Оператор $\hat{\mathbf{P}}$ називається інфінітезимальним генератором унітарного перетворення і є ермітовим (і навіть  самосопряженим) оператором. В даному випадку трансяційної симетрії він співпадає з оператором  імпульсу системи. Представлення \eqref{mom1} має такой же вигляд як і оператор еволюції 
<!--  -->
$$
\hat{S}(t) = e^{\frac{t}{i\,\hbar}\,\hat{H}}
$$
<!--  -->
Тому гамільтоніан можна назвати інфінітезимальним генератором зсувів з часом. 
Множина генераторів групи перетворень є вочевидь (дійсним) векторним простором з додатковою операцією композиції операторів (вона не обовязково є комутативною). Така струкутра називається __алгеброю__. Приклади алгебр: (комутативна) алгебра дійсних чи комплекснх функцій на множині, алгебра матриць лінійних перетворень векторного простору, алгебра 3-векторів із векторним добутком. Остання алгебра є прикладом алгебри Лі, де ''добуток'' є коосиметричною операцією: $X*Y = [X,Y] = -Y*X$.

Кожна група неперервних перетворень (група Лі) таким чином породжує алгебру інфінітезимальних генераторів, яку можна вважати алгеброю Лі якщо добуток представляти у вигляді комутатора операторів.

Взагалі інвариантність динаміки системи відносно певного перетворення симметрії означає, що якщо $\psi$ є решенням рівняння Шредінгера, то $\hat U\psi$ також є його рішенням:
<!--  -->
\begin{equation}
\label{u} i\,\hbar\,\frac{\partial \psi}{\partial\, t}=\hat H\psi
\rightarrow i\,\hbar\,\frac{\partial }{\partial\, t}\hat U\psi=\hat
H\hat U\psi
\end{equation}
<!--  -->
Застосовуючи оператор $\hat U$ (від часу не залежить) до обох частин рівняння Шредінгера, і порівнюючи результат з \eqref{u}, маємо:
<!--  -->
\begin{equation}
\left[\hat U,\hat H\right]=0
\end{equation}
<!--  -->
Використовуючи інфінітезимальне представлення \eqref{mom1}, неважко переконатися що:
<!--  -->
\begin{equation}
\label{momint} \left[\hat X,\hat H\right]=0
\end{equation}
<!--  -->
тобто інфінітезимальний генератор є __інтегралом руху__ якщо динаміка системи інваріантна відностно перетворення яке ним породжується. Цей результат є квантовим аналогом відповідного твердження у класичній механиці. 

Осколькі оператори імпульсу иі координати спряжені, то розглянемо з цієї точки зору оператор координати. Якщо перевести результати [про перетворення хвильової функції при зміні системи відліку](https://colab.research.google.com/github/kulinskij-tech/QuantumMechanics1/blob/main/QM_py/qm1_scheiz_theor.ipynb#galiley) на мову генератора симетрії вдносно перетворення/переходу до ІСВ, то отримуємо, що перехід у миттєву супутну ІСВ описується унітарним оператором:
<!--  -->
\begin{equation}\label{galileoboost}
  \hat{U}_{\mathbf{v}}\,\ket{\psi} = \exp{\left(\frac{m\,\hat{\mathbf{r}}\cdot \mathbf{v}}{i\,\hbar}\right)}\,\ket{\psi}
\end{equation}
<!--  -->
або
<!--  -->
$$
\hat{\mathbf{p}}\to \hat{U}_{\mathbf{v}}\,\hat{\mathbf{p}}\,\hat{U}^{-1}_{\mathbf{v}}
$$
<!--  -->
\begin{multline}
\av{\mathbf{r}}{\hat{U}_{\mathbf{v}}\,\hat{\mathbf{p}}\,
\hat{U}^{-1}_{\mathbf{v}}}{\psi}=
\int\,d\mathbf{r}'\,
\av{\mathbf{r}}{\hat{U}_{\mathbf{v}}\,\hat{\mathbf{p}}}{\vphantom{\hat{U}}\mathbf{r}'}
\av{\mathbf{r}'\vphantom{\hat{U}}}{\hat{U}^{-1}_{\mathbf{v}}}{\psi}
=\\
e^{-\frac{m\,\mathbf{v}\cdot \mathbf{r}}{i\,\hbar}}\,
\int\,d\mathbf{r}'\,\av{\mathbf{r}}{\hat{\mathbf{p}}}{\mathbf{r}'}\,
e^{\frac{m\,\mathbf{v}\cdot \mathbf{r}'}{i\,\hbar}}
\av{\mathbf{r}'}{\psi} =
e^{-\frac{m\,\mathbf{v}\cdot \mathbf{r}}{i\,\hbar}}\,\int\,d\mathbf{r}'\,i\,\hbar\,\frac{\partial\,\delta(\mathbf{r}-\mathbf{r}')}{\partial\,\mathbf{r}'}\,
e^{\frac{m\,\mathbf{v}\cdot \mathbf{r}'}{i\,\hbar}}
\braket{\mathbf{r}'}{\psi}=\\
e^{-\frac{m\,\mathbf{v}\cdot \mathbf{r}}{i\,\hbar}}\,\left(-i\,\hbar\,\frac{\partial}{\partial\,\mathbf{r}}\right)\,
\left(e^{\frac{m\,\mathbf{v}\cdot \mathbf{r}}{i\,\hbar}}
\psi(\mathbf{r})\right) = \left(\mathbf{\hat{p}} - m\,\mathbf{v}\right)\,\psi(\mathbf{r})
\end{multline}
<!--  -->
Т.ч. оператор координати является інфінітезимальным генератором перехода у супутню ІСВ, що еквівалентно зсуву у імпульсному просторі. Оскільки галілеєвська інваріантність є базовим принципом нерелятивістської фізики то оператор координати завжди представляє спостережну величину. На відмну від цього спостережність імпульса визначається існуванням (квази)вільних станів для яких частинка може бути зараєстрована на "нескінченності" завдяки трансляційній інваріантності таких станів.  

### Рух частинки у періодичному полі. Зонна структура.
Природнім застосуванням зв'язку симетрії із структурою хвильових функцій стаціонарних станів є випадок руху у полі періодичного потенціала. В якості реального приклада можна навести розповсюдження електронного збудженя в ідеальному кристалі - базової моделі фізики твердого тіла.

Тут ми розглянемо одновимірний варіант цієї задачі. Система задається гамільтоніаном:
<!--  -->
\begin{equation}\label{hper}
  \hat{H} = \frac{\hat{\mathbf{p}}^{\,2}}{2m} + \hat{V}(x)\,,
\end{equation}
<!--  -->
де потенціал має властивість періодичності із періодом гратки $a$:
<!--  -->
\begin{equation}\label{vper}
  V(x+a) = V(x)\,.
\end{equation}
<!--  -->
Т.ч. гамільтоніан комутує з операторами трансляції $\hat{T}_{na}$. Періодичність потенціала обумовлює властивість симетрії відносно зсувів на ціле кратне сталої
"гратки" $a$: якщо $\psi(x)$ - розв'язок стац. рівняння Шредінгера, то $\psi(x+a)$ - також розв'язок, який відповідає тому ж самому власному значенню гамільтоніана. Очевидно, що власними функціями операторів $\hat{T}_{a}$ так же як і у випадку довільних зсувів є плоскі хвилі $\psi_{k} = e^{ikx}$ і звичайно періодичні функції $u(x) = u(x+a)$
(у випадку інваріантності відносно довільних трансляций $u(x) = const$). Тому власна функція оператора $\hat{T}_{a}$ має вигляд:
<!--  -->
\begin{equation}\label{blokh}
  \psi(x) = e^{ik\,x}\,u(x)\,.
\end{equation}
<!--  -->
Оскільки оператори $\hat{T}_{a}$ и $\hat H$ комутують, у них є спільний набір власних функцій. Т.ч. власні функції гамільтоніана з пеперіодичним потенціалом маютm вигляд
\eqref{blokh}. Цей факт називається __теоремою Блоха__. Параметр $k$, який визначає поведінку власних функцій оператора \eqref{hper} при трансляціях на величину періоду $a$ називається __квазіімпульсом__. Підставляючи \eqref{blokh} у стаціонарне рівняння Шредінгера, отримаємо:
<!--  -->
\begin{equation}\label{schb}
  \left[\frac{\hbar^{\,2}}{2m}\left(\,\frac{d}{d x}+ik\,\right)^{\,2} + E -
  V(x)\right]\,u(x) = 0\,,
\end{equation}
<!--  -->
З огляду на періодичність потенціалу рівняння \eqref{schb} можна розв'язувати в інтервалі $0~\le~x~\le~a$. Крім того, як витікає з \eqref{blokh} значення $k$ і $k+K$, где $e^{iKa} = 1$, еквівалентні, тобто власні функції для таких значень параметра однакові. Звідси витікає висновок що $E(k) = E(k+K)$. Нехай $u_{1}$ и $u_{2}$  - розв'язкі рівняння
\eqref{schb}. Тоді
<!--  -->
\begin{equation}\label{g1}
  \psi(x) = C_1 u_1(x;E)+C_2 u_2(x;E)\,, \quad 0~\le~x~\le~a\,.
\end{equation}
<!--  -->
В інтервалі $a~\le~x~\le~2a$ згідно з \eqref{blokh}:
<!--  -->
\begin{equation}\label{blokh1}
  \psi(x) = e^{ika}\,\left(\, C_1 u_1(x-a;E)+C_2
  u_2(x-a;E)\,\right)\,,\quad a~\le~x~\le~2a\,.
\end{equation}
<!--  -->
Запишемо граничні умови непрервності хвильової функції та її похідної у точці $x=a$:
<!--  -->
\begin{eqnarray}\label{g2}
  C_1 u_1(a;E)+C_2 u_2(a;E) &=& e^{ika} \left(\, C_1 u_1(0;E)+C_2
  u_2(0;E)\,\right)\\
    C_1 u_1'(a;E)+C_2 u_2'(a;E) &=& e^{ika} \left(\, C_1 u_1'(0;E)+C_2
  u_2'(0;E)\,\right)\,.
\end{eqnarray}
<!--  -->
Умова існування нетривіального розв'язку такої лінійної однорідної системи дає дисперсійний закон $E(k)$ (враховуємо сталость вронскіану $W = u_1\,u'_2-u_2\,u'_1$):
<!--  -->
\begin{equation}\label{ek}
  \cos{ka} = \frac{u_1(0;E)u_2'(a;E) + u_1(a;E)u_2'(0;E)
  - u_2(0;E)u_1'(a;E) - u_2(a;E)u_1'(0;E) }{2(u_1\,u_2'-u_1'\,u_2)}
\end{equation}
<!--  -->
Це рівняння має розв'язок тільки за умови
<!--  -->
\begin{equation}
  \left|\,\frac{u_1(0;E)u_2'(a;E) + u_1(a;E)u_2'(0;E)
  - u_2(0;E)u_1'(a;E) - u_2(a;E)u_1'(0;E)
  }{2(u_1\,u_2'-u_1'\,u_2)}\,\right|\le 1
\end{equation}

```python
k,x,c0,c1=symbols('k x c0,c1')
u1=Function('u1')(x)
u2=Function('u2')(x)
# 
a=expand(c0*u1.subs(x,1)+c1*u2.subs(x,1)-exp(I*k)*(c0*u1.subs(x,0)+c1*u2.subs(x,0))).coeff(c0,1)
#
a
```

```python
b=expand(c0*u1.subs(x,1)+c1*u2.subs(x,1)-
         exp(I*k)*(c0*u1.subs(x,0)+c1*u2.subs(x,0))).coeff(c1,1)
b
```

```python
c=expand(c0*diff(u1,x).subs(x,1)+c1*diff(u2,x).subs(x,1)-
         exp(I*k)*(c0*diff(u1,x).subs(x,0)+c1*diff(u2,x).subs(x,0))).coeff(c0,1)
c
```

```python
d=expand(c0*diff(u1,x).subs(x,1)+c1*diff(u2,x).subs(x,1)-
         exp(I*k)*(c0*diff(u1,x).subs(x,0)+c1*diff(u2,x).subs(x,0))).coeff(c1,1)
d
```

```python
(Matrix([[a,b],[c,d]]).det())
```

В такому випадку спектр енергії складається взагалі кажучи з інтервалів - енергетичних зон $E(k)$. Тобто рух частинки у полі періодичного потенціалу реализує проміжний випадок між дискретними станами частинки у полі потенціальної ями (дискретний спектр + неперервний спектр при $E>0$) і вільною частинкою (неперервний спектр), який характеризуется __зонною структурою спектра енергії__. Конкретний приклад обчислення зонного спектру демонструється на прикладі [моделі Кроніга-Пенні](https://colab.research.google.com/github/kulinskij-tech/QuantumMechanics1/blob/main/QM_py/qm1_stac_zad.ipynb#kronigpenni).

### Момент імпульсу. Алгебра кутового момента
Симетрія відносно обертань пов'язана з ізотропністю простору і, як відомо з механіки, із законом збереження момента імпульса. Це відіграє важливу роль у задачах, де зовнішне поле є сферично симетричним. Почнемо з обертань у 2-вимірному просторі. Нагадаємо, що поворот у площині $(x,y)$ задається матрицею:

\begin{equation}\label{rotz_2mtrx}
  R_{z}(\varphi) = \begin{pmatrix}
            \cos{\varphi} & -\sin{\varphi} \\
             \sin{\varphi} & \cos{\varphi} \\
          \end{pmatrix} \Rightarrow \frac{d}{d\,\varphi}
          R_{z}(\varphi) =  \begin{pmatrix}
            -\sin{\varphi} & -\cos{\varphi} \\
             \cos{\varphi} & -\sin{\varphi} \\
          \end{pmatrix} =
         \begin{pmatrix}
   0 & -1 \\
    1 & 0 \\
     \end{pmatrix}\, R_{z}(\varphi)
\end{equation}

що можна представити у вигляді матричної експоненти:

$$ 
R_{z}(\varphi) = 
  \exp{\left(\varphi\,\begin{pmatrix}
   0 & -1 \\
    1 & 0 \\
     \end{pmatrix}\right)}
$$

Аналогічно, матриця повороту навколо вісі $z$ у 3-вимірному просторі має вигляд:

\begin{equation}\label{rotz_mtrz}
  R_{z}(\varphi) = \begin{pmatrix}
            \cos{\varphi} & -\sin{\varphi} & 0 \\
             \sin{\varphi} & \cos{\varphi} & 0 \\
            0 & 0 & 1 \\
          \end{pmatrix} = \exp{\left(\varphi\,
  \begin{pmatrix}
   0 & -1&0 \\
    1 & 0 &0\\
    0&0&0
     \end{pmatrix}\right)}
\end{equation}

\begin{equation}\label{rotz_mtrx}
  R_{x}(\varphi) = \begin{pmatrix}
  1 & 0 & 0 \\
            0&\cos{\varphi} & -\sin{\varphi} \\
             0&\sin{\varphi} & \cos{\varphi}  \\
          \end{pmatrix} = \exp{\left(\varphi\,
  \begin{pmatrix}
   0& 0& 0\\
   0& 0 & -1\\
   0& 1 & 0\\
     \end{pmatrix}\right)}
     \end{equation}

\begin{equation}\label{rotz_mtry}
R_{y}(\varphi) = \begin{pmatrix}
              \cos{\varphi}&0 & \sin{\varphi} \\
              0 & 1 & 0 \\
             -\sin{\varphi}& 0 & \cos{\varphi}  \\
          \end{pmatrix} = \exp{i\left(\varphi\,
  \begin{pmatrix}
   0& 0& 1\\
   0& 0 & 0\\
   -1& 0 & 0\\
     \end{pmatrix}\right)}
\end{equation}
<!--  -->
Звідци отримаємо генератори обертань навколо відповідних вісей.

оператора момента импульсу у квантовій механіці, як інфінітезимального генератора поворотів у випадку обертальної симетрії.

Для цього розглянемо нескінченно мале обертання навколо вісі $\mathbf{n}$, на кут
$\varphi << 1$. Геометрично очевидно, що:
<!--  -->
\begin{equation}
\label{eq8/7} \mathbf{r}' = R(\mathbf{n},\varphi)\mathbf{r} =
\mathbf{r} + \varphi \,\mathbf{n}\times \mathbf{r} + o(\varphi)\,,
\end{equation}
<!--  -->
<a id='unitrot'>тому</a>
<!--  -->
\begin{equation} \label{momunit}
\begin{array}L
(U[R(\mathbf{n},\varphi)]\,\psi)(\mathbf{r}) = \psi (R^{-1}(\mathbf{n},\varphi)\mathbf{r}) = \psi (\mathbf{r} - \varphi \,\mathbf{n}\times \mathbf{r} + o(\varphi )) = \\
\left( 1 - \varphi(\mathbf{n}\times \mathbf{r}) \cdot
\nabla + o(\varphi) \right)\,\psi (\mathbf{r}) = \exp\left(
\varphi\frac{\mathbf{n} \cdot \mathbf{L}}{\rm i\,\hbar}
\right)\,\psi (\mathbf{r})\,,\\
\end{array}
\end{equation}
<!--  -->
де ермітовий оператор
<!--  -->
\begin{equation}
\label{eq10/7} \mathbf{L} = -i\,\hbar \,\mathbf{r}\times \nabla =
\mathbf{r}\times \mathbf{p}
\end{equation}
<!--  -->
$$
L_z = x\,p_y - y\,p_x = x\,\left(-i\,\hbar\,\frac{\partial}{\partial\,y}\right) - y\,\left(-i\,\hbar\,\frac{\partial}{\partial\,x}\right)\\
L_x(L_y\psi(\mathbf{r})) - L_y(L_x\psi(\mathbf{r})) = i\,\hbar L_{z} \psi(\mathbf{r}) \\
x\to y\to z\to x
$$
є інфінітезимальним генератором (він визначає швидкість зміни стана при повороті системи координат) поворотів і називається оператором обертального моменту імпульсу. Можна переконатися, що компоненти цього оператора задовольняють наступним комутаційним співвідношенням:
<!--  -->
\begin{equation}\label{eq11/7}
[\hat{L}_{i},\hat{L}_{j}]=i\,\hbar\,\epsilon_{ijk}\,\hat{L}_{k}
\end{equation}
<!--  -->
$$
\hat{\mathbf{L}}^2 = \hat{\mathbf{L}}\cdot \hat{\mathbf{L}} = L^2_{x}+L^2_{y}+L^2_{z}\\
[\hat{L}_{i},\hat{\mathbf{L}}^2]=0\,.
$$
<!--  -->
які <a href='#rotcomm'>аналогічні до отриманих вище для генераторів обертань</a>.
Таким чином, стан квантової системи може бути заданий власними значеннями операторів $L^{\,2}$ і однією з проекцій. 
Генератори утворюють лінійний простір, який називається алгеброю Лі відповідної групи неперервних перетворень.

Розглянемо питання про власні значення цих операторів. Введемо
наступні оператори:
<!--  -->
\begin{equation} \label{eq13/7} 
L_{\,\pm}  =\hat{L}_{\,x} \pm \,\,i\,L_{\,y}\,,\\
L_{x} = \frac{\hat{L}_{+}+\hat{L}_{-}}{2}\,,\,\,\,\,\,\,L_{y} = \frac{\hat{L}_{+}-\hat{L}_{-}}{2\,i}
\end{equation}
<!--  -->
Завдяки (\ref{eq11/7}) мають місце наступні співвідношення:
<!--  -->
\begin{align}\label{pm1}
\hat{L}_{\pm}^{\dagger} &= \hat{L}_{\mp}\notag\\
\hat{L}_{+}\hat{L}_{-} &= \hat{\mathbf{L}}^2-\hat{L}_{z}^2+\hbar \hat
L_{z}\\
[\hat{L}_{z},\hat{L}_{\pm}]&= \pm \hbar \hat{L}_z\notag\\
[\hat{L}_{\pm} ,\hat{\mathbf{L}}^2]&= 0\,\,.\notag
\end{align}
<!--  -->
Звідси випливає, що дія операторів (\ref{eq13/7}) залишає простір власних станів оператора $\hat{\mathbf{L}}^2$ інваріантним. Крім того, оператори (\ref{eq13/7}) переводять власний вектор оператора проекції моменту імпульсу знов у власний вектор, але із збільшеним
(+) або зменшеним (-) на $\hbar $ значенням проекції. Розглянемо, наприклад, оператор $\hat{L}_{+} $ (для простоти $\hbar =1)$.  Дійсно, будемо позначати власний вектор, операторів $\mathbf{L}^{\,2}$ та $L_{z}$, який відповідає власним значенням $\lambda $ і $\mu $ відповідно $\ket{\lambda,\mu}$, тоді:
<!--  -->
\begin{equation}
\label{eq14/7}\hat{L}_{\,z}\hat{L}_{\,\pm}  \ket{\lambda,\mu} = \left(\hat{L}_{\,\pm}\hat{L}_{\,z} \pm \,\hbar \,L_{\pm} \right)\ket{\lambda,\mu} = \left( \mu \pm \,\hbar \, \right)\,L_{\,\pm}  \ket{\lambda,\mu}\,\,.
\end{equation}
<!--  -->
Розглянемо норму вектора $\hat{L}_{+}\ket{\lambda, \mu}$: 
<!--  -->
$$
\vert\vert\hat{L}_{+}\ket{\lambda, \mu}\vert\vert^2 = \bra{\lambda, \mu}\hat{L}_{-} \,L_{+} \ket{\lambda, \mu} = \bra{\lambda, \mu}(\mathbf{L}^{\,2} -\hat{L}_{z}^{\,2} -\hat{L}_{z} )\ket{\lambda,\mu} \ge 0\,,
$$
<!--  -->
що дає нерівність:
\begin{equation}\label{e15/7}
\lambda - \mu (\mu + \hbar) \ge 0\,.
\end{equation}
<!--  -->
За результатом (\ref{eq14/7}) повторна дія оператора $L_{ +} $ на вектор $\ket{\lambda, \mu}  $ кінець кінцем приведе до того, що нерівність (\ref{e15/7}) порушиться, але це означає, що існує вектор з максимально можливим при даному $\lambda $ значенням $\mu$, яке ми позначимо як $\mu_{\max}  $, тобто $\mu_{\max}  (\mu_{\max}  + \hbar) = \lambda $. Аналогічні міркування для $L_{ -} $ приводять до $\mu_{\min}  (\mu_{\min} - \hbar) = \lambda $. З цих двох рівностей отримуємо:
$$
\mu_{\max}  (\mu_{\max}  + \hbar) - \mu_{\min} (\mu_{\min}  - \hbar) =
(\mu_{\max}  - \mu_{\min}  + \hbar)(\mu_{\max}  + \mu_{\min} ) = 0\,.
$$
Оскільки 
$$
\mu_{\max}  > \mu_{\min}  $$
то 
$$
\mu_{\max}  = - \mu_{\min}  = \hbar\,l\,\,,\,\,\lambda = \hbar^2 \,l\,(l + 1)
$$ 
де $2l$ - ціле число, тому що з стану $\mu_{\min}$ стан з $\mu_{\max}$ отримано в
результаті $2l$ - кратного застосування підвищуючого оператора. Використовуючи явний вираз для оператора проекції моменту імпульсу:
<!--  -->
\begin{equation}\label{eq16/7} 
L_{\,z} = - \,i\hbar \,\frac{\partial}{\partial \,\varphi}
\end{equation}
<!--  -->
і умову періодичності хвильової функції $\psi (\varphi ) = \psi (\varphi + 2\pi )$, отримаємо, що $l$ також повинно бути цілим, оскільки власна функція для (\ref{eq16/7}) має вигляд:
$$
\psi_{m} = C\exp (i\,m\,\varphi )\,.
$$
$$\{q,p_{q}\} = \frac{\partial q}{\partial q}\,\frac{\partial p_{q}}{\partial p_{q}} - \frac{\partial p_{q}}{\partial q}\,\frac{\partial q}{\partial p_{q}} =1 \leftrightarrow [q,p_q] = i\,\hbar $$
Таким чином, виходячи з комутаційних співвідношень між компонентами оператора момента імпульса, отримано:
<!--  -->
- оператор квадрата моменту імпульсу і проекції на певну вісь є одночасно спостережувальними величинами;
- власні значення оператора $\mathbf{L}^{\,2}$ дорівнюють $\hbar^{\,2}l\,(l + 1)$
- власні значення оператора проекції моменту імпульсу на певну вісь $\hbar \,m\,\,,\,\, - l \le m \le l\,\,,\,\,m \in Z$, тобто проекція моменту імпульсу на цю вісь є квантованою.

<!--  -->
Безпосереднім розрахунком можна переконатися, що у сферичних координатах:
<!--  -->
\begin{equation} \label{eq12/7} 
\mathbf{L}^{\,2} = - \hbar^{\,2}
\left[ \frac{1}{\sin \theta}\frac{\partial}{\partial \,\theta}\left(\sin\theta \frac{\partial}{\partial \,\theta} \right) + 
\frac{1}{\sin^{\,2}\theta}\frac{\partial^{\,2}}{\partial \,\varphi^{\,2}} \right]\,.
\end{equation}
<!--  -->
Відмітимо, що у такому представленні оператор залежить тільки від
кутових змінних.<br>

Власні функції оператора (\ref{eq12/7}) у сферичних координатах є сферичними функціями, які виражаються через т.з. <a href="https://dlmf.nist.gov/14.2">поліноми Лежандра</a>.<br>

$$\braket{\theta,\varphi}{l,m} = Y_{l,m}(\theta,\varphi) \sim P^{m}_{l}(\cos{\theta})\,e^{i\,m\,\varphi}$$

які, як витікає з попереднього задовільняють рівнянню (сингулярні при $x=\pm 1$ розвязки т.з. <a href="https://dlmf.nist.gov/14.2"> функції Лежандра 2-го роду</a> $Q^{m}_{n}(x)$ ми відкидаємо):

$$
z=\cos{\theta}\,\,\to\,\,\,\,\,\frac{d}{d\,z}\left((1-z^2)\,\frac{d\,P^{m}_{l}(z)}{d\,z}\right) +\left(l\,(l+1)-\frac{m^2}{1-z^2}\right)\,P^{m}_{l}(z) = 0
$$

Поліноми $P^{m}_{l}(z)\,, m\ne 0$ називаються приєднаними поліномами Лежандра оскільки вони виникають з <a href="https://dlmf.nist.gov/14.2">"базових" поліномів Лежандра $P_{l}(z) = P^{0}_{l}(z)$</a> дією операторів $L_{\pm}$. Поліноми $P_{l}(z)$ повязани з т.з. мультипольним розкладом кулонівського потенціалу (функції Гріна оператора Лапласа):

$$\frac{1}{\vert \mathbf{R} - \mathbf{r} \vert} = \frac{1}{R\,\sqrt{1-2\,z\,u+u^2}} = \frac{1}{R}\,\sum\limits_{l=0}^{\infty} P_{l}(z)\,u^l\\
z= \cos{\theta}\,,\,\,\,u = r/R \ll 1
$$
Використовуючи цю генеруючу функцію можна отримати рекурентні співвідошення:
$$
(1-2\,z\,u+u^2)\frac{d}{du}
\sum\limits_{l=0}^{\infty}\,P_{l}(z)\,u^{l} =  (1-2\,z\,u+u^2)
\sum\limits_{l=0}^{\infty}\,l\,P_{l}(z)\,u^{l-1} =
\frac{x-y}{\sqrt{1-2\,z\,u+u^2}}=\\
\sum\limits_{l=0} z\,P_{l}(z)\,u^{l} - \sum\limits_{l=0} P_{l-1}(z)\,u^{l} \Rightarrow \\
(2\,l+1)\,z\,P_{l}=(l+1)\,P_{l+1}+l\,P_{l-1}
$$

<!--  -->
\begin{equation}
\label{eq17/7} {H} = - \frac{\hbar ^{\,2}}{2\,m}\,\frac{1}{r^{\,2}}\,\frac{\partial}{\partial\,r}\,\left(r^{\,2}\,\frac{\partial}{\partial \,r} \right) + 
\frac{\mathbf{L}^{\,2}}{2\,m\,r^{\,2}} + V(\mathbf{r})\,.
\end{equation}
<!--  -->
У випадку центральної симетрії крім інтеграла руху момента імпульса є також інваріант, пов'язаний із дискретною симетрією - перетворення інверсії відносно центра координат (який співпадає із  силовим центром): ${\rm \mathbf{\mathbf{r}}}\,\, \to \, - {\rm \mathbf{\mathbf{r}}}$. Це означає, що якщо $\psi ({\rm \mathbf{\mathbf{r}}})$ є власною функцією, то $\psi ( - {\rm \mathbf{\mathbf{r}}}) = P\,[\psi ]({\rm \mathbf{\mathbf{r}}})$ також є такою. В силу того, що 
<!--  -->
$$P^{\,2} = P^{\dagger} P = 1$$ 
<!--  -->
власними значеннями цього оператора є $\pm 1$ тому власні функції такого оператора є або парними або не парними. Ця властивість є интегралом руху, тобто не змінюється впродовж еволюції і називається парністю стана. Можна показати, що парність стана з певним значенням $l$ співпадає з парністю $l$ і дорівнює $(-1)^l$.

Момент імпульсу системи, яка складається з підсистем $\Sigma_1, \Sigma_2, \ldots$ визначається за правилом суми моментів: якщо підсистема $\Sigma_1$ є у власному стані обертального моменту з $l_1$, і підсистема $\Sigma_2$ має стан обертального моменту з $l_2$, то комбінована система $\Sigma_{1}\otimes \Sigma_{2}$ перебуває у суперпозиційному стані оператору обертального моменту. 

В цю суперпозицію входять стани із наступними значеннями обертального квантового числа $|l_1-l_2|, |l_1-l_2|+1,\ldots, l_1+l_2$. Доведення цього факту будується на основі представлення базису простору $\Sigma_{1}\otimes \Sigma_{2}$  т.з. "сепарабельними" станами:

$$
\ket{l_1,m_1}\otimes \ket{l_2,m_2}
$$

$$L_{-} = (L^{(1)}_{-}\otimes I+I\otimes L^{(2)}_{-}) \ket{l_1,l_1}\otimes \ket{l_2,l_2}=
c_1\ket{l_1,l_1-1}\otimes \ket{l_2,l_2}+c_2\ket{l_1,l_1}\otimes \ket{l_2,l_2-1}\\
c^{*}_2\ket{l_1,l_1-1}\otimes \ket{l_2,l_2}-c^{*}_1\ket{l_1,l_1}\otimes \ket{l_2,l_2-1}\to l_1+l_2-1 \to 2(l_1+l_2-1)+1
$$
$$2(l_1+l_2)+1\,,2(l_1+l_2-1)+1\, \ldots 2|l_1-l_2|+1$$
які в координатному представленні є простими добутками відповідних функцій $Y_{l_1,m_1}(\theta_1,\varphi_1)\,Y_{l_2,m_2}(\theta_2,\varphi_2)$. В таких станах кожна з підсистем знаходиться в стані з певнимими квантовими числами. Далі діючи наприклад на стани з максимальними значеннями проекції будується лінцюг станів з відповідним значенням квантового числа $L$ системи $\Sigma_{1}\otimes \Sigma_{2}$. <a id='momsum'>В результаті простір розмірності $(2l_1+1)(2l_2+1)$ представляється як ортогональна (пряма) сума просторів з
$L = |l_1-l_2|, |l_1-l_2|+1,\ldots, l_1+l_2$ (див. Рис)</a>.

#### <a id='sec3dosc'> Спектр 3D ізотропного осцилятора</a>
Отримані результати можна використати для аналізу спектра ізотропного 3D осцилятора 
<!--  -->
\begin{equation}
\label{ham3dosc}
\hat{H} = \frac{\hat{\mathbf{p}}^2}{2\,m} + \frac{m\,\omega^2}{2}\,\hat{\mathbf{r}}^2
\end{equation}
<!--  -->
із застосуванням до реальних систем. Оскільки ступені вільності невзаємодіють, то зрозуміло, що
<!--  -->
\begin{equation}
\label{eq14/9}
\begin{array}{l}
[a_{i} ,a_{j}^{\dagger}  ] = \delta_{ij}\,,
H = \hbar\, \omega \,\left(\sum\limits_{i = 1}^{3} a_{i}^{\dagger} a_{i}+
\frac{3}{2} \right)\,, \\
E = \hbar\, \omega \,\left( N + \frac{3}{2} \right)\,,\quad
N = n_{1} + n_{2}+ n_{3}\,.
\end{array}
\end{equation}
<!--  -->
Завдяки сферичній симетрії задачі, гамільтоніан комутує з квадратом моменту імпульсу і його проекцією на вісь. Тому стаціонарні стани можуть бути обраними і як власні вектори також і цих інтегралів руху і характеризуються їх власними числами. Однак, як видно з (\ref{eq14/9}), енергія стаціонарного стану все одне залежить тільки від головного, але ніяк не від орбітального квантового числа, тобто __має місце виродження енергетичних рівнів за орбітальним квантовим числом__. Причина цього факту у більш високій, ніж обертальна симетрія задачі. Образно кажучи, 3-вимірний осцилятор є "сумою" трьох одновимірних. Для одномірного осцилятора групою симетрії є, як ми бачили, $U(1)$. Її узагальненням на 3-вимірний випадок є група $U(3)$, оскільки, як не важко перевірити, гамільтоніан у цьому випадку є інваріантним відносно унітарного перетворення:

\begin{equation}
\label{eq15/9} a_{i} \to \,\,U_{i\,k} \,a_{k} \,,
\end{equation}
де $U_{ik} $ - елементи унітарної матриці $3\times 3$.

Таким чином стаціонарний стан осцилятора $\ket{n_{1},n_{2},n_{3}}$ визначається трійкою цілих чисел які визначають кількість квантів для кожного напряму (моди): 
$$
\psi_{n_{1},n_{2},n_{3}}(x,y,z)  = \braket{x,y,z}{n_{1},n_{2},n_{3}}
$$
Звичайно, що кількість квантів у стац. стані не залежить від вибору базису і тому має місце виродження рівнів з $n>0$.<br> 
Оскільки гамільтоніан \eqref{ham3dosc} інваріантний відносно обертань, то інтегралами руху є компоненти момента імпульсу і відповдіно в якості базиса можна обрати власні вектори операторів $\hat{H}, \hat{L}^2, \hat{L}_{z}$: 
<!--  -->
$$
\ket{\psi} = \sum c_{n,l,m}\,\ket{n,l,m}
$$
<!--  -->
Однак, є певна особливість яка повязана з залежністю властивостей квантових чисел $n$ та $l\, (\le n)$. З одного боку, парність стаціонарного стану визначається парністю $l$: $(-1)^l$, а з іншого - парністю $n$: $(-1)^n$. Тому у стаціонарному стані $n=l ({\rm mod} \,2)$. Зрозуміло, що $l\le n$. 

У термінах операторів народження та знищення оператори орбітального момента мають вигляд:
<!--  -->
\begin{equation}
\label{eq16/9}
\begin{array}{l}
{L}_{i} = \varepsilon_{ijk} x_{j} p_{k} = - \,i\hbar\,\varepsilon_{ijk}
a_{j}^{\dagger}  a_{k} \\
{\mathbf{{L}}}^{\,2} = - \hbar^{\,2}\varepsilon_{ijk} \varepsilon_{imn}
a_{j}^{\dagger}  a_{k} a_{m}^{\dagger}  a_{n} = - \hbar
^{\,2}({A}^{\dagger}{A}- N(N + 1))\,, \\
\end{array}
\end{equation}
<!--  -->
де
<!--  -->
\begin{equation}
\label{eq17/9}{A}= a_{n}a_{n}\,,\quad N = a_{n}^{\dagger}  a_{n}\,.
\end{equation}
<!--  -->
Оператор ${A}$ комутує з ${\mathbf{{L}}}^{\,2}$ і знищує два кванта енергії. Тобто його дія не змінює значення $l$, однак змінює $n\to n-2$: 
$$
\hat{A}\ket{n} \sim \ket{n-2}
$$ 
<!--  -->
З (\ref{eq16/9}) отримуємо:
\begin{equation}
\label{eq18/9} \langle n\,l\,| \hat A^{\dagger} \hat A\,|\, n\,l
\rangle  = n(n + 1) - l(l + 1) \ge 0\,.
\end{equation}

Починаючи зі стану, який відповідає $N = l$, і діючи послідовно оператором ${A}$, отримаємо послідовність станів з фіксованим орбітальним числом, які відрізняються за енергією на два кванти. Таким
чином, кожному значенню головного квантового числа $N$ відповідає послідовність станів з орбітальними числами $l,l - 2,...0(1)$. Це означає, що у стаціонарних станах системи  парність головного квантового числа дорівнює парності орбітального. Це очевидно, оскільки гамільтоніан осцилятора є інваріантним відносно дзеркальної симетрії  $\mathbf{r}\to - \mathbf{r}$, а парність стаціонарних станів, як відомо, визначається як $( - 1)^{l}$, з іншого боку, парність стаціонарного стану тривимірного гармонічного осцилятора визначається як добуток парностей "одновимірних осциляторів", тобто $(-1)^{n_{1} + n_{2} + n_{3}} = ( - 1)^{N}$. Використовуючи отримані результати, спектр 3-вимірного осцилятора
можна подати у вигляді діаграми

<a id='3dosc'>Тому послідовність станів 3D осцилятора в базисі $\ket{n,l,m}$ виглядає так</a>

\begin{align}
E_{0}\to& \ket{0,0,0}\notag \\
E_{1}\to& \ket{1,1,0},\ket{1,1,\pm 1}\notag\\
E_{2}\to& \ket{2,0,0}\oplus\ket{2,2,0},\ket{2,2,\pm 1},\ket{2,2,\pm 2}\\
E_{3}\to& \ket{3,1,0},\ket{3,1,\pm 1}\oplus\ket{3,3,0},\ket{3,3,\pm 1},\ket{3,3,\pm 2},\ket{3,3,\pm 3}\notag\\
\ldots& \notag
\end{align}

Значок $\oplus$ демонструє розклад простору U(3)-представлення на компоненти неприводимі до групи обертань SO(3).

Таким чином, саме більш широка U(3)-симетрія є групою симетрії гамільтоніана \eqref{ham3dosc} вона і спричиняє додаткове виродження енергетичних рівнів за квантовим числом $l$. Кратність виродження (розмірность простору векторів $\ket{N}$): $$\frac{(N+1)(N+2)}{2}$$
-кількість розкладів N на суму 3-х складових $N=n_{1}+n_{2}+n_{3}\,,\,\,n_{i}\ge 0$.
