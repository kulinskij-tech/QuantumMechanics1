Залежність хвильової функції від конкретних величин відображає певну (експериментальну) процедуру вимірювання густини розподілу ймовірності. Те, що стан квантової системи може бути представлений хвильовою функцією, яка залежить від того чи іншого набору параметрів,
аналогічно представленню вектора у певній системі координат. Залежно від вибору представлення, вигляд хвильової функції та її інтерпретація може змінюватись так же, як і аналогічні поняття для координат вектора. Ці міркування підводять до введення поняття
вектора стану, який не пов'язаний з якимось конкретним представленням у якійсь системі координат. Для вектора стану П.А.М.~Дірак ввів позначення $\ket{\psi}$. У випадку координатного або імпульсного представлення стан є функцією з квадратично інтегрованим квадратом модуля на відповідному просторі змінних. Узагальненням результатів, наведених вище, є основний постулат квантової механіки - __принцип суперпозиції__:<br>

## Постулат 1
___Можливі стани квантової системи зображуються векторами комплексного векторного прострору. Якщо квантова система  характеризуєтся тільки просторовими чи імпульсними степенями вільності, то цей простір є
гільбертовим простором функцій $L_{2}$. При цьому всі ненульові скалярні кратні (і тільки вони) ненульового вектора, який зображує стан, відповідають одному і тому ж стану фізичної системи. Тобто стан це промінь $$\lambda\,\ket{\psi}\,, \lambda \in \mathbb{C}$$ 
Нульовий вектор не відповідє ніякому стану.___

У подальшому під станом розуміється довільний нормований вектор. Кріме структури лінійного простору, якая дозволяє визначати разклад векторів у різних базисах, простір станів квантової системи наділений ще однією важлиою операцією - внутршнім, скалярним добутком векторів, яке у координатному представлені має вигляд: 

$$
\braket{\psi_1}{\psi_2} = \int\limits_{V}\psi_1^*(\mathbf{r})\psi_2(\mathbf{r})\,dV =
\int\limits_{\mathbf{k}}c_1^*(\mathbf{k})c_2(\mathbf{k})\,d\mathbf{k}=
\braket{\psi_2}{\psi_1}^{*}\,.
$$ 

З курсу лінійної алгебри відомо що ця операція дозволяє будувати ортогональні проекції векторів. Правило, яке співставляє кожному вектору його ортогональну проекцію на певний векторний підпростір, називається ___ортогональним проекційним оператором___ або ортопроектором. Геометрично очевидна наступна характеристична властивість проектора (і не тільки орто-):
<!--  -->
$$
  \hat{P}^2=\hat{P}\,.
$$
<!--  -->
Оператор ортогонального проектування має наступні <a id='ortoproj'>властивості</a>:
<!--  -->
\begin{eqnarray*}
  \hat P\,(c_{1}\,\ket{\psi_{1}} +
  c_{2}\,\ket{\psi_{2}})&=&c_{1}\,
  \hat P\, \ket{\psi_{1}} +
  c_{2}\,\hat P\, \ket{\psi_{1}}\,,\quad& \text{лінійність} \nonumber \\
  \langle\psi_{2}\vert\hat{P}\psi_{1}\rangle&=&\langle \hat{P}\psi_{2}\vert\psi_{1}\rangle\,,\quad&
  \text{ермітовість} \nonumber \\
  ||\,\psi\,||^2&=&||\,\psi-\hat{P}\psi||^2+||\hat{P}\psi||^2\,,\quad&  \text{ортогональність} \nonumber 
\end{eqnarray*}

## Постулат (проекційний) 2
<!--  -->
Мотивіровкою проекційного постулату, який визначає зв'язок операції вимірювання фізичної величини з математичною операією ортогонального проектування (скалярним добутком в гільбертовому просторі) є приклад вимірювання поляризації світла. В квантовому підході
квадрат модуля вектора стану (напряженность электрического поля) пропорційна до числа квантів - фотонів. Сам фотон є суто квантовим об'єктом - одночастинковий стан ЕМП. У
стані, наприклад, лінійної поляризації ЕМП кожному фотону может бути приписаний відповідний стан лінійної поляризації. Після прохождення поляроїда, який орієнтований вздовж вісі $x$, доля інтенсивності (поляризованого вздовж $x$) світла, що пройде дорівнює (закон Малюса):
<!--  -->
\begin{equation}
\label{prob}
\frac{|\,\mathbf{\mathcal{E}}_{x}\,|^{\,2}}{|\,\mathbf{\mathcal{E}}_{x}\,|^{\,2}+|\,\mathbf{\mathcal{E}}_{y}\,|^{\,2}}
\end{equation}
<!--  -->
Квантова механіка постулює, що статистична
інтерпретація \eqref{prob} може бути застосована і до одиночного фотону. Тобто якщо фотон знаходиться у суперпозиційному стані двох лінійних поляризацій (вектори поляризації вздовж $x$ і $y$ тут вважаються нормованими):
<!--  -->
$$\ket{\psi} = c_x\,\ket{x}+c_y\,\ket{y}\,,\,\,\, c_x\,c_y \ne 0$$
<!--  -->
то неможливо передбачити чи пройде він крізь поляроїд який оріентований вздовж $x$ (або $y$) чи ні. Можна вказати тільки ймовірність:
<!--  -->
$$p_x = \frac{|c_x|^2}{|c_x|^2+|c_y|^2}\,,\,\,\,p_y = 1-p_x$$
<!--  -->
Враховуючи звязок квадрату модуля проекції з інтенсивністю, отримаємо наступний постулат. Зрозуміло, що в якості базисних станів можна обрати будь які ортогональні стани, наприклад, $\ket{\pm}$ циркулярної поляризації.

__Проективний постулат:<br> Система, описується вектором стана $\left|\,\psi\,\right\rangle \in L$. Одразу після вимірювання у якому отримано значення $x$ певної фізичної спостережуваної величини, система буде перебувати у стані $\hat P_{x}
\left|\,\psi\,\right\rangle $, де $\hat P_{x}$ - ортогональний проектор на підпростір власних станів  $\hat P_{x}(L)$, в яких фізична величина має значення $x$. Ймовірність спостереження значення величини $x$ у стані $ \left|\,\psi\,\right\rangle$ визначається квадратом модуля проекції вектора стану $\left|\,\psi\,\right\rangle$ на власний вектор стану в якому фізична величина примає значення $х$:
$$
\braket{\hat P_{x}\psi}{\hat P_{x}\psi}=\braket{\psi}{\hat{P}_{x}\psi}
$$
Стани з підпростіру $\hat P_{x}(L)$ називаються власними, які відпоідають власному значеню $x$ фізичній величині. Якщо підпростір $\hat P_{x}(L)$ одномірний, то говорять, що воно невироджене. У протилежному випадку маєт місце виродження. Кратність виродження є розмірність підпростіру $\hat P_{x}(L)$.__

Таким чином, суперпозиція 
<!--  -->
$$
\ket{\psi}=c_{1}\,\ket{\psi_{1}}+c_{2}\,\ket{\psi_{2}}
$$ 
<!--  -->
векторів описує стан, який демонструє характеристики станів $\psi_{1}$ і $\psi_{2}$ з ймовірностями, визначаємим постулатом. У випадку якщо $x$ є характеристиками даного (нормованого) стану $ \ket{\phi}$, оператор ортогонального проектування може бутий представлений у вигляді
<!--  -->
$$ 
\hat P_{\phi} =
\ket{\phi} \bra{\phi}
$$
<!--  -->
Зміст такого позначеня простий:
<!--  -->
$$
\hat P_{\phi}\left|\, \psi\, \right\rangle =
(\ket{\phi} \bra{\phi})\,
\left|\,\psi\,\right\rangle =
\left\langle\,\phi\,\vert\,\psi\,\right\rangle\,
\ket{\phi}
$$
<!--  -->
Операція ортогонального проектування є прикладом лінійного оператора, тобто правила, співставлення лінійній комбінації векторів лінійної комбінації їх образів з такими ж коефіцієнтами.  Зв'язок формальних алгебраічних дій з квантовомеханічним вимірюванням дається проекційним постулатом.<br>

Разом з повним описом стану квантової системи вектором стану $\vert\psi\rangle$ чи, теж саме, оператором $\hat P_{\psi}$, можна ввести статистичний (у сенсі неповноти знання певних фізчних величин) опис за допомогою ансамбля, аналогічний поняттю класичної (статистичної) фізиці. Сенс такого опису наочно демонструється прикладом приготування станів з певним імпульсом за допомогою термоеміссії. Кожна з емітованих частинок (напр. електронів) буде знаходитись у власному стані оператора імпульса. Однак, завдяки великої кількості ступенів вільності термостата (катода), весь ансамбль емітованих частинок характеризується певним розподілом імпульсів (в стані теплової рівноваги - максвеловським розподілом). Під ансамблем тут розуміється суупність великої кількості однакових систем, які знаходяться у одинакових умовах, але які тим не менше не дозволяють визначити точний мікроскопічний стан. Іншими словами, таким умовам відповідає багато мікроскопічних станів. Визначення ансамбля включає взначення статистичної ваги з якою певний мікростан який зображується вектором стану (хвильовою функцією), входить до ансамбля. Безпосередньо з цього витікає, що оператор, аналогічний $\hat P_{\psi}$, в даному випадку має вигляд:
<!--  -->
\begin{equation}\label{stat}
\hat \rho = \sum_{n}p_{n}\hat P_{\psi_{n}}\,,\quad p_{n}\ge 0\,,\quad \sum_{n}p_{n}=1
\end{equation}
<!--  -->
Оператор \eqref{stat} називаєтся статистичним оператором чи матрицею густини. Тут $p_{n}$ є стат. вага
стана $ \left|\,\psi_n\right\rangle$ у ансамблі. Зрозуміло, що в частному випадку $p_{n}=1$, ми маємо найбільш повний опис квантової системи. Такі стани називаються чистими. В загальному випадку
\eqref{stat} це змішаний стан, і відповдно до \eqref{stat} матриця густини є среднім зваженим проекційних операторів чистих станів які формують ансамбль.<br>

Геометрично чисті та змішані стани утворюють опукле тіло. Для $q$-бита - це шар, границя якого $\mathbb{S}^2$ представлена чистими, а внутрішність - змішаними станами. В базисі станів $\left\{|\,\phi_{n}\,\rangle\right\}$\, стат. оператор чистого стану $\left|\, \psi\, \right\rangle=\sum_{n}c_{n}\,|\,\phi_{n}\,\rangle$ має вигляд:
<!--  -->
$$
  \hat P_{\psi}=\ket{\psi}\,\bra{\psi} 
  = \sum_{n,m}c^*_{n}\,c_{m}\ket{\phi_{m}}\bra{\phi_{n}}
 $$
 з якого очевидна властивість ермітовості $\hat{P} = \hat{P}^{\dagger}$.
