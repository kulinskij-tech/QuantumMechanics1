## Fresnel integrals for rectangular packet

```
rect02 = HeavisidePi[x/2]/Sqrt[2];
Plot[(HeavisidePi[x/2]/Sqrt[2]), {x, -2, 2},TicksStyle->4, Filling -> Bottom,ImageSize->100]
rectk0 = FourierTransform[rect02, x, k]
```

${\rm Re}\, \psi = $ 

```
wfre = Simplify[1/(Pi*Sqrt[2])*Assuming[{t > 0, x \[Element] Reals},
Integrate[rectk0*Cos[(k*x - k^2*t)], {k, -\[Infinity], +\[Infinity]}]]]
(*Export["wfre.txt",TeXForm[wfre]]*)
```

${\rm Im}\, \psi = $ 

```
wfim = Simplify[1/(Pi*Sqrt[2])*Assuming[{t > 0, x \[Element] Reals},
Integrate[Sin[k]/k*Sin[(k*x - k^2*t)], {k, -\[Infinity], +\[Infinity]}]]]
(*Export["wfim.txt",TeXForm[wfim]]*)
```
