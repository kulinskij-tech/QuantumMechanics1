## Математичне доповнення: Елементи теорії самоспряжених поширень ермітових операторів.
$\newcommand{\ket}[1]{\left.\left\vert\, #1\,\right. \right \rangle}$
$\newcommand{\bra}[1]{\left. \langle \, #1\, \right\vert}$
$\newcommand{\braket}[2]{\langle \,#1 \vert #2 \, \rangle}$
$\newcommand{\av}[3]{\left\langle \,#1 \left \vert\, #2 \,\right\vert #3 \,\right \rangle}$

У нескінченовимірному просторі побудова оператора якій відповідає фізичній величині є сложнішою ніж у скінечнновимірному, де ермітові оператори представляються самоспряженими матрицями $\hat{A} = \hat{A}^{\dagger}$, оскільки випадок дискретного спектра не вичерпує всіх можливостей. Тільки обмежені ермітови оператори можуть бути визначені на всьому просторі $L_2$ (див. теорему \ref{ht}). Необмежені ермітові оператори не можуть бути визначені повсюди у $L_2$ . Це видно на прикладі операторів координати і імпульсу.
Обидва ці оператори є ермітовими і необеженими (це не відноситься до оператора координати у випадку, якщо конфігураційний простір - обмежена область). Тому область визначення операторного виразу який претендуватиме на відповідність фізичної величини треба поширити.
Можна вказати природню область визначення (яку у подальшому будемо вважати всюди щільною у $L_2$) для необмеженного оператора, як лінійну підмножину векторів, на яких значення оператора $\hat{A} \,\psi$ також є елементом гільбертова простору фізичних станів $L_2$:

\begin{equation}\label{da}
  D(\hat{A})=\{\psi\in L_{2}:\hat{A}\psi\in L_2\}
\end{equation}

З фізичної точки зору така область визначення для ермітова оператора означає, що розглядаються стани для яких визначена дисперсія спостережної величини $\hat{A}$:

$$ \braket{\hat{A}\,\psi}{\hat{A}\,\psi} = \av{\psi}{\hat{A}^2}{\psi} $$

Суто математично до цього можна прийти вводячи поняття __графіка__ оператора як підмножини ортогональної суми лінійних просторів аргументів та значень $L_2\oplus L_2$ виду:

\begin{equation}\label{gra}
  \Gamma_{\hat{A}} =\{\psi,\hat{A}\psi\}\subseteq L_2 \oplus L_2
\end{equation}

У просторі $L_2\oplus L_2$ вводиться відповідна норма:

\begin{equation}\label{dirsum}
\|\{\psi_1,\psi_2\}\,\|^{\,2}_{L_2\oplus L_2}=\|\,\psi_1\|^{\,2}+\|\psi_2\|^{\,2}
\end{equation}

тоді для графика маємо

\begin{equation}\label{dirsum1}
  \|\{\psi,\hat{A}\psi\}\|^{\,2}_{L_2\oplus L_2}=\|\,\psi\,\|^{2}+\|\hat{A}\psi\|^{2}
\end{equation}

Очевидно, визначення лінійного оператора еквівалентно визначенню його графіка, тому логічно зробити область визначення оператора замкненою відносно норми \eqref{dirsum}. Це однак не означає, що вона стане такою у нормі $L_{2}$, оскільки норма \eqref{dirsum1} сильніша за норми $L_2$ (відносно неї менше послідовностей збігаються, оскільки вимагається збігання значень). Крім того, може статися, що замикання по нормі \eqref{dirsum1} веде до появи елементів виду $\{0,\psi\ne 0\}$. Такі пари векторів не можуть належати графіку лінійного оператора (порушується лінійність). В такому випадку говорять, що оператор не допускає замкнення. Якщо ж оператор допускає замкнення, то:

$$\forall \, \psi_{n}: \|\,\psi_{n}\,\|\rightarrow 0\,,\,\,\,
\hat{A} \,\psi_{n}\in L_{2} \Rightarrow \|\hat{A}
\,\psi_{n}\,\|\rightarrow 0.$$ 

Оператор природньо назвати __замкнненим__, якщо його графік є замкненим подпростором відносно норми \eqref{dirsum}. Неважко зрозуміти, шо ортогональна сума гільбертових просторів також є гільбертовим простором. Скалярний добуток у $L_2\oplus L_2$ будується  очевидним чином:

\begin{equation}\label{scaprod}
  \{\psi_{1},\hat{A} \,\psi_{1}\}\cdot  \{\psi_{2},\hat{A} \,\psi_{2}\}=
  \braket{\psi_{1}}{\psi_{2}}+\braket{\hat{A} \,\psi_{1}}{\hat{A} \,\psi_{2}}
\end{equation}

Використовуючи \eqref{scaprod}, наприклад, графік спряженого оператора можна розглядати як ортогональне доповнення до "графіка" повязного з вихідним оператором (графік оператора "повернутий на $90^{\rm{o}}$"):

\begin{equation}\label{opa}
  O\Gamma_{\hat{A}}=\{-\hat{A} \,\psi,\psi\}
\end{equation}

Дійсно, ортогональність до \eqref{opa}, завдяки \eqref{scaprod} означає що:

\begin{equation}\label{ortgr}
\{\phi,\tilde\phi\}\cdot\{-\hat{A} \,\psi,\psi\}=-\av{\phi}{\hat{A}}{\psi}+\braket{\tilde\phi}{\psi\vphantom{\tilde\phi}}=0
\end{equation}

це співпадає, за означенням, з правилом дії спряженого оператора: $\tilde \phi=A^\dagger \phi$. Звідси витікає, що графік спряженого оператора, а значить і сам оператор
$A^\dagger$, є замкненим (оскільки ортогональне доповнення до довільної множини є замкненим завдяки неперервності скалярного добутку). Тому, для оператора (з густою у $L_2$ областю визначення) його спряжений визначений, на більш широкій множині, тобто $D(A)\subset D(A^\dagger)$. Область визначення всякого іншого замкненого розширення вихідного оператора менша за $D(A^\dagger)$. Якщо вихідний оператор ермітов, то $\hat{A}$ і $\hat{A}^\dagger$ співпадають на $D(A)$. Т.ч. ермітовий оператор (не обовязково замкнений) завжди має замкнене поширення у вигляді свого спряженого. Якщо додатково розглянути подвійно спряжений $A^{\dagger\,\dagger}$, то його графік як раз і є найменшою замкненою множиною, яка містить графік вихідного оператора (як подвіне ортогональне доповення), співпадаючи тим самим із замкненням оператора. Ермітовий оператор A називається __самоспряженим__ якщо:

$$D(\hat{A})=D(\hat{A}^\dagger)$$

Якщо ермітовий оператор замикаемий, і його замкнення самоспряжено, то такий оператор називается __суттєво самоспряженим__, оскільки в такому випадку замкнення цього оператора є единим самоспряженим розширенням цього оператора. Очевидно, що в этом випадку такое розширення может бути ототожнено з $A^{\dagger\dagger}$.

Таким чином, ермітовий оператор может бути зроблений самоспряженим за рахунок розширення його області визначення, або, що еквівалентно, шляхом звуження області визначення його спряженого. У загальному випадку замкнений ермітов оператор може мати декілька (навіть нескінсенне число) самоспряжених розширень, які відповідають різним фізичним ситуаціям, або не мати жодного. У останньому випадку оператор не може відповідати ніякій фізичній величині. Тому, вказання області визначення ермітового оператора є важливим фізичним моментом, який визначає постановку задачи. Оскільки спектр ермітового оператора належить дійсній вісі, розглянемо величину:

\begin{equation}
\label{zz}
\begin{array}{l}
 \vert \vert (A - z\,I)\psi \vert \vert^{\,2} = \braket{(A - zI)\psi}{(A - zI)\psi} =
\vert \vert (A - x\,I)\psi \vert \vert^{\,2} - i\,y\,\braket{(A - x\,I)\psi
}{\psi} + \\
\,\,i\,y\,\braket{\psi }{(A - x\,I)\psi } + y^{\,2}\vert \vert \psi \vert \vert^{\,2}
\ge y^{\,2}\vert \vert \psi \vert \vert^{\,2} 
 \end{array}
\end{equation}

тобто оператор $\hat{A}-z\hat I$ має неперевний зворотній для ${\rm Im}\, z\! \ne 0 $, який визначений на замкненому (завдяки неперевності $(\hat{A}-z\hat I)^{-1}$) підпросторі - області значень $\mathcal{R}(\hat{A}-z\hat I)$. Цей підпростір може не співпадати з усім $L_2$. Тому важливо розглянути розмірності наступних підпросторів які є відповідними ортогональними доповненнями до області значень $\mathcal{R}(\hat{A}-z\hat I)$:

\begin{eqnarray}\label{deff}
  N_{z_1}&=& \dim \left( {\Re (A - z_{1} \,I)^{\bot}}  \right) =
  \dim \,\left( {\rm Ker}(A^{\dagger}  - z_{1}^{^*} \,I) \right),  \,{\rm Im} z_{1} > 0
  \nonumber \\
N_{z_2}&=& \dim \left( \Re (A - z_{2} \,I)^{\bot}  \right)\, =
\dim \,\left( {\rm Ker}(A^{\dagger}  - z_{2}^{\ast} \,I) \right),
\,{\rm Im} z_{2} < 0
\end{eqnarray}

Це т.з. __дефектні числа__. Можна показати, що вони є сталими у відповідних полуплощинах (розмірність є цілою, і неперервно залежить від $z$, тому така функція є сталою). В якості $z_1$, $z_2$ можно обрати $+i$ и $-i$. Основний результат полягає в тому, що ермітов оператор допускає самоспряжене розширення, т.і.т.т. якщо дефектні числа однакові. При цьому, якщо вони дорівнюють 0, то оператор суттєво самоспряжений ($A^{\dagger\dagger}$- єдине самоспряжене розширення). Завдяки неперервності дефектних чисел від $z$, обмежені за нормою (знизу чи зверху) ермітови оператори мають самосоряжені розширення (оскільки верхню полуплощину можна з'єднати неперевним шляхом з нижньою ніде не перетинаючи спектр оператора). Власне кажучи, для цього достатньо, щоб на дійсній вісі існувало хоча б одне регулярне значення, зокрема наявність дискретного спектра у оператора гарантує її існування. Ідея розширення ермітова оператора до самоспряженого геометрично доволі проста, якщо дефектні числа скінчнні. Наявність дефектних підпросторів означає, що:

$$\exists\, u\in L_2\,:\, \braket{u}{(\hat{A}-z\hat I)\,\psi} = 0\,,\,
\forall\, \psi \in D(A)$$

$$\exists\, w\in L_2\,:\,
\braket{w}{(\hat{A}-z^*\hat I)\,\psi} = 0\,,\,\forall \,\psi \in D(A)$$ 

і тому область визначення $\hat{A}^{\dagger}$ представляется у вигляді прямої суми:

\begin{equation}\label{adag}
  D(\hat{A}^{\dagger})=D(\hat{A})\oplus N_{z}\oplus N_{z^*}.
\end{equation}

Сама дія оператора $\hat{A}^{\dagger}$ визначається наступним
чином:

\begin{equation}\label{adag1}
  \hat{A}^{\dagger}\psi=\hat{A} \phi + z^* u + z w\,, u \in N_{z}\,,
  w \in N_{z^*}
\end{equation}

оскільки оператор $\hat{A}-z\hat I$ має зворотній при ${\rm Im}\, z\!\ne 0$, на власній області значень $\mathcal{R}(\hat{A}-z\hat I)$, розглянемо оператор:

\begin{equation}\label{kelly}
  \hat V_{z}=(\hat{A}-z^*\hat I)(\hat{A}-z\hat I)^{-1}
\end{equation}

якій визначений на $\mathcal{R}(\hat{A}-z\hat I)$. У скінченновимірному випадку \eqref{kelly} це відоме перетворення Келі, яке кожній ермітовій матриці ставить у відповідність унітарну. Зокрема
для одновимірній ситуації де гільбертовим простором є $\mathbb{C}$, \eqref{kelly} це просто дробно-лінійне відображення комплексній прямій

$$x \to \frac{x-i}{x+i}\,\in \,U(1)$$

при якому дійсна вісь (одновимірні ермітови оператори) відображається на одиничне коло (одновимірні унітарні оператори). З такої аналогії зрозуміло, що дефектні підпростори для ізометричного оператора є 

$$\left(\mathcal{R}(\hat V_{z}-w\hat I)\right)^{\bot} = N_{z},\,\, |w|>1$$ 

і

$$\left(\mathcal{R}(\hat V_{z}-w\hat I)\right)^{\bot} =
N_{z^*},\,\, |w|<1.$$ 

Як вже зазначалось, у скінченновимірному випадку поняття ізометричного і унітарного оператора співпадають (так як і ермітовість та самоспряженність). У нескінченновимірному випадку
\eqref{kelly} здійснює відповідність між ермітовими і ізометричними операторами з відповідними областями визначення. Изометричность оператора \eqref{kelly} на области визначення следует непосредственно из визначення и ермітовийости оператора $\hat{A}$:

$$  \forall \, u_{1}\,:\,\,u_{1} = (\hat{A}-z\hat I)\psi_{1}\,,\quad
\forall \, u_{2}\,:\,\, u_{2} = (\hat{A}-z\hat I)\psi_{2}$$

$$\braket{\hat V u_{1}}{\hat V u_{2}}=\braket{(\hat{A}-z^*\hat I)\psi_{1}}{(\hat
A-z^*\hat I)\psi_{2}}=(\hat{A}\psi_{1},\hat{A} \,\psi_{2})+
|z|^{\,2}\braket{\psi_{1}}{\psi_{2}}-\\
z^*\braket{\hat{A}\psi_{1}}{\psi_{2}}-z\braket{\psi_{1}}{\hat{A}\psi_{2}} $$ 

що співпадає з:

$$\braket{u_{1}}{u_{2}}=\braket{(\hat{A}-z\hat I)\psi_{1}}{(\hat{A}-z\hat I)\psi_{2}}=\braket{\hat{A}\psi_{1}}{\hat{A} \,\psi_{2}}+
|z|^{\,2}\braket{\psi_{1}}{\psi_{2}}-z\braket{\hat
A\psi_{1}}{\psi_{2}}-z^*\braket{\psi_{1}}{\hat{A}\psi_{2}}
$$

завдяки ермітовостs оператора $\hat{A}$. Очевидно, що 

$$D(\hat
V)=\mathcal{R}(\hat{A}~-~z\hat I),\qquad
\mathcal{R}(V)=\mathcal{R}(\hat{A}~-~z^*\hat I).$$

Тепер зрозуміло, яким чином розповсюдити дію на весь гільбертів простір із збереженням ізометричності, якщо дефектні числа рівні. Необхідно розширити \eqref{kelly} до унітарного оператора доозначив його довільним ізометричним, а, завдяки скіченновімірності, відповідно і унітарним відображенням між дефектними підпросторами:

\begin{equation}
\label{uni} W:\,\left( {\Re (A - z\,I)} \right)^{\bot}  \to \left(
{\Re (A - z^* \,I)} \right)^{\bot}
\end{equation}

Множина всіх таких операторів параметризується множиною унітарних матриц $U(n)$, где $n$-розмірність дефектних підпросторів. Кожному такому унітарному розширенню відповідає самоспряжене розширення вихідного оператора:

\begin{equation}\label{ext}
  \tilde A = (z\tilde V - z^*I)(\tilde V - I)^{-1}
\end{equation}

область визначення якого представляється у вигляді прямої суми:

\begin{equation}\label{dext}
D(\tilde A) = D(\hat{A})\oplus (\hat W-\hat I)\left(N_{z}\right)
\end{equation}

тобто:

\begin{equation}\label{dext1}
\tilde \psi = \psi - \phi + \hat W\, \phi\,, \quad \psi\in D(\hat
A)\,,\, \phi \in N_{z}\,,
\end{equation}

і

\begin{equation}\label{aext}
\tilde A\tilde \psi = \hat{A} \,\psi - z\phi + z^*\hat W\, \phi\,,
\end{equation}

Вирази \eqref{adag} и \eqref{aext} називаються __першою та другою формулами Неймана__ відповідно.

Зазначимо, що у всіх виразах $z$ було фіксованим. В якості $z$ можна обрати наприклад $i$.

Продемонструємо загальну теорію на конкретному прикладі вже введенних раніше операторів координати та імпульсу. Для простоти розглянемо одновимірний випадок необмеженного руху, тобто
простором станів є $L_2(\mathbb{R})$. Оператор координати, визначений на щільній множині:

\begin{equation}\label{dx}
  D(\hat x)=\{\psi: \|\,\psi\|^{\,2}+\|\hat x\, \psi\|^{\,2}<\infty\}
\end{equation}

і є самоспряженим (очевижно, що оба дефектних числа дорівнюють 0). Область визначення оператора імпульсу:

\begin{equation}\label{dp}
  D(\hat p)=\{\psi: \|\,\psi\|^{\,2}+\|\hat p\, \psi\|^{\,2}<\infty\}
\end{equation}

також є щільною у $L_2$ і є гільбертовим простором відносно відповідного скалярного добутку. Цей простір позначається $W^{1}_{2}(\mathbb{R})$ і називається __простором Соболєва__ типу (1,2) $W^{1}_{2}(\mathbb{R})$). Він складається з неперевних функцій, чия похідна існує як елемент простору $L_2$. Достатньо скористатися фактом того, що множина $C^{\infty}_{0}$ нескінченно диференційованих функцій, які зникають разом зі всіма своїми похідними назовні певної обмеженної області (т.з. функції з компактним носієм), щільно у $L_2$. Розглянемо послідовність $f_{n}\in C^{\infty}_{0}$, яка збігається до $\frac{d\psi}{dx}$ за нормою $L_2$ (для спрощення вважаємо функції дійсними):

$$\left|\left|f_{n}-\frac{d\psi}{dx}\right|\right|_{L_2}\rightarrow 0
\Rightarrow \forall \, u \in C^{\infty}_{0}: \lim_{n\rightarrow
\infty} \int\limits_{-\infty}^{+\infty}u
f_{n}dx=-\int\limits_{-\infty}^{+\infty}\frac{du}{dx} f_{n}dx=
\int\limits_{-\infty}^{+\infty}\frac{df_{n}}{dx} udx,
$$ 

але

$$
\int\limits_{-\infty}^{+\infty}u
f_{n}dx = \int\limits_{-\infty}^{+\infty}u dF_{n}(x) =-
\int\limits_{-\infty}^{+\infty}\frac{du}{dx} F_{n}dx
$$ 

Тут $F_{n}$ першообразна функції $f_{n}$. Зазначимо, що вибір першообразної не впливає на значення інтеграла. Тому покладемо:

$$F_{n}(x) = \int\limits_{0}^{x}f_{n}(y)dy$$

Застосовуючи нерівність Коши, отримаємо:

$$\left|\int\limits_{0}^{x}f_{n}(y)dy -
\int\limits_{0}^{x}\psi'(y)dy\right|^2\le
|x|\,\int\limits_{0}^{x}\,|f_{n}(y)-\psi'(y)|^{\,2}dy\le
|x|\,\|f_{n}-\psi'\|^{\,2}_{L_2}$$ 

що означає рівномірну збіжність першообразних на довільному скінченному проміжку. Завдяки відомій теоремі, границі послідовності таких першообразних є неперервною функцією. Далі, рівняння

\begin{equation}\label{p}
  \frac{d\psi}{dx}=\pm \psi
\end{equation}

не має розв'язків у $L_2$, тому індекси дефекта - $(0,0)$ і оператор $\hat p$ суттєво самоспряжений з областю визначення \eqref{dp}. З фізичної точки зорения область визначення
\eqref{dp} оператора імпульса це стани з обмеженною дисперсиєю розподілу імпульса.

Теперь розглянемо випадок обмеженного руху, де простір состояний (у координатному представленні) є $L_{2}[a,b]$:

$$L_{2}[a,b] = \left\{\psi: \int\limits_{a}^{b}\,|\,\psi|^{\,2}dx<\infty\right\}$$

В цьому випадку оператор координати є обмеженим і його самоспряженість очевидна. Відмітимо, що тим не менше, оператор координати не має жодного власного вектора у $L_2$. Для оператора імпульса ситуація складніша. Рівняння \eqref{p} вже має квадратично інтегровані на проміжку розв'язки:

\begin{equation}\label{p1}
  \psi_{\pm} = c_{\pm} \exp(\pm x)
\end{equation}

Тому індекси дефекта є $(1,1)$ і оператор має різні самоспряжені розширення. Кожне таке розширення характеризуется комплексним числом виду 

$$\exp(i\theta)\,, \,\theta\in [0,2\pi)\,,$$ 

тобто є елементом з $U(1)$. Покажемо це.

Оператор $\hat p\,\psi = -i\frac{\partial\, \psi}{\partial \, x}$ ермітовий, якщо в якості області визначення обрати простір функций \eqref{dp} з додатковою умовою:

\begin{equation}\label{dpo}
D(\hat p)=\left\{\psi:\|\hat p\, \psi\|<\infty,\,
\psi(a)=\psi(b)=0\right\}.
\end{equation}

При цьому спряжений оператор визначений на більш широкій множині
\eqref{dp}. З \eqref{dext1} маємо:

$$\tilde \psi(x) = \psi(x)-
c\frac{\exp\,(x)}{\sqrt{\frac{2}{\exp(2b)-\exp(2a)}}}+c
\frac{\exp(-x+i\alpha)}{\sqrt{\frac{2}{\exp(2b)-\exp(2a)}}}\exp(b+a)$$

Неважко перевірити, що областю визначення самоспряженого розширення $\hat p_{\alpha}$ ермітова оператора $\hat p$ є:

$$D(\hat p_{\alpha}) = \left\{\psi:\|\hat p\, \psi\|<\infty,\,
\psi(a)=\exp(i\,\theta(\alpha))\psi(b)\right\},
$$

$$\exp(i\,\theta(\alpha)) = \frac{\exp(b+i\alpha/2)-\exp(a-i\alpha/2)}
{\exp(b-i\alpha/2)-\exp(a+i\alpha/2)}.
$$

Тепер розглянемо випадок руху на піввісі $x>0$ для формально того ж оператора "імпульсу" $\hat{p} = -i\,\frac{d}{d\,x}$. Область ермітовості цього оператора визначається функціями з відповідного соболівського простору які задовільняють граничній умові $\psi(0) = 0$. Спряжений оператор $p^{\dagger}$ визначений на функціях на які не накладаються ніякі умови (окрім того що $\psi' \in L_2$). Очевидно, що тут неможливо розширити область визначення $\hat{p}$ так щобі не порушти його ермітовість. Відповідно це означає, що такий ермітовий оператор не може бути поширений до самоспряженого. Застосовуючи формальну теорію, визначемо дефектні числа як розмірності ядер операторів $(\hat{p}^{\dagger} \pm i)$:
$$
\psi'(x) = \pm \,\psi(x)\,,\,\,\,x>0
$$
Очевидно, що 

$${\rm dim\, Ker}\,(\hat{p}^{\dagger} + i) = 0 \ne 1 = {\rm dim\, Ker}\,(\hat{p}^{\dagger} - i)$$ 

тобто даний ермітов оператор немає самоспряженного поширення.
