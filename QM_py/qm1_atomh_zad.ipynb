__Задача__ Побудуйте класичні дужки Пуасона для інтегралів руху $H, L_{i},R_{k}$ у полі Кулона $U = -\gamma/r$
<br>
__Розв'язок__ Продемонструємо як такі розрахунки проводити за допомогою пакета SymPy. Визначимо необхідні величини

```python
x,y,z,gamma, mu,px,py,pz = symbols('x y z gamma mu px py pz',real=True)
posc = [x,y,z]
# 
rc = sqrt(x**2+y**2+z**2)
# 
impc = [px,py,pz]
# 
hamc = (px**2+py**2+pz**2)/(2*mu) - gamma/rc
# 
momc=[]
for i in range(3):
    p0=0
    for j in range(3):
        for k in range(3):
            p0=p0+LeviCivita(i,j,k)*posc[j]*impc[k]
    momc.append(p0)
# 
rungec = []
for i in range(3):
    p0=0
    for j in range(3):
        for k in range(3):
            p0=p0+LeviCivita(i,j,k)*impc[j]*momc[k]
    rungec.append(p0-mu*gamma*posc[i]/rc)
# 
def poiss(a,b):
    p0=0
    for i in range(3):
        p0=p0+diff(a,posc[i])*diff(b,impc[i])-diff(b,posc[i])*diff(a,impc[i])
    return p0
```

```python
print('Перевірка канонічного співвідношення {x,p}')
xp=eye(3) #identity 3x3 matrix 
for i in range(3):
    for j in range(3):
        xp[i,j]=poiss(posc[i],impc[j])
xp
```

```python
print('Дужка {x[i],L[j]}=x[k]')
xmom=eye(3)#identity 3x3 matrix
for i in range(3):
    for j in range(3):
        xmom[i,j]=poiss(posc[i],momc[j])
xmom
```

```python
print('Дужка {p,L}')
pmom=eye(3) #identity 3x3 matrix
for i in range(3):
    for j in range(3):
        pmom[i,j]=poiss(impc[i],momc[j])
pmom
```

```python
print('Дужка {H,L} - перевірка того що L зберігається')
for i in range(3):
    print('[H,L(',posc[i],')]=',simplify(poiss(hamc,momc[i])))
```

```python
print('Дужка {H,R} - перевірка того що R зберігається')
for i in range(3):
    print('[H,R(',posc[i],')]=',simplify(poiss(hamc,rungec[i])))
```

```python
print('Дужки {R[i],R[j]}: перевірка того що {R[i],R[j]} = - 2 m H L[k] ')
rr=eye(3) #identity 3x3 matrix
for i in range(3):
    for j in range(3):
        rr[i,j]=simplify(poiss(rungec[i],rungec[j])/(2*mu*hamc))
rr
```

  [R_{\,i}, R_{j}\, ] = -2\,i\,\hbar\,m \,\varepsilon_{i\,j\,k}\, L_k\,

Згідно з \eqref{aa} власні значення операторів $\mathbf{R}^2$, $\mathbf{L}^2$ і $H$ пов'язані певним співвідношенням:

\begin{equation}\label{ha}
  \mathbf{R}^2 = \left(\varepsilon

\end{equation}

яке різниться з класичним наявністю члена пропорційного до $\hbar^2$. На підпросторі зв'язаних станів 
($E =
-\frac{\varkappa^2}{2\,m}<0$):

\begin{eqnarray}\label{ha1}
[R_i,R_j\,]  = i\,\hbar\, \varkappa^2 \epsilon_{i\,j\,k}\,L_k\,,\\
  \mathbf{R}^2 = - \varkappa^{\,2}\,
(\mathbf{L}^2+\hbar^2) +m^2\,\gamma^2\,.
\end{eqnarray}

Розглянемо оператори:

\begin{equation}\label{lp}
  \mathbf{W}^{\pm} = \frac{1}{2}\left(\mathbf{L} \pm \frac{1}{\varkappa} \mathbf{R}\right)
  \end{equation}

Виражаючи $\mathbf{W}^{\pm}$ через $\mathbf{L}$ и $\mathbf{R}$, неважко перевірити співвідношення для стаціонарних станів з енергією $E$:

\begin{gather}
  [W^{\pm}_i,W^{\pm}_j] = i\,\hbar\, \varepsilon_{i\,j\,k}
  W^{\pm}_k \,,\notag \\
[W^{\pm}_i,W^{\mp}_j] = 0\,, \notag\\ \mathbf{W}^{+^2} =
\mathbf{W}^{-^2}\,, \notag \\ \mathbf{W}^{+^2} + \mathbf{W}^{-^2}
= \frac{\gamma^2\,m^2}{2\,\hbar^2\,\varkappa^2} - \frac{\hbar^2}{2}\,.
\label{hw}
\end{gather}

Іншими словами, оператори $\mathbf{W}^{\pm}$ мають ті ж самі властивості,
як і оператор кутового моменту. Тому власні значення операторов $\mathbf{W}^{\pm^2}$ мають вігляд:

$$
\mathbf{W}^{\pm^2} \to \hbar^2 j(j+1)\,.
$$

При цьому, $j$ може приймати і полуцілі значення. Остаточно, для рівней енергії з \eqref{hw} маємо:

\begin{equation}\label{enrg_h}
  E = -\frac{m\,e^4}{2\hbar^2}\frac{1}{(2j+1)^2} = -Z^2\frac{\rm Ry}{n^2}\,.
\end{equation}

Знову зкористаємось пакетом Sympy

```python
psi = symbols('psi',cls=Function)
x,y,z,gamma = symbols('x y z gamma',real=True)
pos = [x,y,z]
#
def coord(f):
    p=[]
    for x in pos:
        p.append(x*f)
    return p
# 
def imp(f):
    p=[]
    for x in pos:
        p.append(-I*diff(f,x))
    return p
# 
def mom(f):
    p=[]
    for i in range(3):
        p0=0
        for j in range(3):
            for k in range(3):
                p0=p0+LeviCivita(i,j,k)*pos[j]*imp(f)[k]
        p.append(p0)
    return p
#
def ham(f):
    s0=0
    for x1 in pos:
        s0=s0-diff(f,x1,2)/2
    h=s0 - gamma/sqrt(x**2+y**2+z**2)*f
    return h
print('Hamiltonian:')
ham(psi(x,y,z))
```

```python
# [p,H]
i=1
simplify(imp(ham(psi(x,y,z)))[i]-ham(imp(psi(x,y,z))[i]))/I
```

```python
# [x,H]
for i in range(3):
    print(simplify(coord(ham(psi(x,y,z)))[i]-ham(coord(psi(x,y,z))[i]))/I)
# 
i=0
simplify(coord(ham(psi(x,y,z)))[i]-ham(coord(psi(x,y,z))[i]))/I
```
