Рішення задач цього розділу базується на алгебрі операторів $a$ і $a^\dagger$. Алгебраічний метод побудови спектра гармонічного осциллятора може бути узагальнено наступним чином. Нехай є оператор Шредингера $\hat{H}$ з дискретним невиродженим спектром $E_n\,, n=0,1\ldots$. Визначимо сходинкові оператори:
$$
\hat{a}
\left|\,E_n\,\right\rangle = \sqrt{E_n - E_0}
\left|\,E_{n-1}\,\right\rangle\,,\quad \hat{a}^{\dagger}
\left|\,E_n\,\right\rangle = \sqrt{E_{n+1}- E_0}
\left|\,E_{n+1}\,\right\rangle\,,
$$
вони мають комутаційні властивості:
$$
 \left\langle\, m
\,\right|\left[\hat{a},\hat{a}^{\dagger}\right] \left|\,n
\,\right\rangle = (E_{n+1} - E_n)\delta_{m n}\,.
$$
В такому представленні, аналогічно до \eqref{eq_osc_creatanih2},  всі стани можуть бути отримані в результаті дії оператора народження на основний стан $\vert E_0 \rangle$.  
Очевидно має місце представлення:
$$
\hat H_{0} = a^{\dagger}\,a +
E_0\,.
$$
Крім того, кожен гамільтоніан зі спектром $E_{0}<E_{1}<\ldots$ природньо породжує послідовність гамільтоніанів зі спектрами:
$$
{\rm Spec} H_{n} = E_{n}<E_{n+1}< \ldots
$$
Спробуємо побудувати таку послідовність. Якщо $\hat{a}_{0}$ обрати у вигляді:
$$\hat{a}_{0} = p + i\,\frac{\psi_0'}{\psi_0}\,,\quad  
\hat{a}_{0}\,\psi_0 = 0 $$ де $\psi_0$ - хвильова функція основного стана, тобто $\hat{a}_{0}$ знешкоджує основний стан, тоді:
$$
\hat{a}^{\dagger}_{0} \hat{a}_{0} = \left(\hat p - i\,\frac{\psi_0'}{\psi_0}\right)
\left(\hat p + i\,\frac{\psi_0'}{\psi_0}\right) = p^2 +
\frac{\psi_0''}{\psi_0} = \hat{p}^2+V(x)-E_0 = H_{0}-E_0\,,\quad m=1/2
$$
Послідовно виключаючи зі спектру значення $E_n$, знайдемо оператори:
$$
\hat{H}_{n} = \hat{a}_{n-1}\hat{a}_{n-1}^{\dagger} + E_{n-1} = 
\hat{a}_{n}^{\dagger}\,\hat{a}_{n} + E_n\,,\\
\hat{a}_n
\left|\,E_k\,\right\rangle = \sqrt{E_k - E_n}
\left|\,E_{k-1}\,\right\rangle\,,\quad \hat{a}^{\dagger}_{n}
\left|\,E_k\,\right\rangle = \sqrt{E_{k+1}- E_n}
\left|\,E_{k+1}\,\right\rangle\,.
$$
Зрозуміло, що мають місце включення $Spec \hat{H}_{n+1} \subset Spec \hat{H}_{n}$ і виконується співвідношення:
\begin{eqnarray}
  \hat{a}_{n} \, \hat{H}_{n} =& \hat{H}_{n+1}\,\hat{a}_{n} \notag\\
  a^{\dagger}_n\, \hat{H}_{n+1} =& \hat{H}_{n}\, a^{\dagger}_n \label{han}
\end{eqnarray}
з якого витікає, що:
\begin{equation}\label{anh}
\hat{H}_{n+1} = \hat{a}_{n} a^{\dagger}_n + E_n\,.
\end{equation}
Тобто, задача о знахождені спектра оператора $\hat{H}_{0}$
вирішується побудовою послідовності операторів $\hat{H}_{n}$.
