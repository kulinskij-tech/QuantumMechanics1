В рамках формализму квантової механіки спектр оператора пов'язаний із значеннями фізичної величини, яка представляється цим оператором. Фізична величина, з якою пов'язується процедура її вимірювання, зображується лінійним оператором на векторному просторі.

## Спектр оператора. Спектральний розклад
Ті значення $\lambda$ для яких оператор $\hat{A} - \lambda \,\hat{I}$ має неперевний зворотній називаеються __регулярними__. Доповнення до множини регулярних значень називається __спектром__ оператора. Тобто спектр - це множина значень $\lambda$ для яих не існує ограниченого оператора зворотнього до $\hat{A} - \lambda\, \hat I$.
Нагадаємо що вектор $\left|\, \psi \right\rangle$ є власним вектором $\hat{A}$, і відповідає власному значенню $\lambda$, якщо
<!--  -->
\begin{equation}
\hat{A} \ket{\psi_{\lambda}} =
\lambda \, \ket{\psi_{\lambda}}\,,\quad \lambda \in
\mathbb{C}
\end{equation}
<!--  -->
Наприклад, якщо фізична величина приймає злічений ряд значень $\lambda_{n}$, то відповідний оператор для такої величини є:
<!--  -->
\begin{equation}\label{op1}
  \hat{A} = \sum\limits_{n} \lambda_n\,\hat{P}_{n}\,,
\end{equation}
<!--  -->
де $\hat{P}_{n}$ - проекційний оператор на простір станів в яких величина приймає значение $\lambda_n$.  "Фізичність" величини означає, що $\lambda_n \in \mathbb{R}$. Враховуючи властивості ортопроекторів і [їх звязок із процедурою вимірювання](https://colab.research.google.com/github/kulinskij-tech/QuantumMechanics1/blob/main/QM_py/qm1_ermitop_theor.ipynb#postpro), отримаємо:

- оператори, які зображують фізичні величини, є ермітовими
- їх власні стани утворюють повну систему векторів
- власні вектора, які відповідають різним власним значенням є взаємно-ортогональними.

Матричне представлення такого оператора у зліченному базисі $\{\left|\,\phi_{n}\,\right\rangle \}$ легко будується за допомогою діраковських позначень:<br>
<!--  -->
\begin{equation}
  \hat{A} = \sum_{k,n} A^{k}_{n}\,
  \braket{\phi_{k}}{\phi_{n}}\,, \quad
A^{k}_{n} = \av{\phi_{k}}{\hat{A}}{\phi_{n}}\,.\notag
\end{equation}
<!--  -->
Ермітово спряжений оператор визначається правилом:
<!--  -->
\begin{equation}
 \braket{\psi_1}{\hat{A}\psi_2} = \braket{\hat{A}^{\dagger}\psi_1}{\psi_2}\,.
\end{equation}
<!--  -->
Використовуючи проекційний оператор можна показати, що:
<!--  -->
\begin{equation}
\langle \hat{A} \rangle_{\psi} = {\rm Tr} \left(\hat{A}\,\hat{P}_{\psi}\right)
\end{equation}
<!--  -->
Таке представлення середніх значень фізичних величин демонструє, що можна взагалі не використовувати поняття хвильової функції чи вектора стану.

### Стуктура простору фізичних величин для Q-bit
Тут простір фізичних величин є скінченновимірним. Оскільки ф.в. зображуються ермітовими матриціями

$$\hat{A}^{\dagger} = \hat{A}$$

у випадку $\mathbb{C}^2$ ми маємо 4-вимірний простір ермітових матриць $2\times 2$, оскільки такі матриці мають вигляд:
<!--  -->
\begin{equation}
\label{aopqbit}
\hat{A} = 
\begin{pmatrix}
x_0+x_3&x_{1}-i\,x_{2}\\
x_{1}+i\,x_{2}&x_0-x_3
\end{pmatrix}\,,\,\,\,x_{i}\in \mathbb{R}
\end{equation}

Це фактично відповідає структурі [ортопроектора](https://colab.research.google.com/github/kulinskij-tech/QuantumMechanics1/blob/main/QM_py/qm1_qbitspace.ipynb#ortho). Оператори з нульовим слідом, наприклад мають досить просту структуру:
<!--  -->
$$
\hat{A}_{0} = \mathbf{a}\cdot\boldsymbol{\sigma}\,,\,\,\,\mathbf{a}\in \mathbb{R}^3
$$
<!--  -->
тобто асоційовані із звичайними викторами 3-вимірного дійсного простора.

__Вправа:__ Доведіть, що ермітовий оператор \eqref{aopqbit} може бути представлений як лінійна комбінація двох ортопроекторів на взаємно ортогональні стани.</p> 

### Неперервний спектр операторів 
В квантовії механиці доводиться мати справу з нескінченновимірним простором $L_2$. Основна різниця такого випадку у порівнянні з  скінечннвимірним полягає у тому що у скінечннвимірному нормованому лінійному просторі всі лінійні оператори (функціонали) є неперервними. У нескінченновимірному просторі це вже не так:  існують лінійні функції на векторах, яки не є неперервними (!). Такі лінійні оператори (функціонали) можуть бути визначені не на всьому лінійному просторі, а лише на його підмножині. Наприклад, оператор диференціювання корректно визначений на множині диференційованих функцій, яке значно меньше за простір $L_2$. Ця обставина призводить до низки математичних важкощів при означенні операторів, які відвідають певним фізичним величинам. При цьому необхідно стежити за множиною означення операторних виразів, для того, що би всі дії мали сенс (у фізичних застосуваннях природньо вимагати, щоб область означення оператора була густою підмножиною у $L_2$).
Всі точки спектра\, $\lambda$\, оператора можуьб відповідати наступним ситуаціям:

- Оператор $\hat A-\lambda\hat I$ незворотній, тобто ядро ${\rm Ker}(\hat A-\lambda\hat I)$\, нетривіально (дискретний спектр)
- множина значень оператора $\hat A-\lambda\hat I$\, не співпадає з всім простором.

Останнє може означати, що:

- Область значень оператора\, $\hat A-\lambda\hat I$\, густа в\, $L_{2}$, але зворотній оператор необмежений (неперервний спектр)
- Область значень оператора\, $\hat A-\lambda\hat I$\, не густа в $L_{2}$ (остатковий спектр)

Для самосопряжених і унітарних операторів остатковий спектр пустий і ми його розглядати не будемо. 

### Стани неперевного спектра оператора фізичної величини

Станам, які відповідають значеню $\lambda$ з неперервного спектра можна надати <a id='continuumstates'>фізично змістовну інтерпретацію</a>. Оскільки оператор 
$$
(\hat A-\lambda\hat I)^{-1}
$$ 
необмежений, то це означає, що :
$$\forall\, \epsilon\,>\,0\,,\, \exists \, \psi\in L_{2}:
\vert\vert\,(\hat A-\lambda\hat I)\psi\vert\vert\le\epsilon
$$
Неформально це означає, що з довільною точністю можна знайти "майже власний" вектор оператора. Тобто дисперсія розподілу значень із середнім значенням\, $\lambda$\, може бути зроблена як завгодно малою. Ситуацію з неперервним спектром демонструє ермітовий оператор координати $\hat x$. Власний вектор, якщо би такий існував, задовольняв би рівнянню:

$$
\hat{x}\,\psi(x)=x_0 \,\psi(x)\,,
$$

яке має тільки тривіальний розв'язок\, $\psi(x)=0$. Оператор $(\hat{x}-x_0 \hat{I})^{-1}$\, не обмежений, оскільки не визначений на функціях які не прямують (досить швидко) до $0$ в точці $x_0$. В той же час зрозуміло, що до власних значень оператора координати природньо відносити всі дійсні числа - координати точок, де можна зареєструвати появу системи. Тому необхідно узагальнити поняття власного вектора для значень, які відповідють неперервному спектру. У випадку неперервної множини значень фізичної величини слід говорити скоріше про ймовірність спостереження у інтервалі $[x_0,x_0+\Delta x]$ в даному стані (порівняйте зі  статистичною інтерпретацією хвильової функції у $x$-представленні) ніж про ймовірність того, що вона приймає значення $x_0$. Нехай\, $\psi(x,x_0)$\, є узагальнений власний вектор, який відповідає значенню\, $x_0$\, неперервного спектра. Введемо величину:

$$
  \Delta_{x_0}\psi(x)=\int
  \limits_{x_0}^{x_0+\Delta x}\,\psi(x,x')
  dx'
$$

-власний диференціал. Не важко довести, що саме власні диференціали (які є елементами $L_2$, і в силу принципа суперпозиції, представляють стани, в яких вимірювальна величина приймає значення з певного проміжка), можуть бути нормировані:

$$
  \braket{\Delta_{\lambda_{1}}\psi}{\Delta_{\lambda_{2}}\psi}=
  \left\{\begin{array}{ll}
  0 & \lambda_{1}\ne\lambda_{2}\,;\,
  (\lambda_{1},\Delta \lambda_{1})\cap
  (\lambda_{2},\Delta \lambda_{2})=\oslash\\
  \Delta \lambda+o(\Delta \lambda) & \lambda_{1}=\lambda_{2}\,;\,
  (\lambda_{1},\Delta \lambda_{1})\cap
  (\lambda_{2},\Delta \lambda_{2})=(\lambda,\Delta \lambda)
\end{array}\right.
$$
ябо якщо ввести $\delta$-функціонал Дірака:
$$
\braket{\phi_{\lambda_{1}}}{\Delta_{\lambda_{2}}\psi}= \int
dx\int\limits_{\lambda_{2}-\Delta\lambda/2}^{\lambda_{2}+\Delta\lambda/2}
\phi^{*}(x,\lambda_{1})\,\psi(x,\lambda)=
\label{c0} \int\limits_{\lambda_{2}}^{\lambda_{2}+\Delta\lambda}
\delta(\lambda_{1}-\lambda)d\lambda=\chi_{I}(\lambda_{1})
$$
де $\chi$ - характеристична функція інтервала. Безпосередньо з означення витікає, що для оператора координати
$$
\psi(x,x_{0})=\delta(x-x_{0})\,.
$$
Таким чином, при наявності неперевного спектра довільний стан (пакет) представляється у вигляді:
$$
\label{cont}
\psi(x)=\sum_{n}\braket{\phi_{n}}{\psi}\,\phi_{n}(x)+\int\limits_{cont}d\lambda\,
c(\lambda)\,\phi(x,\lambda)\\ c(\lambda)=\int dx\,
\phi^{*}(x,\lambda)\,\psi(x)\nonumber
$$
З математичної точки зору узагальнені вектори є (неперервними) лінійними функціоналами (узагальненими функціями) на області визначення оператора $D(\hat{A})$, яка є всюду густою у $L_{2}$. В діраковських позначеннях їм відповідають бра-вектори - вектори узагальнених станів:
$$
  \langle\,\lambda\,|:\, \forall\, |\,
  \psi \rangle\,\in\, D(\hat A):\,\,
  \langle\,\lambda\,|(|\, \psi\rangle)\,=\,
  \langle\,\lambda\,|\, \psi\rangle
$$
Використовуючи \eqref{x} можна вважати, що значення неперервної хвильової функції $\psi(x)$ у точці є координатой вектора стан в базисі власних узагальнених станів оператора координати:
$$
  \psi(\mathbf{r})\, = \,\langle\,\mathbf{r}\,|\, \psi\,\rangle
$$
(порівняйте з означенням координати вектора: $v^{i} = (\mathbf{e}^{i},\mathbf{v})$). Аналогічно:
$$
  \phi(\mathbf{p})\, =\, \langle\,\mathbf{p}\,|\, \psi\,\rangle
$$
і
$$
\langle \mathbf{r}\,|\,\mathbf{p} \rangle \, =\,
  \frac{1}{(2\,\pi\,\hbar)^{3/2}}
  \exp\left(i\frac{\mathbf{p}\cdot\mathbf{r}}{\hbar}\right)\,.
$$
узагальнені стани можно також уявляти як границі послідовностей векторів з $D(\hat{A})$ у сенсі т.з. слабкої збіжності:

\begin{equation}
\label{w} f_{n}\stackrel{{\rm weak}}{\longrightarrow}f
\leftrightarrow \forall \, \psi \in D(\hat{A}):\,\,
\langle f_{n}\vert\psi \rangle \rightarrow \langle f \vert \psi \rangle
\end{equation}

У такому ж слабкому сенсі слід розуміти і означення узагальнених власних станів з непреревного спектру. Дійсно, необмеженість оператора $\hat A-\lambda\hat I$ означає, що:

\begin{equation}
\label{bnd0} \forall \, c >0, \exists \psi\in
L_{2}:\vert\vert \,\psi\,\vert\vert=1,\,\,\, c
< \vert\vert\left(\hat{A}-\lambda\hat I\right)^{-1}
\psi\vert\vert<\frac{1}{\vert\vert\hat{A}-\lambda\hat
I\vert\vert}
\end{equation}

Іншими словами:

\begin{equation}\label{bnd1} 
\forall\, \epsilon>0, \exists\, \psi\in
L_{2}:\vert\vert\,\psi\,\vert\vert=1,\quad \vert\vert\left(\hat{A}-\lambda\hat
I\right)\psi\,\vert\vert<\epsilon
\end{equation}

Це, в свою чергу означає, що:

\begin{equation}
\label{bnd2} \forall\, \epsilon>0, \forall\, \phi \in L_{2} \,
\exists\, \psi\in L_{2}:\vert\vert\,\psi\,\vert\vert=1,\quad \left\vert\left(
\left(\hat{A}-\lambda\hat
I\right)\psi,\phi\right)\right\vert\,<\,\epsilon
\end{equation}

Тобто стани неперервного спектра дозволяють побудувати "майже" (в сенсі \eqref{bnd1}) власні вектори які вже належать $L_{2}$. В якості приклада розглянемо послідовність:

$$
\label{chi} n\,\chi_{\left[0,\frac{1}{n}\right]}\in L_{2}
$$

Вона не є фундаментальной і тому не збігається по нормі в $L_{2}$. Однак, у сенсі слабкої збіжності вона збігається до $\delta$-функціоналу Дірака. Таким чином, узагальнені власні вектори слід розуміти як стани, які можуть бути скіль завгодно точно апроксимовані (в слабому сенсі) послідовностями звичайних станів з $L_2$. За допомогою $\delta$-функціонала формули спрощуються і набирають одноманітний вигляд для дискретного і неперевного спектрів. Так, співвідношення ортогональності і нормування для узагальнених власних функцій має вигляд:

\begin{equation}
\label{nc1} \braket{\psi(x,\lambda_{2})}{\psi(x,\lambda_{1})}=
\delta(\lambda_{2}-\lambda_{1})
\end{equation}

Формально можно ввести аналог матричного представлення оператора, оскільки $\delta$-функціонал грає роль одиничної матриці ($\delta$-символа Кронекера) у випадку базиса узагальнених власних векторів. А саме оператору може бути співставлено "інтегральне ядро":

\begin{equation}
\label{krn} (\hat{A}\psi)(x)\rightarrow \int\limits_{V}
A(x,x')\,\psi(x')\,dx'
\end{equation}

Так для операторів координати та імпульса у базисі узагальнених власних векторів оператора координати:

\begin{equation}
\label{k2} (\hat x)_{x,x'}=x'\,\delta(x'-x)\,,\,\,\,\,\,\qquad (\hat
p_{x})_{x,x'}=i\,\hbar\, \frac{\partial}{\partial x'}\,\delta(x'-x)
\end{equation}

Тепер неважко узагальнити цей результат на випадок довільної величини, з неперервним спектром. <br>

### Умова сумісності фізичних величин
Як правило, для опису станів системи одієї физической величини недостатньо. Величина:

\begin{equation}\label{comm0}
[\hat A\,,\hat B]=\hat A\,\hat B-\hat B\,\hat A
\end{equation}

є лінійним оператором і називається комутатором операторів
$A$ и $B$. Два оператора $\hat A$ та $\hat B$ які представляють фізичні величини називаются комутуючими якщо:

\begin{equation}\label{comm}
  [\hat A\,,\hat B]=0\,.
\end{equation}

Очевидно, що якщо (самоспряжені) оператори (фізичні величини) мають спільні власні стани, тобто можуть буть одночасно виміряні, то вони комутують. Для операторів $\hat A$ и $\hat B$ фізичних величин і стана $ \left|\,\psi\,\right\rangle $ виконується співвідношення
невизначенності:
<!--  -->
\begin{gather}
\langle\,\Delta\hat{A}^2\rangle_{\psi} \langle\,\Delta
\hat{B}^2\rangle_{\psi}\ge \frac{1}{4} \vert \langle\,i\,[\hat A\,,\hat
B]\,\rangle_{\psi}\vert^2\,,\label{uncert}\\ \Delta\hat{A} = \hat{A} -
\left\langle\,\hat{A} \,\right\rangle_{\psi}\,.\notag
\end{gather}
<!--  -->
яке витікає з очевидної нерівності
<!--  -->
$$0\le \left\vert\left\vert\left(\hat{A}-\langle \hat{A}\rangle_{\psi}+i\,\lambda\,\left(\hat{B}-\langle \hat{B}\rangle_{\psi}\right)\right)\psi\right\vert\right\vert^2$$
Коммутатор ермітових операторів є антиермітовим оператором:
$$
[\hat{A}\,,\hat{B}]^\dagger=-[\hat{A}\,,\hat{B}]\,.
$$
В загальному випадку
$$[A,B]^{\dagger} = (A\,B-B\,A)^{\dagger} = (AB)^{\dagger}-(BA)^{\dagger} = B^{\dagger}\,A^{\dagger} - A^{\dagger}\,B^{\dagger}  = [B^{\dagger},A^{\dagger}]$$

По суті, \eqref{uncert} це операторна форма співвідношення невизначенностей. У випадку операторів координати та імпульсу:

\begin{equation}\label{xp}
\langle\,\Delta\hat x^2\rangle \langle\,\Delta\hat p_{x}^2\rangle
\ge \frac{1}{4} \vert\langle
i\,[\hat{x},\hat{p}_{x}]\,\rangle\vert^2=\frac{\hbar^2}{4}\,.
\end{equation}

\begin{equation}\label{heiz}
  [\hat x_{i},\hat p_{j}] = i\,\hbar\,\delta_{ij}\\
  [n_{i}\hat{x}_{i},m_{j}\hat p_{j}] = n_{i}m_{j}\,[\hat{x}_{i},\hat p_{j}]=
   i\,\hbar\,n_{j}m_{j}\,\delta_{ij} = i\,\hbar\, \mathbf{n}\cdot\mathbf{m}
\end{equation}
