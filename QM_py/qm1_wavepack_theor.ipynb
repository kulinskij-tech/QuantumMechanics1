# Квантова природа світла та розповсюдження частинок.

### Основні експериментальні факти корпускулярно-хвильового дуалізму.

Як відомо, історично квантова теорія виникла при розгляді фундаментальної проблеми встановлення теплової ріноваги у системі ЕМП+речовина. Гіпотеза про дискретний характер обміну енергією між матерією та рівноважним випромінюванням порціями чи квантами величини $\varepsilon =\hbar\, \omega$ спочатку розглядалась як суто математичний трюк (очікувалось, що при переході $\varepsilon = \hbar\, \omega \to 0$ розбіжність зникне), який дозволяв би коректно розглянути проблему розбіжності повної енергії рівноважного випромінювання. Ця розбіжність була притаманна класичній формулі Релєя-Джинса для енергетичної густини на одиничний інтервал частот:
<!--  -->
\begin{equation}\label{rj} 
\rho(\omega)=\frac{1}{V}\frac{d\,E}{d\,\omega}= U_{\omega}\, \frac{d\,N_{\omega}}{d\,\omega} = 
4\frac{k_\,T}{2}\frac{4\,\pi\,k^2 \,d\,k}{(2\,\pi)^3\,d\,\omega}=
\frac{\omega^{\,2}}{\pi^2\,c^{3}}\,T\,.
\end{equation}
<!--  -->
Суттєво, що вивід цього виразу заснований на припущенні про як завгодно малий обмін енергією між середовищем та випромінюванням, а також рівного розподілу енергії за степенями свободи (на кожну степінь свободи у стані теплової рівноваги, як відомо з класичної термодинаміки, приходиться енергія $T/2$).
Закон розподілу густини енергії за частотами, отриманий М. Планком, приводив до певної залежності енергетичної густини випромінювання даної частоти
<!--  -->
<img src="./figs/fig_plank1.png" alt="Image" width="600">
<!--  -->
\begin{equation}\label{eq_plank} 
\rho (\omega ,T) = \frac{\omega^{2}}{\pi^{\,2}\,c^{3}}
\frac{\hbar\,\omega}{\exp \left(\frac{\hbar\,\omega}{T} \right) - 1}\,.
\end{equation}
<!--  -->
Цей вираз досить точно описував експериментальні дані, і при $\hbar~\to~0$ давав класичний закон \eqref{rj}. <br>

Ідея дискретного обміну енергією в індивідуальних (а не тільки статистичних) явищах мікросвіту повною мірою "запрацювала" у роботах А. Айнштайна з
фотоефекту в яких запропоновано формулу закон збереження енергії у вигляді:

$$\hbar\,\omega = W_0 + \frac{m\,v^2}{2}$$

За її допомогою факт залежності енергії електронів тільки від частоти і її незалежності від інтенсивності падаючого світла набував логічного з'ясування. Подальший розвиток ідеї квантування руху мікроскопічних систем дозволив Бору сформулювати
"компромісні" принципи, прийняття яких забезпечує стабільність атомних систем і приводить до емпіричного комбінаційного принципу Рідберга-Ритца. 
Пряме експериментальне підтвердження дискретного обміну енергиєю між електронами і атомами було продемонстровано у дослідах Франка і Герца по вимірюванню залежності струма електронів у парах ртуті.

Розвиток цих ідей для взаємодії атомів і випромінювання привело В. Хайзенберга до побудови матричної механіки - представленню квантової механіки яке засноване на рівняннях руху для спостережувальних величин, які пов'язані із частотами випромінювання.

Інший підхід до побудови квантової механіки було розвинуто де Бройлем та Шредінгером, але ж і він суттєво засновується на узагальненні співвідношення між енергією та частотою $E = \hbar\,\omega$, імпульсом і хвильовим вектором $\mathbf{p} = \hbar\,\mathbf{k}$, відомими з квантової теорії випромінювання, на рух інших мікрочастинок. Факт спостереження явища дифракції електронів на щілинах дозволяє, за відповідною статистичною інтерпретацією про локалізацію електрона в даній області фотопластинки, дозволило висунути припущення про співвідношення між квантовим станом та деяким уявним хвильовим процесом, який описується комплексною функцією -- хвильової функцією протилежно стану класичної системи, яке зображується точкою фазового простору. Власно кажучи, функцію також можна розглядати як точку деякого, взагалі, нескінченновимірного функціонального простору. У подальшому ми з'ясуємо структуру множини станів
квантової системи.

### Хвильова функція як математичне формулювання корпускулярно-хвильового дуалізму
Врахування квантової природи світла як потоку фотонів дозволяє надати статистичну інтерпретацію інтенсивності ЕМП $\vert \mathbf{E}_{\omega}\vert^2$ і повязати її з кількістю фотонів відповідної частоти. 
Явище дифракції електронів на щілинах і утворення дифракційній картини яка аналогічна тієї що утворюється при реєстрації фотонів (дифракція Юнга) дозволив Л. де Бройлю висунути припущення про співвідношення між квантовим станом та деяким хвильовим процесом $\psi(\mathbf{r},t)$, який визначає поле реєстрації частики. При цьому інтенсивність такого хвильового поля $|\psi(\mathbf{r},t)|^2$ має статистичну інтерпретацією яка пов'язана з ймовірністю детектування частинки (електрона та інш). Цей хвильовй процес описується комплексною функцією -- хвильової функцією протилежно стану класичної системи, яке зображується точкою фазового простору узагальненних координат та імпульсів (Гамільтон) чи швидкостей (Лагранж). Власно кажучи, функцію також можна розглядати як точку (вектор) деякого, взагалі, нескінченновимірного функціонального простору. [У подальшому ми з'ясуємо структуру множини станів квантової системи](https://colab.research.google.com/github/kulinskij-tech/QuantumMechanics1/blob/main/QM_py/qm1_spaceortheor.ipynb).
Це призводить до неможливості приписати певну координату та траєкторію у класичному сенсі елементарним частинкам типу фотона, електрона та інш.

Вирішальна роль квантової природи світла яке є носієм інформації про вимірювання фізичних характеристик мікрочастинок призводить до необхідності враховувати вплив процедури вимірювання на стан системи, тому що намагання точного вимірення координати призводить до необхідності зменшення довжини світла, що в свою чергу означає збільшення частоти, тобто енергії та імпульсу. Це вочевидь порушує суто класичне припущення про нехтування впливу процедури вимірювання на такі фізичні характеристики як координата та імпульс. Нагадаємо, що аналогічним чином закони росповсюдження світла, які були встановлені класичною електродинамікою Максвела, призвели також до необхідності перебудови класичної механіки і заміни принципа відносності Ньютона-Галілея на принцип відносності Айнштайна. Саме обмеженність швидкості світла призвела, як відомо, до обмеження концепції абсолютного часу нерелятивистськими швидкостями і необхідності визнання відносності поняття одночасності.
