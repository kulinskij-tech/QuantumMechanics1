Задача про спектр енергії зарядженої частинки у полі кулонівського потенціалу є однією з основних точно розв'язуваних задач квантової механіки (разом із задачею про гармонічний осцилятор). Її можна вважати "нульовим" наближенням для більш складних атомних систем. Гамільтоніан моделі має вигляд:
<!---->
\begin{equation}\label{ham_coulomb}
\hat{H} = -\frac{\hbar^{\,2}}{2m}\,\Delta - \frac{\gamma}{r}
\end{equation}
<!---->
де $m = \frac{m_e M}{m_e+M}$ -- приведена маса системи ядро + електрон,
$\gamma = Z e^{\,2}$. 

Енергію основного стану, як відомо, можна отримати із квазикласинчих міркувань квантування орбіт (більш детально дивись [Квазикласичне наближення](./qm2_quasiclass.ipynb)). Дійсно, якщо вважати основний стан відповідаючий руху по колу із мнімальним значенням моменту імпульса, то:
<!--  -->
$$
m\,v^2/r_0 = \frac{Z\,e^2}{r^2_0}\,,\,\,\,\, m\,v\,r_0 = L_0 = \hbar\\
r_0 = \frac{\hbar^2}{Z\,m\,e^2}
$$
<!--  -->
звідки отримаємо "радіус орбіти" Бора та енергію (згідно з теоремою віріала)

```python
from scipy import constants
# 
ab=constants.hbar**2*(4*np.pi*constants.epsilon_0)/(constants.m_e*constants.e**2)
print('Bohr radius = ',ab, 'm')
# 
constants.physical_constants['Bohr radius']
```

З теореми віріала отримуємо повну енергію:

$$
E_0 = -\frac{Z\,e^2}{2\,r_0} = -Z^2\,\frac{m\,e^4}{2\,\hbar^2}
$$
В перерахунку на електрон-вольти це складає ($Z=1$)

Оскільки кулонівський потенціал є
сферично симетричним, то інтегралами руху (крім гамільтоніана) є
компоненти оператора момента імпульса $L_{i}$. Таким чином стани можна характеризувати трьома квантовими числами, які відповідають трьом комутуючими інтегралам руху:
<!---->
$$
\hat{H}, \,\,\hat{\mathbf{L}}^2\,,\,\,\,\hat{L}_z
$$
<!---->
Як і у класичному випадку можна скористатись сферичною симетрією для розділення змінних:
<!---->
\begin{equation}\label{ham_sph}
\psi_{n,l,m} (r,\theta,\varphi) = R_{n,l}(r)\,Y_{l,m}(\theta,\varphi)\\
\hat{H} \psi_{n,l,m} = E\, \psi_{n,l,m} \Rightarrow \left( -\frac{\hbar^{\,2}}{2m}\,\frac{1}{r^2}\frac{\partial}{\partial\,r}
\left(r^2\,\frac{\partial}{\partial\,r}\right)+ \underbrace{\frac{\hbar^2\,l(l+1)}{2\,m\,r^2} - \frac{\gamma}{r}}_{U_{eff}(r)}\right)\,R_{n,l} = E\,R_{n,l}
\end{equation}
<!--  -->
Після введення безрозмірної змінної $\tilde{r}\to  r/r_0$, де

$$r_0 = \frac{\hbar^2}{Z\,m\,e^2}$$

радіальне рівняння має вигляд:
<!--  -->
$$
 -\frac{1}{\tilde{r}}
\frac{d^2}{d\tilde{r}^2}\left(\tilde{r}\,R \right) + \left(\frac{\,l(l+1)}{\tilde{r}^2} - \frac{2}{\tilde{r}}\right)\,R_{n,l} = k^2\,R_{n,l}\\
k^2 = r^2_0\,\frac{2\,m\,E}{\hbar^2}
$$
<!--  -->
Враховуючи асимптотику при $r\to \infty$ можна спростити це радіальне рівняння, проводячи заміну: 
<!--  -->
$$
R = \tilde{r}^l\,e^{-k\,\tilde{r}}\,L(\tilde{r})
$$
<!---->
Незважаючи на явну залежність ефективного потенціалу від $l$, наявність додаткової симетрії негеометричного походження (тобто яка не повязана із інваріантністю відносно перетворень координат) призводить до незалежності власних значень енергії гамільтоніана \eqref{ham_sph} від квантового числа $l$. З аналогічною обставиною ми стикались для ізотропного 3-вимірного осцилятора.
<!--  -->

<!--  -->

Для додатних енергій $E>0$ параметр $\sqrt{- \mu / 2E}$ стає уявним. Замість цього ми масштабуємо вектор, використовуючи додатне значення під коренем:$$\hat{\mathbf{T}} = \sqrt{\frac{\mu}{2E}} \hat{\mathbf{R}}$$Це змінює знак останнього комутатора: $[\hat{\mathbf{T}}_i, \hat{\mathbf{T}}_j] = -i\hbar \epsilon_{ijk} \hat{\mathbf{L}}_k$. Це алгебра Лі групи $SO(3,1)$, яка ізоморфна обмеженій групі Лоренца. Таким чином, стани розсіяння мають внутрішню структуру симетрії релятивістського простору-часу.Радіальні хвильові функціїРадіальне рівняння для $E > 0$ дає розв'язки, які не затухають експоненціально на нескінченності. Визначаючи хвильове число $k = \sqrt{2\mu E}/\hbar$, хвильові функції нормуються на дельта-функцію ($\delta(k - k')$) і набувають вигляду:$$R_{k,l}(r) = C_{k,l} \, (2kr)^l \, e^{-ikr} \, _1F_1\left(l + 1 + \frac{i}{\nu}; 2l + 2; 2ikr\right)$$де $_1F_1$ — вироджена гіпергеометрична функція, а $\nu = \frac{\hbar k}{\mu \gamma} = k a_0$ — безрозмірний параметр імпульсу. Оскільки кулонівський потенціал спадає повільно ($\sim 1/r$), радіус його дії є нескінченним. Як наслідок, асимптотична поведінка хвильових функцій при $r \to \infty$ містить логарифмічний внесок: $$R_{k,l}(r) \xrightarrow{r \to \infty} \frac{2}{r} \sin\left(kr - \frac{l\pi}{2} + \frac{1}{\nu}\ln(2kr) + \sigma_l\right)$$де $\sigma_l = \arg \Gamma\left(l + 1 - \frac{i}{\nu}\right)$ — кулонівський зсув фази. Наявність члена $\ln(2kr)$ означає, що стани розсіяння ніколи не поводяться як повністю вільні частинки, навіть на нескінченній відстані.

## Класична задача (коротка довідка)

З класичної механіки відомо, що при русі у кулонівському(гравітаційному)
полі існує одатковий інтеграл руху -- вектор Лапласа-- Рунге -- Ленца:
<!---->
$$\label{laprunglenz}
\mathbf{R} = \mathbf{p}\times \mathbf{L} - \gamma \,\mu\,\frac{\mathbf{r}}{r}
$$
<!---->
Дійсно:
<!---->
$$\label{laprunglenz_vyvod}
\frac{d}{d\,t}\left({\mathbf{p}}\times {\mathbf{L}}\right) = \dot{\mathbf{p}}\times \mathbf{L} = 
-\frac{\gamma}{r^3}\,\mathbf{r}\times (\mathbf{r}\times \mathbf{p}) =
-\frac{\gamma}{r^3}\,\left(\mathbf{r} (\mathbf{r}\cdot \mathbf{p}) - \mathbf{p}\,r^2\right) =\\
\gamma\,\mu\,\left(\frac{\dot{\mathbf{r}}}{r} - \frac{\mathbf{r} (\mathbf{r}\cdot \dot{\mathbf{r}})}{r^3}\right) = \gamma\,\mu\,\frac{d}{d\,t}\left(\frac{\mathbf{r}}{r}\right)\,.
$$
<!---->
Знайдемо модуль цього вектора:
<!---->
\begin{multline}\label{r2}
  \mathbf{R}^2= p^2\,L^2-2\,\mu\,\frac{\gamma}{r}\,\mathbf{r}\cdot (\mathbf{p}\times \mathbf{L})+m^2\,\gamma^2 = 
  2\,\mu\,E\,L^2 + m^2\,\gamma^2 =\\ m^2\,\gamma^2\left(1+\frac{2\,E\,L^2}{\mu\,\gamma^2}\right)=
  m^2\,\gamma^2\,e^2\,.
\end{multline}
<!---->
\[R = \gamma\,\mu\,e\]
<!---->
де $e$ - ексцентриситет:
$$
e = \sqrt{1+\frac{2\,E\,L^2}{\mu\,\gamma^2}}
$$
Рівняння траекторії безпосередньо витікає з факту інваріантності вектора $\mathbf{R}$:
<!---->
$$\label{R_traj}
  \mathbf{R}\cdot \mathbf{r} = R\,r\,\cos\varphi = L^2 - \gamma\,\mu\,r
$$
<!---->
Звідки:
<!---->
$$\label{eq_classtraj}
r = \frac{r_0}{1+e\,\cos{\varphi}}\,,\,\,\,\,\, r_0 = \frac{L^2}{\mu\,\gamma}\,.
$$

```python
x = np.linspace(0.1,10,200)
# 
def ueff(x): 
    return 4*x**2 - 10*x
# 
x1=2
y1=-2
coeff=[-y1,-10,4]
roots=np.roots(coeff)
plt.plot(x,ueff(1/x))
plt.plot(x,y1*x/x,'r',dashes=[2,3])
plt.plot(roots[0],y1,'ro',ms=12)
plt.plot(roots[1],y1,'ro',ms=12)
plt.axhline(y=0,xmin=-0.1,xmax=10,color='black',dashes=[3,3])
plt.text(8,2.5,'E>0',fontsize=16)
plt.text(8,-3.5,'E<0',fontsize=16)
plt.title('Effective potential energy $U_{eff}(r)$\n',fontsize=14)
plt.ylim(-8, 4)
plt.show
```

\begin{align}\label{eq_poissonbrackets_class}
\{L_{i},L_{j}\} =&\,\hphantom{2\,\mu}\varepsilon_{ijk}\,L_{k}\notag\\
\{R_{i},R_{j}\} =&\, -2\,\mu\,E\,\varepsilon_{i\,j\,k}\,L_{k} \\
\{L_{i},R_{j}\} =&\,\hphantom{2\,\mu} \varepsilon_{ijk}\,R_{k} \notag
\end{align}

Це дійсно збігається із квантовомеханічним розражунком, але слід пам'ятати що ніякої "кругової орбіти" у основному стані немає, оскільки в цьому стані взагалі момент імпульсу відсутній $l=0$ тобто розподіл координати електрона подібний до стоячої сферичносиметричної хвилі. Стаціонарне рівняння Шредінгера для цього стану має вигляд:

$$\frac{1}{r}\,(r\,\psi_0)'' - \frac{2\,m\,Z\,e^2}{\hbar^2}\,\frac{\psi_0}{r} = \varkappa^2\,\psi_0$$

Підставляючи $\psi_0 = A\,e^{-\varkappa\,r}$ знаходимо:

$$ \varkappa =  Z\,\frac{m\,e^2}{\hbar^2}\,\, \Rightarrow \,\, E_0 = -Z^2\,\frac{m\,e^4}{2\,\hbar^2}$$

## Квантовий випадок. Алгебра інтегралів руху

В квантовій механиці величині \eqref{laprunglenz} може бути
співставлений векторний оператор з компонентами:
<!---->
\begin{equation}\label{laprunglenz_qm} 
\hat {R}\,_{i} \, = \frac{1}{2}
\varepsilon_{i\,j\,k} \{\hat {p}_{j} ,\,\hat {L}_{\,k} \} - 
\gamma\,m\,\frac{\hat {x}_{i}}{r}\,
\end{equation}
<!---->
Нагадаємо, що $\{A,B\}=AB+BA$. 

Безпосереднім обчисленям комутатора можна пересвідчитись, що
у випадку коли хвильові функції стаціонарних станів обмежені у точці $r=0$:
<!---->
\begin{equation}\label{hlrl_commut} 
[H,\hat {R}\,_{i} ]\, = 0\,.
\end{equation}
<!---->
З означення витікає:
<!---->
\begin{equation}
\label{la_commut} {\mathbf{\hat {R}}}\, \cdot {\mathbf{\hat {L}}} = {\mathbf{\hat {L}}} \cdot {\mathbf{\hat {R}}}\,\, = 0\,.
\end{equation}
<!---->
Крім того, завдяки тому що \eqref{laprunglenz_qm} є вектором:
<!---->
\begin{equation}
\label{LR_commut}
[L_{\,i} ,R_{j} ] = i\,\hbar\,\varepsilon
_{i\,j\,k} \,R_{k}\,.
\end{equation}
<!---->
Для полегшення аналітичних (досить простих, по суті, алгебраічних) розрахунків скористаємося пакетом SymPy.

```python
psi = symbols('psi',cls=Function)
x,y,z,gamma = symbols('x y z gamma',real=True) 
pos = [x,y,z] # coordinates
#
def coord(f): # coordinate operator
    p=[]
    for x in pos:
        p.append(x*f)
    return p
# 
def imp(f):  # momentum operator
    p=[]
    for x in pos:
        p.append(-I*diff(f,x))
    return p
# 
def mom(f):  # angular momentum operator
    p=[]
    for i in range(3):
        p0=0
        for j in range(3):
            for k in range(3):
                p0=p0+LeviCivita(i,j,k)*pos[j]*imp(f)[k]
        p.append(p0)
    return p
#
def ham(f):  # hamiltonian operator
    s0=0
    for x1 in pos:
        s0=s0-diff(f,x1,2)/2
    h=s0 - gamma/sqrt(x**2+y**2+z**2)*f
    return h
ham(psi(x,y,z))
```

```python
# [p_y,H]
i=1 #y-component
simplify(imp(ham(psi(x,y,z)))[i]-ham(imp(psi(x,y,z))[i]))/I
```

```python
# [x,H]
i=0 #x-component
simplify(coord(ham(psi(x,y,z)))[i]-ham(coord(psi(x,y,z))[i]))/I
```

Розглянемо комутаційні співвідношення для компонент
оператора Лапласа -- Рунге -- Ленца. Зауважимо, що завдяки \eqref{LR_commut} вираз для \eqref{laprunglenz_qm} може бути записаний (самостійно) як:
<!---->
\begin{equation}
\label{R_alt} R\,_{i} \, = \varepsilon _{i\,m\,n} p_{m}\,L_{n} - i\,\hbar \,p_{i} - m\,\gamma\,\frac{x_{i}}{r}\,\,.
\end{equation}
<!---->
Для подальших перетвореннь скористаємось комутаторами:
<!--  -->
$$
[L_{\,i} ,p_{j} ] = i\,\hbar\,\varepsilon _{i\,j\,k} \,p_{k}
\,\,\,,\,\,[L_{\,i} ,x_{j} ] = i\,\hbar\,\varepsilon _{i\,j\,k}
\,x_{k} \,\,,\,\,\,[L_{\,i} ,\,r] = 0\,,
$$
<!--  -->
і тотожністю $[A,B\,C] = [A,B]\,C + B\,[A,C]$:
<!---->
\begin{equation}
\label{RR_commut} [R_{\,i} ,R_{\,j} ] = \, - 2i\,m\,\hbar\,\varepsilon _{i\,j\,k} \,L_{k} \,H
\end{equation}
<!---->
Крім того, використовуючи \eqref{R_alt} прямим обчисленням можна
показати, що:
<!---->
\begin{equation}
\label{R2} \mathbf{R}^{2} = 2\,m\,H({\rm L}^{2} + \hbar^{\,2})
+ \gamma^{\,2}m^{\,2}
\end{equation}
<!---->
Рзглянемо тепер певний підпростір S$_{E}$ стаціонарних
станів, які відаповідють значенню енергії $E$<0 дискретного
спектра. Оператори момента импульсу і Лапласа -- Рунге -- Ленца
комутують з гамільтоніаном. Обмежуючи дію цих операторів
на S$_{E}$ для \eqref{RR_commut} маємо:
<!---->
\begin{equation}
\label{RR_commut1} {\left. {[R_{\,i} ,R_{\,j} ]} \right|}_{S_{E}}
=i\,\hbar\, \varkappa^2 \epsilon_{i\,j\,k}\,L_k
\,\,\,,\,\,\varkappa^{\,2} = - 2\,m\,E
\end{equation}
<!--  -->
Введемо оператори:
<!---->
\begin{equation}
\label{WRL} {\rm W}^{(\pm)} = \frac{1}{2}\left( {\rm L} \pm
\frac{1}{\varkappa }{\rm R} \right)
\end{equation}
<!---->
Для цих операторів виконуються співвідношення:
<!---->
\begin{equation}\label{WW_commut}
\begin{array}{l}
 [W^{(\pm)}_{i} ,W^{(\pm)}_{j} ] = i\,\hbar \,\varepsilon _{i\,j\,k} W^{(\pm)}_{k}\,\delta_{+,-}\\
  {\rm W}^{(+)^{2}} - {\rm W}^{(-)^{2}} = 0\,,\,\,\,\,
  {\rm W}^{(+)^{2}} + {\rm W}^{(-)^{2}} = \frac{\gamma
^{\,2}m^{\,2}}{2\,\varkappa^{\,2}} - \frac{\hbar^{\,2}}{2} \\
 \end{array}
\end{equation}
<!---->
З \eqref{WW_commut} витікає, що компоненти оперторів
\eqref{WRL} задовільняють комутаційним співвідношенням для
компонент оператора кутового момента, а власні значення
квадратів цих оперторів є рівними. Поэтому позначимо:
<!---->
\begin{equation}
\label{W2} {\mathbf{W}^{(-)}}^{2} = {\mathbf{W}^{(+)}}^{2} = \hbar ^{\,2}j(j + 1)\,,
\end{equation}
<!---->
де $2j$ - ціле число. З цього співвідношення остаточно
отримаємо:
<!---->
\begin{equation}
\label{eq_enrg_coulomb} E = - \frac{\varkappa^{\,2}}{2\,m} = -
\frac{m\,\gamma^{\,2}}{2(2j + 1)^{\,2}\hbar^{\,2}} = 
-\frac{m\,Z^2e^4}{2(2j + 1)^{\,2}\hbar^{\,2}}
\end{equation}
<!---->
Вводячи позначення $n = 2j+1$, отримаємо формулу для енергії:
<!---->
\begin{equation}\label{enrg_n_hydrogen}
 E_{n} = -Z^2\frac{ {\rm Ry} }{n^2}\,, \quad {\rm Ry} = \frac{m\,e^4}{2\,\hbar^{\,2}}
\end{equation}
<!---->

$$
\bra{n, l, m}\mathbf{R}^2\ket{n, l, m}=
m^2\,\gamma^2\left(1 - \frac{(l(l+1)+1)}{n^2}\right)
$$
<!--  -->

$$\bra{1, 0, 0}\mathbf{R}^2\ket{1, 0, 0}= \bra{1, 0, 0}\mathbf{L}^2\ket{1, 0, 0}=0$$
